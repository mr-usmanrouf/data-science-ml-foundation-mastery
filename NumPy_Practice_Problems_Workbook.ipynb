## Jupyter & NumPy Pro-Tips

A few shortcuts and habits that will save you real time across these 105 problems:

| Shortcut | What it does |
|---|---|
| `Shift + Enter` | Run the current cell and move to the next one |
| `Esc`, then `A` | Insert a new cell **above** the current one |
| `Esc`, then `B` | Insert a new cell **below** the current one |
| `Esc`, then `M` | Convert the current cell to **Markdown** |
| `Esc`, then `Y` | Convert the current cell to **Code** |
| `Shift + Tab` | Show the docstring/tooltip for a function while your cursor is inside its parentheses |

One NumPy-specific habit worth building early: read shape-mismatch errors carefully. `ValueError: operands could not be broadcast together with shapes ...` is the single most common error you'll hit in this workbook, and it always tells you the exact two shapes that don't align. Print `.shape` liberally while debugging instead of guessing.
