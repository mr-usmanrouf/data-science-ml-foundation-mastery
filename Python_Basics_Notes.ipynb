## 1. What a program actually is

Before any syntax, it's worth answering a question most courses skip: when you "run" a Python file, what is actually happening?

A program is a sequence of instructions. That sounds obvious, but the important word is *sequence* — a computer doesn't understand your goal, only the individual steps you give it, one after another, in order. Python is one way of writing those steps down in a form that's readable by a person and, after translation, followable by a machine.

The text you write — the `.py` file, or the contents of a notebook cell — is called **source code**. It's just text. Nothing about it runs on its own; a `.py` file sitting on your disk is no more "running" than a letter sitting in an envelope is being read.

Something has to turn that text into action. That something is the **Python interpreter** — a program (yes, a program that runs other programs) that reads your source code and carries out the instructions in it. When people say "I ran my Python script," what they mean is: I handed my source code to the interpreter, and the interpreter executed it.

It's worth being a little more precise about "executed," because the word hides a two-step process. The interpreter first reads your source code and translates it into a lower-level set of instructions it can work with efficiently — you don't need the details of this yet, only that it happens. Then it carries out those instructions one at a time, top to bottom, following the control-flow rules of whatever language constructs you used (we'll meet several of those constructs later — conditions, loops, function calls — and each one is really just a way of telling the interpreter to deviate from strict top-to-bottom execution in a controlled way).

None of this requires you to know how CPython — the specific interpreter you're almost certainly using — is implemented internally. That's a topic for later. What matters now is the mental shape: **source code is a static description; execution is what happens when the interpreter reads and acts on that description, one instruction after another.**

### Script vs. interactive execution

There are two common ways to hand code to the interpreter.

The first is writing a `.py` file and asking the interpreter to run the whole thing from top to bottom. This is called running a **script**. You write the file, save it, and execute it; the interpreter goes through it start to finish.

The second is typing code directly into an interactive prompt — often called the **REPL**, short for Read-Eval-Print Loop. You type one line, the interpreter reads it, evaluates it, prints a result if there is one, and waits for your next line. This is what you're doing right now in a Jupyter notebook, cell by cell, except a notebook also lets you keep the results and explanations next to each other permanently, which is why it's a good format for learning.

The two modes aren't different languages — they're different ways of feeding the same interpreter. A script commits to running everything in order without pausing for you; the REPL lets you experiment one step at a time and see immediately what each step did. You'll use both. Scripts are how real programs get run; the REPL (and notebooks) are how you explore an idea before you're sure of it.

Two separate things happened in that one line, and it's worth pulling them apart.

`"Hello"` is a **string** — a piece of text data, written between quotes. On its own, `"Hello"` is a valid piece of Python: it's a value, and Python could evaluate it and hand you back that exact string. Try it in its own cell.

````python
"Hello"    # just a value, evaluated on its own
````

Notice this produces output too, but it looks slightly different from what `print("Hello")` produced — in a notebook, a bare value on the last line of a cell is shown as the cell's **result**, formatted in a particular way (with quotes, in this case, since it's showing you a string value, not printing its contents).

`print(...)` is different. It's a **function call** — you're asking Python to run a piece of pre-built behavior named `print`, and handing it `"Hello"` as the thing to act on. What `print` does is send text to the screen, without quotes, as plain output. We'll spend an entire chapter on functions later; for now, treat `print(x)` as "display `x`."

An **expression** is anything Python can evaluate down to a value. `"Hello"` is an expression. `2 + 2` is an expression. `print("Hello")` is also an expression — it evaluates to a special "nothing" value after it does its job of printing (you'll meet that value, `None`, later) — but its point isn't the value it evaluates to, it's the *action* it performs on the way there. Some code is written for what it evaluates to; some is written for what it does. `print` is written for what it does.

A comment that just restates the code adds nothing:


That comment tells you nothing you couldn't already see. A comment earns its place when it explains something the code itself can't — a reason, a warning, a piece of context. You'll see this distinction respected throughout the book: comments here explain *why*, not *what*.

## 4. Variables — storing and naming values

`print("Hello")` displayed a value and then that value was gone — nothing kept it around. Most useful programs need to hold on to values and use them again later, possibly after changing them. That's what a variable is for.

````python
age = 19
````

Nothing printed, because we didn't ask for output — we asked Python to remember something. The line `age = 19` is called an **assignment**. Reading it aloud as "age equals nineteen" is a bad habit to build, because `=` in Python does not mean mathematical equality. It means something closer to *"attach the name on the left to the value on the right."*

A useful (and honest) way to picture it:

```text
age ───────► 19
```

`age` is a name. `19` is a value — specifically, an **integer object** sitting somewhere for Python to use. The assignment doesn't put `19` "inside" a box called `age`; it makes `age` a way of referring to that `19`. This distinction feels pedantic with numbers, but it stops being pedantic very soon, so it's worth taking seriously now, while the example is simple enough to see clearly.

Once the name exists, you can use it in place of the value it refers to:

````python
age = 19
print(age)
print(age + 1)
````

`age + 1` doesn't change `age`. It looks up what `age` currently refers to, computes `19 + 1`, and produces `20` as a new value — which `print` then displays. `age` itself still refers to `19` after that line runs. If you want `age` to actually become `20`, you have to say so:

````python
age = 19
age = 20
print(age)
````

### What just happened to the old value?

When you write `age = 20`, Python doesn't edit the `19` into a `20`. It creates (or locates) a new value, `20`, and points `age` at it instead. The old `19` is simply no longer referred to by `age`. Picture it as redirecting an arrow, not editing a box:

```text
before:   age ───────► 19

after:    age ───────► 20
```

This matters because a name can be reassigned to a value of a completely different kind without complaint:

````python
thing = 19
thing = "nineteen"
print(thing)
````

`thing` first pointed at an integer, then at a string. Nothing about the name `thing` restricts what kind of value it can point to — the name is just a label you're free to move. (Whether reassigning a name to a wildly different kind of value is a *good idea* is a separate question — usually it isn't, because it makes code harder to follow — but Python itself has no objection.)

### Two names, one value

Here's the example that makes the "name → value" model earn its keep, rather than just being a nice picture.

````python
x = 10
y = x

print(x, y)
````

`y = x` does not create a copy of `x`'s value living somewhere new. It looks up whatever `x` currently refers to — `10` — and makes `y` refer to that same `10`.

```text
x ─────┐
       ├────► 10
y ─────┘
```

Both names point at the same value. Now:

````python
x = 10
y = x

x = 20

print(x)
print(y)
````

What do you think `y` prints, before you run it?

Run it. `y` is still `10`. The line `x = 20` redirected `x` to a new value; it did nothing to `y`, because `y` was never attached to `x` — it was attached to the *value* `x` happened to be pointing at, at the moment the assignment ran. Once made, that connection between `y` and `10` doesn't care what `x` does afterward.

```text
x ─────► 20            10 ◄───── y
          (x moved on)   (y still points here)
```

This will look almost too obvious right now, with plain numbers. It stops being obvious once you meet lists in a few chapters, where a very similar-looking piece of code behaves completely differently — because lists, unlike numbers, can be changed in place. Hold on to this diagram; you'll need it again.

### Reassignment and multiple assignment

You've already seen reassignment — pointing a name at a new value:

````python
score = 0
score = score + 10
print(score)
````

Read the right-hand side first: Python evaluates `score + 10` using score's *current* value, gets `10`, and only then makes `score` refer to that result. The name on the left is never consulted for its old value until the right-hand side has been fully worked out — order matters, and it always goes right-to-left in terms of "what happens first," even though you read the line left-to-right.

Python also lets you assign several names at once:

````python
a, b, c = 1, 2, 3
print(a, b, c)
````

and to swap two names' values without a temporary variable, which is a fairly elegant consequence of the same mechanism:

````python
a, b = 1, 2
a, b = b, a
print(a, b)
````

The right-hand side `b, a` is fully evaluated — producing the pair `(2, 1)` — *before* any assignment happens, so there's no risk of overwriting `a` before you've read its old value for `b`.

### Naming things

A variable name must start with a letter or underscore, and can only contain letters, digits, and underscores after that — no spaces, no punctuation. Names are case-sensitive: `Age` and `age` are different names.

```python
age = 19        # fine
_age = 19       # fine, if unusual
1age = 19       # not allowed — can't start with a digit
my age = 19     # not allowed — no spaces
```

Beyond what's *legal*, there's what's *conventional*. Python code overwhelmingly uses `snake_case` for variable names — lowercase words separated by underscores, like `total_score` or `first_name` — rather than `camelCase` or `PascalCase`. Nothing enforces this; the interpreter doesn't care. But code that follows the convention is easier for other Python programmers (including a future version of you) to read at a glance, and that's worth more than it sounds.

## 5. Numbers — integers and floats

Python has already been quietly using two different kinds of numbers in the examples so far. It's time to name them.

`19`, `0`, and `-3` are **integers** — whole numbers, with no fractional part. `19.99`, `0.5`, and `-3.0` are **floats** (short for *floating-point numbers*) — numbers with a decimal point, used for anything that isn't a whole quantity.

You can check which kind a value is with `type()`: 

````python
print(type(19))
print(type(19.99))
````

`int` and `float` are the names Python uses for these two kinds internally. The distinction isn't cosmetic — Python genuinely represents them differently in memory, because a whole number and a number with a fractional part need different storage strategies to be handled efficiently and precisely. You don't need the representation details yet (they matter more once you start asking why `0.1 + 0.2` doesn't print exactly `0.3` — a topic for later, once you have the tools to appreciate why), but it's worth knowing the two are not just "numbers formatted differently" — they're genuinely distinct kinds of value.

One consequence you can see immediately: mixing the two in an operation produces a float, even if the float involved has no fractional part.

````python
print(3 + 2)      # int + int -> int
print(3 + 2.0)    # int + float -> float
print(3.0 + 2.0)  # float + float -> float
````

Python decides "if a float is anywhere in the calculation, the result should be a float," because a float is the only type of the two that can represent every value the other one can, plus more. This is a small rule, but it explains a lot of surprising output later if you don't have it.

### Arithmetic operators

```python
+   addition
-   subtraction
*   multiplication
/   division (always produces a float — its own chapter is coming)
**  exponentiation
```

````python
print(2 + 3)
print(5 - 8)
print(4 * 2.5)
print(2 ** 10)
````

`-8` above is a **negative number**, written with a leading `-`. Be careful reading `-` in code: it means subtraction between two values (`5 - 8`) but negation in front of a single value (`-8` on its own). Python tells these apart from position, and so should you when reading code — `5 - -3` (subtract negative three) is legal and equals `8`, but it's exactly the kind of thing worth writing with a space or parentheses for clarity: `5 - (-3)`.

### Order of operations

Python evaluates expressions using the same precedence rules you already know from mathematics: exponentiation before multiplication and division, which come before addition and subtraction. Parentheses override all of it.

````python
print(2 + 3 * 4)      # multiplication first: 2 + 12
print((2 + 3) * 4)    # parentheses first: 5 * 4
````

When an expression has more than two or three operators, don't try to hold the whole precedence table in your head while reading it — add parentheses, even ones that aren't strictly required, if they make the intended grouping obvious. Code is read far more often than it's written, and a slightly redundant parenthesis costs nothing.

## 6. `/` vs `//` vs `%`

Run this before reading anything else:

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
````

Three different results from the same two numbers. This chapter is about understanding all three well enough that none of them ever surprises you again — including in the negative-number cases that trip up almost everyone the first time.

### `/` — division

`/` is ordinary division, and in Python it **always** produces a float, even when the numbers divide evenly:

````python
print(6 / 2)     # 3.0, not 3
print(7 / 2)     # 3.5
````

This is a deliberate design choice. Older versions of Python (and many other languages) made `/` behave differently depending on whether the operands were integers, which caused a specific, well-known class of bugs: `7 / 2` silently truncating to `3` when a programmer expected `3.5`. Python decided `/` should always mean "true division," full stop, and gave floor division its own separate operator instead.

### `//` — floor division

`//` divides and then rounds *down* — toward negative infinity — to the nearest whole number, returning that whole number.

````python
print(7 // 2)     # 3
print(8 // 2)     # 4
print(9 // 2)     # 4
````

For positive numbers, this matches the "how many whole times does 2 go into 7" intuition, and it's tempting to stop there. Don't — that intuition breaks the moment a negative number is involved.

### The part that surprises everyone

> What do you think `-7 // 2` evaluates to?

Most people guess `-3`, reasoning "seven into two is three-point-five, so drop the decimal." Run it.

````python
print(-7 // 2)
````

It's `-4`, not `-3`. The reason is in the name: **floor** division rounds toward negative infinity, not toward zero. `-7 / 2` is `-3.5`. The floor of `-3.5` — the largest whole number that is *not greater than* `-3.5` — is `-4`, because `-4` is further left on the number line than `-3.5`, and `-3` is to the *right* of `-3.5`, which would mean rounding up, not down.

```text
...  -5   -4   -3.5   -3   -2  ...
            ▲    ▲
          floor  actual value
```

"Truncate toward zero" and "floor toward negative infinity" agree for positive numbers and disagree for negative ones. Python's `//` is always floor, never truncation — this is consistent, not arbitrary, and once you've seen the disagreement once you won't forget which one Python does.

### `%` — the remainder (modulo)

`%` gives you what's left over after floor division has taken out as many whole groups as it can.

````python
print(7 % 2)     # 1
print(8 % 2)     # 0
print(9 % 2)     # 1
````

And the same negative-number question applies here too:

````python
print(-7 % 2)
````

It's `1`, not `-1`. This isn't a separate rule to memorize — it falls directly out of a mathematical identity that ties all three operators together, and this identity is the actual definition of what `//` and `%` mean in Python, more fundamental than any intuition about "leftover pieces":

$$
a = (a \mathbin{/\!/} b) \times b + (a \bmod b)
$$

Check it against `-7` and `2`: `(-7 // 2) * 2 + (-7 % 2)` = `(-4) * 2 + 1` = `-8 + 1` = `-7`. It holds exactly. Python guarantees this identity is true for every pair of numbers you give it (except dividing by zero, which raises an error rather than pretending to have an answer). Once `//` is defined as "floor toward negative infinity," `%` is *forced* to be whatever value makes this equation balance — there's no independent choice being made about the remainder's sign.

A consequence worth noticing directly: `%`'s result always has the same sign as the second operand (the one on the right), not the first.

````python
print(7 % -2)      # -1, matches the sign of -2
print(-7 % -2)     # -1, still matches -2's sign
````

### Why this matters in real code

These aren't abstract number tricks — they're the standard tools for a specific family of problems: turning a raw quantity into "how many whole groups, and what's left."

**Even or odd:**

````python
n = 17
print(n % 2 == 0)   # False -> n is odd
````

**Extracting digits**, right to left, from a number:

````python
n = 4207
last_digit = n % 10
rest = n // 10
print(last_digit, rest)
````

`% 10` isolates the ones place; `// 10` drops it, leaving the rest of the number to process the same way again (this is the seed of an idea called recursion, which is well beyond this chapter, but the pattern is worth recognizing early).

**Wrapping a position** — for example, cycling through a fixed number of positions and looping back to the start:

````python
position = 0
for _ in range(7):
    print(position)
    position = (position + 1) % 4   # wraps back to 0 after reaching 3
````

(We haven't covered `for` yet — don't worry about the loop syntax itself, just notice that `% 4` is what keeps `position` inside `0, 1, 2, 3` no matter how many times it increases.)

**Converting minutes into hours and minutes** — exactly the shape of the practice problem below:

````python
total_minutes = 137
hours = total_minutes // 60
remaining_minutes = total_minutes % 60
print(hours, "hours and", remaining_minutes, "minutes")
````

**Grouping items into fixed-size batches** — how many full batches of 6, and how many are left over:

````python
items = 20
batch_size = 6
full_batches = items // batch_size
leftover = items % batch_size
print(full_batches, "full batches,", leftover, "left over")
````

In every one of these, `//` answers "how many complete groups," and `%` answers "what remains once the complete groups are removed." That's the actual meaning to internalize — not "`//` rounds down" in isolation, but `//` and `%` as a matched pair that always, together, account for the whole original number.

## 7. Strings

A string is a sequence of characters, written between quotes — either kind works, and they're interchangeable:

````python
a = "hello"
b = 'hello'
print(a == b)
````

Having both is mostly a convenience: it lets you write a string that contains one kind of quote by wrapping it in the other, without extra fuss.

````python
quote = "She said 'hi' to me"
print(quote)
````

### Building strings out of other strings

`+` between strings **concatenates** them — joins them end to end. It does not add anything numerically; that would make no sense, and Python won't try to guess what you meant if you mix a string and a number this way.

````python
first = "Ali"
last = "Khan"
full = first + " " + last
print(full)
````

````python
print("ab" + "cd")   # "abcd" -- concatenation, not addition
print("3" + "4")     # "34" -- these are strings, not numbers
````

`*` between a string and an integer **repeats** the string that many times:

````python
print("ha" * 3)
print("-" * 20)
````

That second line is a common, genuinely useful pattern for drawing a quick separator line in output, and it's worth remembering rather than reconstructing from scratch each time.

### Strings don't change — they're replaced

Here's something that looks like it should work and doesn't:

````python
word = "cat"
word[0] = "b" 
````

That raises an error: `'str' object does not support item assignment`. Strings in Python are **immutable** — once created, a string's contents cannot be changed. This isn't a missing feature; it's a deliberate property, and it has real consequences later (immutable values, for instance, can be used in places mutable ones can't, which you'll run into with dictionaries).

If you want what looks like "changing" a string, what you're actually doing is building a *new* string and reassigning the name to it: 

````python
word = "cat"
word = "b" + word[1:]
print(word)
````

(That `word[1:]` is a slice — the next chapter is entirely about that syntax.) The important idea for now: `word` didn't get edited in place. A new string, `"bat"`, was created, and `word` was pointed at it — exactly the same "redirect the arrow" behavior you saw with numbers in Chapter 4, just applied to a value that happens to look like text.

### A few methods worth knowing now

Strings come with a large set of built-in methods, and dumping all of them on you here wouldn't teach you anything — you'd forget most of them before using them once. A small, genuinely common set is enough to start:

````python
name = "  Ali Khan  "
print(name.strip())        # remove leading/trailing whitespace
print(name.upper())
print(name.lower())
print(name.strip().replace("Ali", "Usman"))
````

Every one of these **returns a new string** rather than modifying `name` — a direct consequence of immutability. Check for yourself: after running the cell above, what does `name` itself still hold?

````python
print(name)   # unchanged
````

One more that comes up constantly — checking whether one string contains another, using the `in` keyword:

````python
sentence = "the quick brown fox"
print("quick" in sentence)
print("slow" in sentence)
````

## 8. String indexing and slicing

A string being a *sequence* of characters means each character has a position, and Python lets you reach in and grab one directly.

````python
s = "Python"
print(s[0])
print(s[1])
````

Indexing starts at `0`, not `1`. This trips up almost every beginner at least once, so it's worth internalizing the reason rather than just the rule: the index isn't naming "the first character, the second character..." — it's naming *how many characters to skip from the start* before reading one off. `s[0]` means "skip zero characters, then read one." `s[1]` means "skip one, then read one." Thought of that way, starting from `0` stops being arbitrary.

Here's the full map of a six-character string, forward and backward:

```text
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

Negative indexes count from the end, and `-1` is the last character — genuinely useful, since you often want "the last item" without needing to know or compute the string's length first.

````python
s = "Python"
print(s[-1])
print(s[-6])
print(s[5])    # same character as s[-1], reached from the other end
````

### Slicing — taking a range instead of one character

`s[start:stop]` gives you the characters from index `start` up to, but **not including**, index `stop`.

````python
s = "Python"
print(s[0:2])   # 'Py'
print(s[2:6])   # 'thon'
````

Why is `stop` excluded rather than included? Because it makes a very common calculation clean: the length of `s[start:stop]` is exactly `stop - start`, with no off-by-one adjustment needed. `s[0:2]` has length `2 - 0 = 2`. If `stop` were inclusive, you'd constantly be writing `stop - start + 1` instead, and that `+1` is exactly the kind of detail that produces bugs when people forget it under pressure. Python's designers chose the version that keeps the arithmetic simple, at the one-time cost of "stop feels like it should be included but isn't."

Both boundaries can be left out, and each has a specific meaning when omitted:

````python
s = "Python"
print(s[:3])    # from the start up to index 3
print(s[3:])    # from index 3 to the end
print(s[:])     # the whole string (a full copy, in effect)
````

### Adding a step

`s[start:stop:step]` adds a third piece: how many positions to move forward each time, instead of always moving one at a time.

````python
s = "Python"
print(s[0:6:2])   # every second character
print(s[::2])      # same thing, using the omitted-boundary defaults
````

A negative step reverses the direction entirely, and this is where people's mental model usually breaks, so it's worth slowing down.

````python
s = "Python"
print(s[::-1])   # the whole string, reversed
````

With a negative step, Python walks *backward*, which quietly flips what "start" and "stop" mean in practice: `start` should be to the *right* of `stop` on the string, because you're now moving right-to-left.

Here is the pair the syllabus specifically wants you to sit with, because it looks like it should behave the same way twice and doesn't:

````python
s = "Python"
print(s[-1:-5])       # ???
print(s[-1:-5:-1])    # ???
````

What do you expect for each, before running them?

Run them. `s[-1:-5]` prints an **empty string**. `s[-1:-5:-1]` prints `'noht'`.

Here's why. `s[-1:-5]` uses the *default* step, which is `1` — meaning "move forward, left to right." But `-1` (the position of `'n'`) is to the *right* of `-5` (the position of `'h'`). Moving forward from a position that's already past your stopping point can never reach it — there's nothing to collect, so you get nothing: an empty string. Python doesn't raise an error here; it just quietly produces an empty result, which is arguably more dangerous than an error, because silent wrong answers are harder to notice than loud ones.

`s[-1:-5:-1]` fixes this by setting the step to `-1`, telling Python to move *backward*. Now starting at `-1` (`'n'`) and moving backward toward, but not including, `-5` (`'h'`) makes sense as a direction, and it collects `'n'`, `'o'`, `'h'` — wait, let's actually trace it index by index instead of guessing.

### Let's trace it

`s = "Python"`, indices `0..5` forward, `-6..-1` backward. Starting index for the slice is `-1`, which is position `5`, the character `'n'`. Step is `-1`, so each step moves one position to the left (toward index `0`). Stop is `-5`, position `1`, the character `'y'` — and stop is excluded, same rule as always, just now approached from the right.

```text
index:     0    1    2    3    4    5
char:      P    y    t    h    o    n
                ▲                   ▲
             stop (excl.)        start
```

Walking backward from index `5`: collect `n` (index 5), step to `4`: collect `o`, step to `3`: collect `h`, step to `2`: collect `t`, step to `1` — that's the stop index, excluded, so we halt *before* collecting `y`.

Result: `n`, `o`, `h`, `t` → `"noht"`. That matches what ran above.

The rule underneath all of this, stated once and for all: a slice always moves in the direction its step implies, starting at `start` and continuing until it reaches `stop`, without ever including `stop` itself. If moving in that direction from `start` can never reach `stop` at all, you silently get nothing back, rather than an error. Whenever a slice looks wrong, walk through it exactly like this — index by index, direction by direction — rather than guessing.

## 11. Logical operators and truthiness

Comparisons give you one boolean at a time. `and`, `or`, and `not` combine them.

````python
age = 20
has_ticket = True

print(age >= 18 and has_ticket)
print(age >= 18 or has_ticket)
print(not has_ticket)
````

`and` is true only if both sides are true. `or` is true if at least one side is true. `not` flips a boolean. None of that is surprising, and if it were the whole chapter, it wouldn't need to be a deep dive. The part that does need care is what happens when `and` and `or` are given values that *aren't* already boolean.

### Every value has a truth value

Python doesn't require a condition to literally be `True` or `False` — it will treat almost any value as one or the other, according to a rule called **truthiness**. Most values count as true. A specific, short list of values count as false:

```python
0
0.0
""
None
[]
{}
set()
```

Zero (in any numeric form), an empty string, `None` (a special "no value" value you'll meet properly with functions), and every empty collection — these are **falsy**. Everything else — including negative numbers, non-empty strings (even `"False"` as text!), and non-empty collections — is **truthy**.

````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("False"))    # truthy! it's a non-empty string
print(bool([]))
print(bool([0]))        # truthy! a list containing something, even a falsy something
````

That `bool("False")` line is worth pausing on: the *text* `"False"` is a non-empty string, and non-empty strings are truthy regardless of what they say. Truthiness is about the value's type and emptiness, never about its meaning as text.

This is why `if some_list:` is idiomatic Python for "if the list is not empty," and why `if some_string:` means "if the string is not empty" — you'll see both constantly once loops and conditions are combined.

### `and` / `or` don't always return `True` or `False`

Here's the part that genuinely surprises people coming from other languages. `and` and `or` in Python don't evaluate down to a boolean — they evaluate down to *one of their two operands, unchanged*.

````python
print("hello" and 10)
print(0 or 10)
print("" or "default")
````

The rule: `a and b` evaluates `a`. If `a` is falsy, the whole expression short-circuits and returns `a` immediately, without even looking at `b` — because if `a` is falsy, the `and` can't possibly end up true no matter what `b` is, so there's no point evaluating it. If `a` is truthy, the result of the whole expression is whatever `b` is (which could itself be truthy or falsy — `and` doesn't force a boolean at the end, it just returns whichever operand *decided* the outcome).

`"hello" and 10`: `"hello"` is truthy, so `and` moves on and returns `10` — whatever `b` turned out to be, without judging it further. `0 or 10`: `0` is falsy, so `or` moves on and returns `10`. `"" or "default"`: `""` is falsy, so `or` returns `"default"`.

This isn't a quirky edge case to memorize and forget — it's a genuinely common idiom for supplying a **default value**:

````python
username = ""
display_name = username or "Guest"
print(display_name)
````

If `username` is a non-empty string, `display_name` becomes that string (since it's truthy, `or` returns it as-is). If `username` is empty, it's falsy, and `or` falls through to `"Guest"`. This one line replaces what would otherwise be several lines of `if`/`else`.

### Short-circuiting

The mechanism above has a name — **short-circuit evaluation** — and it does something beyond just deciding a value: it can skip evaluating the second operand entirely.

````python
def noisy_true():
    print("evaluated the right side")
    return True

print(False and noisy_true())
print(True or noisy_true())
````

Neither line printed `"evaluated the right side"`. In the first, `False and ...` already knows the answer must be falsy the instant it sees `False`, so it never bothers calling `noisy_true()` at all. In the second, `True or ...` already knows the answer must be truthy, for the same reason. (We haven't covered `def` yet — treat `noisy_true()` here only as "a thing that would print a message if it actually ran," to observe whether it runs.)

This matters in real code whenever the second condition would be unsafe or wasteful to evaluate on its own — checking that something exists *before* checking a property of it, for instance:

```python
if name != "" and name[0] == "A":
    ...
```

If `name` is empty, `name[0]` would raise an error (there's no index `0` in an empty string) — but because `name != ""` is checked first and short-circuits when false, `name[0]` never gets evaluated when it would have failed.

### Precedence, briefly

`not` binds tighter than `and`, which binds tighter than `or` — meaning `not` is applied first, then `and`-groups are formed, then `or` combines those groups.

````python
print(True or False and False)   # and happens first: True or (False and False)
````

As with arithmetic, the right move when an expression combines more than two of these isn't to recite the precedence table — it's to add parentheses that make the grouping visible to the next reader, who very well might be you in three months.

## 12. `if` / `elif` / `else`

Every piece of code you've written so far runs unconditionally, top to bottom. Real programs need to make choices — do one thing if something is true, another thing otherwise. That's what `if` is for.

````python
age = 20

if age >= 18:
    print("You can vote.")
````

The structure: `if`, a condition, a colon, then an indented block. The indentation isn't decoration — it's how Python knows which lines belong to the `if`. Change the age and predict what happens before running:

````python
age = 15

if age >= 18:
    print("You can vote.")

print("This always runs.")
````

The second `print` isn't indented under the `if`, so it runs regardless of the condition. Indentation in Python isn't just a style preference the way it is in many languages — it's part of the syntax. A misplaced indentation level changes what the code actually does, not just how it looks.

### Adding an alternative

`else` gives you a block that runs exactly when the `if` condition was false: 

````python
age = 15

if age >= 18:
    print("You can vote.")
else:
    print("Not old enough yet.")
````

### More than two branches

`elif` (short for "else if") lets you check additional conditions in sequence, without nesting one `if` inside another `else`.

````python
score = 72

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)
````

Something worth noticing here: only **one** branch runs, even if more than one condition would have been true. `score >= 70` is also technically true when `score` is `95`, but because `score >= 90` is checked *first* and matches, Python never even evaluates the later conditions. Order matters — conditions should generally go from most specific to least specific when they overlap like this.

### Nesting

Conditions can contain other conditions:

````python
age = 25
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed.")
    else:
        print("Bring ID next time.")
else:
    print("Too young.")
````

This works, but it's worth noticing that the same logic could often be written flatter, using `and` from the last chapter:

````python
age = 25
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")
elif age >= 18:
    print("Bring ID next time.")
else:
    print("Too young.")
````

Neither version is "more correct" — but as conditions nest more than two or three levels deep, code tends to become harder to follow, and it's worth periodically asking whether a boolean expression could replace a level of nesting.

## 13. `while` loops

Everything so far runs each line at most once. A loop lets a block of code run repeatedly, for as long as some condition holds.

````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

### Tracing it step by step

It's worth watching this loop the way the interpreter does, one iteration at a time, rather than skimming it as "prints 1 to 5":

1. `count` is `1`. Check `count <= 5` → `True`. Enter the loop body: print `1`. Then `count = count + 1` makes `count` become `2`.
2. Back to the top. Check `2 <= 5` → `True`. Print `2`. `count` becomes `3`.
3. Check `3 <= 5` → `True`. Print `3`. `count` becomes `4`.
4. Check `4 <= 5` → `True`. Print `4`. `count` becomes `5`.
5. Check `5 <= 5` → `True`. Print `5`. `count` becomes `6`.
6. Check `6 <= 5` → `False`. The loop body does not run. Execution continues on whatever comes after the loop.

Every `while` loop needs three ingredients, and it's worth naming them because forgetting one is the most common way loops go wrong: something to check (the condition), a **starting state** (`count = 1`, set up *before* the loop), and something inside the loop that **changes that state** (`count = count + 1`) so the condition eventually stops being true.

### What happens if you forget the update

Leave out `count = count + 1` and the condition never changes — the loop runs forever. This is called an **infinite loop**, and it's a real risk with `while` specifically (you'll see in the next chapter that `for` loops over a fixed range don't have this particular failure mode, because the sequence to iterate over is already fully decided before the loop starts). If a notebook cell seems to hang without producing output, an infinite loop is the first thing to suspect — interrupt it and check your loop's update step.

### `while` for "repeat until something happens," not just counting

Counting loops are the easiest example, but `while`'s real strength is a condition that isn't a simple counter — for instance, repeating until a user provides valid input:

````python
# (This cell will actually wait for your input if you run it.)
number = int(input("Enter a positive number: "))

while number <= 0:
    print("That wasn't positive.")
    number = int(input("Enter a positive number: "))

print("Got it:", number)
````

Here the "state" being tracked isn't a counter — it's whatever the user most recently typed. The loop keeps running exactly as long as that value stays invalid.

## 14. `for` loops and `range()`

It's tempting to describe `for` as "a loop that repeats a fixed number of times," but that description undersells what's actually going on, and it'll cause confusion later when you meet `for` loops that don't count anything at all. The better description: **`for` repeats a block once for each item in some sequence of values, in order.**

Start with something you already understand as a sequence — a string:

````python
for letter in "cat":
    print(letter)
````

No indexing, no counting, no manual "current position" variable. `letter` is automatically set to each character in turn — `'c'`, then `'a'`, then `'t'` — and the loop body runs once per value. This is called **iteration**, and it's the actual concept `for` exists to express: "do this once for every item in that."

### `range()` — generating a sequence of numbers to iterate over

Most of the time, you don't already have a sequence lying around — you want to repeat something a specific number of times, or count through a range of numbers. `range()` produces exactly that: a sequence of numbers, which `for` can then iterate over just like it iterated over the characters of a string.

````python
for n in range(5):
    print(n)
````

`range(5)` produces `0, 1, 2, 3, 4` — five values, starting at `0`, stopping *before* `5`. If that "stop is excluded" rule sounds familiar, it should: it's the exact same convention slicing used in Chapter 8, and for the same reason — it keeps `range(n)` producing exactly `n` values, with no off-by-one arithmetic required.

`range()` accepts up to three arguments, mirroring slicing again:

```python
range(stop)               # 0, 1, ..., stop-1
range(start, stop)        # start, start+1, ..., stop-1
range(start, stop, step)  # start, start+step, ..., up to (not including) stop
```

````python
for n in range(2, 8):
    print(n)
````

````python
for n in range(0, 10, 2):
    print(n)
````

````python
for n in range(10, 0, -1):
    print(n)
````

A negative step, exactly as with slicing, means "count downward" — and just as with slicing, `start` needs to be on the correct side of `stop` for that direction, or you'll get an empty sequence rather than an error.

### The relationship between `for` and iterable values

Strings and the values `range()` produces are both examples of something Python calls **iterable** — a value you can hand to `for` and get its items one at a time, in order. You don't need the formal definition yet (it becomes precise later, once you've met more collection types), but it's worth planting the term now: *iterable* is the general category, and strings and ranges are two specific members of it. Over the next several chapters — lists, tuples, dictionaries, sets — you'll keep meeting new iterable types, and `for` will work on every one of them the same way, without you needing to learn a new looping syntax each time. That reuse is exactly why the concept is worth naming precisely instead of just saying "for repeats things."

### `for` when you also need the count

Sometimes you're iterating over something and also want to know which position you're at. Rather than maintaining a separate counter by hand, `enumerate()` gives you both at once — mentioned here briefly, because you'll see it used naturally in later chapters:

````python
for index, letter in enumerate("cat"):
    print(index, letter)
````

## 15. `break`, `continue`, and loop `else`

Sometimes a loop needs to stop early, or skip part of its body, based on something discovered *inside* the loop rather than in the loop's original condition. `break` and `continue` handle those two cases.

### `break` — stop the loop entirely

````python
for n in range(1, 20):
    if n == 7:
        break
    print(n)

print("loop ended")
````

As soon as `n == 7`, `break` exits the loop immediately — not just that iteration, the *whole loop*. Execution jumps straight to whatever comes after it. Notice `7` itself never got printed: `break` runs before the `print(n)` below it gets a chance.

A realistic use — searching for something and stopping the moment it's found, rather than needlessly checking everything after it:

````python
numbers = [4, 9, 15, 22, 30, 41]
target = 22

for n in numbers:
    if n == target:
        print("found it")
        break
````

### `continue` — skip to the next iteration

````python
for n in range(1, 10):
    if n % 2 == 0:
        continue
    print(n)
````

`continue` doesn't end the loop — it ends the *current* iteration early and jumps straight back to the loop's next value. Even numbers hit `continue` before reaching `print(n)`, so they're skipped; odd numbers fall through past the `if` and get printed. The loop as a whole still runs through every value in `range(1, 10)` — it's just that some iterations do nothing.

### Loop `else` — the part almost nobody expects

`while` and `for` loops can both have an `else` clause, and it means something specific enough that it's worth being precise about it rather than guessing from the word "else."

**The loop's `else` block runs if the loop finished normally — that is, if the loop was *not* stopped early by a `break`.**

````python
for n in range(5):
    print(n)
else:
    print("loop completed without break")
````

Now compare it with a `break` involved: 

````python
for n in range(5):
    if n == 3:
        break
    print(n)
else:
    print("loop completed without break")
````

The `else` block's message doesn't print this time, because the loop was cut short by `break`. This is genuinely the entire rule — `else` on a loop means "only if `break` never happened" — and it's worth restating because "loop else" is one of the more commonly misunderstood corners of Python, precisely because the word `else` suggests something closer to "otherwise," which isn't quite what's happening here.

This is most useful for exactly the kind of search shown a moment ago — checking whether something was found, without a separate flag variable to track it:

````python
numbers = [4, 9, 15, 22, 30, 41]
target = 99

for n in numbers:
    if n == target:
        print("found it")
        break
else:
    print("not found anywhere in the list")
````

Without loop `else`, you'd normally need a separate boolean — something like `found = False`, set to `True` inside the loop, checked afterward — just to know whether the loop ever broke. The loop `else` clause removes the need for that bookkeeping variable, for this one specific pattern.

## 16. Lists

You've been iterating over strings and ranges, both sequences of a single fixed kind of thing (characters, numbers). A **list** is a sequence too, but it can hold a mix of any values, and — this is the important new part — it can be changed after it's created.

````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(type(fruits))
````

Everything you learned about indexing and slicing strings applies to lists without modification, because both are sequences and Python's sequence operations are shared across sequence types rather than reinvented for each one.

````python
fruits = ["apple", "banana", "cherry", "date"]

print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
print(fruits[::-1])
````

The index diagram from Chapter 8 applies here exactly as it did for strings — a list of four items has valid indexes `0` through `3`, or `-4` through `-1` counting backward.

`in` checks membership, same as it did for strings:

````python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("mango" in fruits)
````

And `for` iterates over a list's elements directly, the same way it iterated over a string's characters — because a list, like a string, is iterable:

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

### The difference that actually matters: mutability

A string can't be changed after creation; a list can. This single difference is worth returning to the two-names-one-value diagram from Chapter 4, because it behaves *differently* here in a way that catches almost everyone off guard the first time.

````python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
print(b)
````

What do you expect `a` to print, before running it?

Both `a` and `b` show `[1, 2, 3, 4]`. This looks alarming at first, but it's the direct, logical consequence of what you already learned about assignment: `b = a` doesn't copy the list — it makes `b` refer to the *same* list object `a` already refers to, exactly like `y = x` made `y` refer to the same `10` as `x` back in Chapter 4.

```text
a ─────┐
       ├────► [1, 2, 3]
b ─────┘
```

The difference is what happens *next*. With numbers, `x = 20` redirected `x` to a brand-new value, leaving `y` untouched. But `b.append(4)` does not redirect `b` to a new list — it reaches into the *existing* list object and changes it in place. Since `a` and `b` still both point at that same object, both names see the change, because there was only ever one list to see.

```text
a ─────┐
       ├────► [1, 2, 3, 4]   (same object, modified in place)
b ─────┘
```

This is the payoff of the distinction the book asked you to hold onto back in Chapter 4: "redirecting a name" and "modifying the object a name points to" are two completely different actions, and they behave identically for immutable values (because there's no way to modify what can't change) but very differently for mutable ones. If you actually want an independent copy, you have to ask for one explicitly:

````python
a = [1, 2, 3]
b = a.copy()

b.append(4)

print(a)
print(b)
````

Now `a` and `b` refer to two separate list objects, and changing one doesn't touch the other.

## 17. Modifying lists

Because lists are mutable, Python gives them methods that change the list in place rather than producing a new one. This is the chapter to read carefully if the previous one's `a`/`b` surprise still feels uncomfortable — every method here relies on exactly that same behavior.

````python
numbers = [3, 1, 4]

numbers.append(1)      # add to the end
print(numbers)

numbers.insert(0, 9)   # insert 9 at index 0
print(numbers)

numbers.remove(1)      # remove the FIRST 1 it finds, not all of them
print(numbers)

last = numbers.pop()   # remove and return the last item
print(last, numbers)
````

Notice `.remove(1)` removed only the first `1` it found, even though the list originally had two — that's specific to `remove()`, which always removes the first matching value and stops. `.pop()` is slightly different in kind from the others: it doesn't just modify the list, it also *hands back* the value it removed, which is why `last = numbers.pop()` makes sense as a line but `last = numbers.append(1)` would just set `last` to `None` — `append` doesn't return anything meaningful, because there's nothing useful to hand back from "add this to the end."

`.sort()` orders the list in place:

````python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
numbers.sort()
print(numbers)
````

### `.sort()` versus `sorted()` — a distinction worth keeping precise

Here's a pair that looks almost identical and behaves in genuinely different ways.

````python
a = [3, 1, 4, 1, 5]
result = a.sort()

print(a)
print(result)
````

`a.sort()` reordered `a` in place — and returned `None`, which is why `result` printed as `None` rather than as a sorted list. This is Python's convention across the mutating list methods: a method that changes something in place generally returns `None`, as a signal that its purpose was the *side effect*, not a value to use.

Now the alternative:

````python
b = [3, 1, 4, 1, 5]
result = sorted(b)

print(b)
print(result)
````

`sorted(b)` is a **function**, not a method on the list, and it leaves `b` completely untouched — it builds and returns a brand-new sorted list instead. This distinction — "operations that mutate the existing object and return `None`" versus "operations that leave the original alone and return something new" — is one of the more important habits to build now, because it generalizes: you'll meet the same split again with strings' methods (all of which return new strings, since strings can't be mutated at all) and, later, with several list operations that come in both a mutating and non-mutating flavor. When in doubt about a specific method, the fastest way to check is exactly what you just did above: assign its result to a name and print both that name and the original.

## 18. Tuples

A tuple looks almost like a list — an ordered sequence of values — but with parentheses instead of square brackets, and one deliberate restriction: once created, it cannot be changed.

````python
point = (3, 4)
print(point)
print(point[0], point[1])
````

````python
point = (3, 4)
point[0] = 10
````

That error — the same `does not support item assignment` you saw for strings back in Chapter 7 — is the whole point of tuples. A tuple is Python's **immutable** sequence, the way a string is Python's immutable sequence of characters. That raises a real question, though: why have an immutable version of something a list already does, mutably?

### Why tuples exist

The honest answer is: immutability is a *feature* in specific situations, not a limitation. A value that can't change is one you can hand to other code with a guarantee attached — "this will still look the same later," no matter what that other code does with it. Lists don't offer that guarantee; tuples do.

One concrete place this matters, sooner than you might expect: dictionary keys, which you'll meet in the next chapter, must be immutable — you can use a tuple as a key, but never a list, for exactly this reason (a key that could silently change out from under the dictionary would break the whole idea of "look something up by this key").

There's also a simple readability signal in choosing a tuple: writing `point = (3, 4)` for a coordinate pair suggests "this is a fixed, small, structural grouping of values," where a list of the same two numbers would (correctly) invite the reader to wonder whether it might grow, shrink, or get reordered later. Neither type is "better" in general — they answer different questions about intent.

### Tuple unpacking

Tuples work naturally with the multiple-assignment syntax from Chapter 4, and this combination is used constantly in real code, well beyond anything explicitly called a "tuple":

````python
point = (3, 4)
x, y = point
print(x, y)
````

This is called **unpacking** — the tuple's values are distributed into `x` and `y` in order, in one step. It works for any sequence with the right number of items, not just literal tuples, but tuples are where you'll see it introduced most often, because a function that needs to hand back more than one value almost always does so by returning a tuple:

````python
def divide_with_remainder(a, b):
    return a // b, a % b   # this builds a tuple, even without explicit parentheses

quotient, remainder = divide_with_remainder(17, 5)
print(quotient, remainder)
````

(We haven't covered `def` yet — the point here is only to notice that `a // b, a % b` silently becomes a tuple, and that the calling side unpacks it right back into two separate names. You'll return to this exact pattern once functions get their own chapter.)

## 19. Dictionaries

Every collection so far has organized values by *position* — index `0`, index `1`, and so on. Sometimes position isn't the natural way to look something up. If you're storing a person's age, you don't want to remember "age is at position 3" — you want to ask for it by name. A **dictionary** stores values by a key you choose, instead of by position.

````python
person = {"name": "Ali", "age": 25, "city": "Lahore"}
print(person)
````

Each entry is a **key** (`"name"`, `"age"`, `"city"`) paired with a **value** (`"Ali"`, `25`, `"Lahore"`). Look something up by key, not by position — there's no `person[0]` here; position was never the organizing idea.

````python
person = {"name": "Ali", "age": 25, "city": "Lahore"}
print(person["name"])
print(person["age"])
````

Looking up a key that doesn't exist raises an error rather than guessing:

````python
person = {"name": "Ali", "age": 25}
print(person["email"])
````

### `.get()` — looking something up without risking that error

`.get()` returns `None` (rather than raising an error) if the key isn't present, and optionally lets you specify a different fallback value: 

````python
person = {"name": "Ali", "age": 25}
print(person.get("email"))
print(person.get("email", "not provided"))
````

The choice between `person["email"]` and `person.get("email")` is really a choice about what a missing key *means* in your program. If a missing key is a bug — data that should always be there — let the error happen; it tells you something is wrong immediately. If a missing key is a normal, expected situation, `.get()` with a sensible fallback is the better tool.

### Insertion, modification, deletion

Dictionaries are mutable, same as lists — assigning to a key that exists changes its value; assigning to a key that doesn't exist yet creates it.

````python
person = {"name": "Ali", "age": 25}

person["age"] = 26          # modify existing key
person["email"] = "ali@example.com"   # create a new key

print(person)
````

````python
del person["email"]
print(person)
````

### Iterating through a dictionary

A dictionary offers three different views for iteration, depending on what you actually need — keys alone, values alone, or both together.

````python
person = {"name": "Ali", "age": 25, "city": "Lahore"}

for key in person.keys():
    print(key)
````

````python
for value in person.values():
    print(value)
````

````python
for key, value in person.items():
    print(key, "->", value)
````

`.items()` is the one worth reaching for most often once you need both pieces at the same iteration — it hands you a `(key, value)` pair each time through, which is then unpacked into two names in the same motion you saw with tuples in the last chapter (that's not a coincidence; each item really is a tuple under the hood).

Plain `for key in person:` (with no `.keys()` at all) also works and means the same thing as `.keys()` — worth knowing, because you'll see both styles in real code.

### Building the concept from the problem it solves

It's worth stepping back and asking why any of this is better than, say, two parallel lists — one of names, one of ages, kept in matching order. That approach technically works, but it forces you to keep two separate structures perfectly synchronized by hand, and looking something up means searching for a position in one list to then use in the other. A dictionary collapses that into one structure where "look this key up" is the entire operation — no searching, no separate position-tracking. That's the actual problem dictionaries exist to solve: fast, direct lookup by a meaningful label, rather than an arbitrary position you'd otherwise have to remember or recompute.

## 20. Sets

A **set** is an unordered collection with one defining rule: it can never contain duplicate values.

````python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)
````

Notice the duplicates simply vanished — a set doesn't complain about them, it just never stores more than one of the same value in the first place. This makes sets the natural tool whenever "does this collection contain duplicates" is exactly the wrong question to be asking, because the answer is always no by construction.

A very common use: removing duplicates from an existing list, by converting it to a set and back.

````python
names = ["Ali", "Sara", "Ali", "Bilal", "Sara"]
unique_names = list(set(names))
print(unique_names)
````

(The order of `unique_names` isn't guaranteed to match the original list — sets don't track position, which is the trade-off for the uniqueness guarantee and for the fast membership checks described next.)

### Membership and creation

`in` works on sets exactly as it does on the other collections you've seen, and it's typically *faster* on a large set than on an equivalently large list — a detail you don't need to act on yet, but worth knowing exists, since it's part of why sets are chosen over lists in real code, not just for the deduplication trick.

````python
primes = {2, 3, 5, 7, 11}
print(7 in primes)
print(9 in primes)
````

One easy trap: `{}` alone creates an empty **dictionary**, not an empty set — a leftover of dictionaries getting `{}` first. Use `set()` to create an empty set.

````python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
````

### Set operations — connecting to the mathematical idea

If you've seen set notation in mathematics, this will feel immediately familiar, because Python's set operations are named after exactly those operations.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # union -- everything in either
print(a & b)   # intersection -- only what's in both
print(a - b)   # difference -- in a, but not in b
print(a ^ b)   # symmetric difference -- in exactly one, not both
````

These read almost exactly like a Venn diagram would describe them: `a | b` is the whole shaded area of both circles; `a & b` is only the overlap; `a - b` is a's circle with the overlap cut out; `a ^ b` is everything *except* the overlap. If you've never seen the mathematical versions, that's fine — the code above is a perfectly good way to learn what they mean, since the two are really the same idea expressed twice.

## 21. Nested data structures

Real data is rarely as flat as the examples so far. A list can hold dictionaries; a dictionary can hold lists; any of these can hold each other, as deeply as the problem actually needs. Nothing new needs to be learned here syntactically — this chapter is about getting comfortable reading structures that combine what you already know.

````python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Usman", "marks": 92},
]

print(students)
````

This is a list containing two dictionaries. Reading `students[1]["marks"]` isn't one operation — it's two operations, applied one after the other, and the way to avoid confusion is to do them one at a time in your head (or, while learning, literally in separate cells) rather than trying to parse the whole expression in one glance.

### Breaking it down step by step

`students[1]` — indexing the list first, same as any list indexing you've done before:

````python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Usman", "marks": 92},
]

second_student = students[1]
print(second_student)
print(type(second_student))
````

`students[1]` evaluates to a single dictionary — `{"name": "Usman", "marks": 92}` — because that's what's sitting at index `1` of the outer list. Nothing about the presence of a dictionary inside changes how list indexing works; the list doesn't know or care what its elements are.

Now the second step operates on *that* dictionary:

````python
marks = second_student["marks"]
print(marks)
````

Put back together as one expression, `students[1]["marks"]` means exactly "take index `1` of `students`, then look up `"marks"` in whatever that gave you" — read left to right, each bracket operating on the result of everything before it, not on the original `students` list directly.

````python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Usman", "marks": 92},
]

print(students[1]["marks"])
print(students[0]["name"])
````

### Modifying something nested

Because both lists and dictionaries are mutable, you can reach into a nested structure and change a value in place, exactly the way you'd change a top-level one — the nesting doesn't add any new rule, it just means the target of the assignment is a longer expression.

````python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Usman", "marks": 92},
]

students[0]["marks"] = 90
print(students)
````

### Iterating over nested structures

Combine `for` with the indexing you already know, and you can process every element of a nested structure without writing anything new — just the tools from Chapters 14 and 19, used together.

````python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Usman", "marks": 92},
    {"name": "Sara", "marks": 78},
]

for student in students:
    print(student["name"], "scored", student["marks"])
````

Each pass through the loop, `student` refers to one dictionary from the list — exactly as `letter` referred to one character of a string back in Chapter 14. The habit to build here is: whenever a structure looks intimidating, ask what the *outermost* layer is (a list? a dictionary?), handle that one layer using tools you already have, and only then look at what's sitting inside each piece. Nested data is never a new concept — it's familiar tools applied more than once in sequence.

This is also, deliberately, as far as this book takes structured data. Real data analysis work — spreadsheets, tables, large datasets — uses dedicated tools (Pandas, among others) built on top of exactly these fundamentals, and that's a subject for later in the curriculum, not here. What you should walk away with is the ability to read and trace a nested structure by hand, which is the skill those later tools all assume you already have.

## 22. Writing your first functions

Look back at the last chapter's problem solution (once you've written it) and imagine needing to find "the top scorer" again, for a different list, somewhere else in a longer program. Copying the loop again would work, but it duplicates logic that should only need to be written once. A **function** is how you package a piece of behavior under a name, so it can be reused without being retyped.

````python
def greet():
    print("Hello!")
````

Running that cell does nothing visible — and that's expected. `def` **defines** the function; it doesn't run it. This is worth sitting with, because it's a genuinely new kind of statement compared to everything before this chapter: you're not computing a value or printing anything, you're teaching Python a new piece of behavior that it will remember and can perform later, on request.

To actually run it, **call** it — by writing its name followed by parentheses:

````python
greet()
````

Now `"Hello!"` prints. `greet` (no parentheses) refers to the function itself, as a value — you could print it, or pass it around, though you won't need to yet. `greet()` (with parentheses) means "run it now." That distinction — the name of a thing versus calling the thing — matters enough that mixing them up is one of the most common early function mistakes:

````python
print(greet)     # the function object itself
print(greet())   # calling it -- greet() runs, prints "Hello!", and evaluates to None
````

Notice `print(greet())` produced *two* lines: `"Hello!"` (printed by the function while it ran) and then `None` (the value `greet()` itself evaluated to, since `greet` never explicitly returns anything — you'll cover `return` properly next chapter). Those are two separate events, easy to conflate: the function *doing* something, and the function *evaluating to* something, are not the same thing, exactly as was true for `print()` itself back in Chapter 3.

### Why functions exist

The obvious answer — "to avoid retyping code" — is true but incomplete. A function also gives a piece of behavior a **name**, and a well-chosen name is documentation that never goes out of date, because it's read every time the function is called. Compare reading fifteen lines of loop logic inline versus reading `top_scorer(students)` — the second tells you *what* is happening without forcing you to first work out *how*. This separation — using something without needing to know its internals at the moment you use it — is one of the most important ideas in programming generally, and functions are the first place you meet it properly.

### Defining versus calling — one more angle

A function's body — everything indented underneath the `def` line — does not run when the function is defined. It only runs each time the function is *called*, and it runs fresh every single time.

````python
def show_message():
    print("This runs every time show_message() is called.")

show_message()
show_message()
show_message()
````

Three calls, three separate executions of the same body — not something computed once and replayed. Each call is a fully independent run of the code inside the function.

## 23. Parameters, default values, and return values

`greet()` from the last chapter always does exactly the same thing, no matter what. Most useful functions need to behave differently depending on what they're given.

````python
def greet(name):
    print("Hello, " + name + "!")

greet("Ali")
greet("Sara")
````

`name` is a **parameter** — a name the function's body can use, standing in for whatever value gets passed in when the function is *called*. `"Ali"` and `"Sara"` are **arguments** — the actual values supplied at the call site. The distinction (parameter = the placeholder in the definition, argument = the real value at the call) is worth keeping precise, because you'll read error messages that use these words specifically.

A function can take more than one parameter, separated by commas, matched up with the arguments in the same order:

````python
def greet(name, greeting):
    print(greeting + ", " + name + "!")

greet("Ali", "Good morning")
````

### `print` versus `return` — a distinction worth taking seriously

Here is a function that seems, at a glance, to do the same thing as a function using `return`: 

````python
def add_v1(a, b):
    print(a + b)

def add_v2(a, b):
    return a + b

add_v1(3, 4)      # prints 7
result = add_v2(3, 4)
print(result)      # prints 7 too
````

Both printed `7` — but they got there in very different ways, and the difference matters the moment you try to *use* the result rather than just look at it.

````python
result_v1 = add_v1(3, 4)
result_v2 = add_v2(3, 4)

print(result_v1)
print(result_v2)
````

`add_v1(3, 4)` printed `7` (as a side effect) but the *call itself* evaluated to `None`, because `add_v1` never returns anything — `print` inside a function displays a value to the screen; it does not hand that value back to whoever called the function. `result_v1` ends up holding `None`, not `7`.

`add_v2(3, 4)` printed nothing on its own — but the call evaluated to `7`, because `return a + b` explicitly hands that value back to the caller. `result_v2` correctly holds `7`.

The rule underneath this: **`print` is for showing something to a person; `return` is for handing a value back into the program, so other code can keep using it.** A function meant to be *used* — its result stored, passed to something else, computed with further — needs `return`. A function meant only to *display* something and be done can use `print` and skip `return` entirely (in which case, as you saw last chapter, it implicitly returns `None`).

### `return` also ends the function immediately

Once a `return` statement runs, the function stops — nothing after it, in that function, runs during that call.

````python
def classify(n):
    if n < 0:
        return "negative"
    print("this only runs for non-negative n")
    if n == 0:
        return "zero"
    return "positive"

print(classify(-5))
print(classify(5))
````

For `classify(-5)`, the first `return` fires immediately, and the `print` line below it never executes for that call at all — not "executes but is ignored," genuinely never runs.

### Default parameter values

A parameter can specify a value to use automatically, if the caller doesn't provide one.

````python
def greet(name, greeting="Hello"):
    print(greeting + ", " + name + "!")

greet("Ali")
greet("Sara", "Welcome")
````

`greeting="Hello"` in the definition means "if nothing is given for `greeting`, use `\"Hello\"`." The caller is always free to override it, as `greet("Sara", "Welcome")` does. This is genuinely useful the moment a function has a parameter that's *usually* the same value, with occasional exceptions — it avoids forcing every single call site to repeat the common case explicitly.

## 24. Variable scope — local vs global

Run this, and think about what you expect before you do:

````python
def show_number():
    x = 10
    print(x)

show_number()
print(x)
````

The second `print(x)` fails with `NameError: name 'x' is not defined`. `x` was created *inside* `show_number`, and it simply doesn't exist outside of it. This isn't a bug to route around — it's a deliberate feature called **scope**, and it's one of the more important structural ideas a function introduces.

### Namespaces

Think of each function call as getting its own private notebook of names — a **local namespace** — separate from the notebook everything outside functions uses (the **global namespace**). A variable created inside a function is written into that function's own notebook, and once the function finishes running, that notebook is gone; nothing outside ever had access to it in the first place. This is exactly why `x` in the example above doesn't leak out — it was never in the outer notebook to begin with.

This is a genuine convenience, not just a restriction: it means a function's internal variable names can't accidentally collide with variables somewhere else in a large program, because each function's names live in their own separate space.

### Reading a global from inside a function

A function *can* read a variable defined outside it, as long as it isn't also assigning to a variable of the same name inside itself: 

````python
message = "Hello from the outside"

def show_message():
    print(message)

show_message()
````

### What happens when a local name shadows a global one

Here's the case that catches people, because it looks like it should update the global variable and doesn't.

````python
count = 100

def reset_count():
    count = 0
    print("inside the function:", count)

reset_count()
print("outside the function:", count)
````

What do you expect the last line to print, before running it?

It prints `100` — the global `count` was never touched. The line `count = 0` inside `reset_count` doesn't modify the outer `count` at all; because there's an *assignment* to `count` somewhere inside the function, Python treats `count` as a brand-new **local** variable for the entire duration of that function call, entirely separate from the global one that happens to share its name. The local `count` shadows the global one inside the function's body — the global still exists, untouched, one namespace over.

This is worth restating precisely, because the rule is about *assignment*, not just *use*: if a function only *reads* a name (as `show_message` did above) without ever assigning to it, it reaches out to the global namespace. But the moment a function *assigns* to a name anywhere in its body, Python decides that name is local to the function for that entire call, even on lines *before* the assignment — which is a subtlety worth knowing exists, even without needing to construct that exact example right now.

If you genuinely want a function to modify a global variable, Python requires you to say so explicitly, with the `global` keyword — but that's a tool to reach for rarely, and only once you understand exactly why it's necessary, which is beyond what this chapter needs. For now, the important habit is: assume a function's variables are private to it, and use `return` (from the last chapter) to hand a value back out, rather than relying on a function to quietly change something outside itself.

## 25. Handling errors with `try` / `except`

Run this: 

````python
number = int("not a number")
````

A `ValueError`, and the program stops right there — nothing after this line, anywhere in the script, would run. Every crash you've seen so far in this book behaved this way: an error occurred, and execution simply halted. Sometimes that's the correct behavior — a genuine bug should be loud. But sometimes a failure is *expected* and *recoverable* — a user might reasonably type something that isn't a valid number, and the program shouldn't die because of it. `try` / `except` is how you handle that second case.

````python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("That wasn't a valid number.")
````

Python attempts everything inside `try`. If it completes without error, `except` is simply skipped, as if it weren't there. If an error occurs anywhere inside `try`, execution jumps immediately to the matching `except` block, and the rest of `try` — everything after the line that failed — never runs.

### Exceptions are runtime events, not compile-time checks

It's worth being precise about what an exception actually is: it's something that happens *while the program is running*, triggered by a specific line of code encountering a situation it can't handle — dividing by zero, converting invalid text, looking up a missing dictionary key, indexing past the end of a list. Python doesn't (and can't, in general) know in advance that a line will fail; it discovers this the moment that line actually executes, and at that point it **raises** an exception, which either gets *caught* by a `try`/`except` or, if there is none, propagates all the way up and stops the program.

### Naming the specific error

`except ValueError:` only catches a `ValueError` — the specific problem you anticipated. A different kind of error, like `TypeError`, would not be caught by that block, and would still crash the program, which is usually the correct behavior: catching errors you didn't anticipate and didn't design for often hides real bugs rather than fixing them.

````python
try:
    result = 10 / 0
except ValueError:
    print("This won't catch it -- wrong exception type.")
````

That's a `ZeroDivisionError`, and since the `except` clause only listens for `ValueError`, it doesn't match, and the program crashes anyway. Fix it by naming the actual exception the code can raise: 

````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Can't divide by zero.")
````

You can catch more than one kind in the same block when several distinct problems deserve the same response: 

````python
try:
    number = int(input("Enter a number to divide 100 by: "))
    print(100 / number)
except (ValueError, ZeroDivisionError):
    print("That didn't work -- give me a nonzero number.")
````

### `else` and `finally`

Two additional clauses exist, and each has one specific job worth knowing, without overcomplicating this chapter with every possible combination.

`else` runs only if `try` completed with **no** exception at all — the counterpart to the loop `else` you saw back in Chapter 15, and the same underlying idea: "only if nothing went wrong." 

````python
try:
    number = int("42")
except ValueError:
    print("invalid number")
else:
    print("parsed successfully:", number)
````

`finally` runs **no matter what** — whether `try` succeeded, failed and got caught, or failed and *wasn't* caught. It's for cleanup that has to happen regardless of outcome (closing a file, for instance, which the next chapter covers in more convenient form).

````python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("caught it")
finally:
    print("this always runs")
````

## 26. Reading and writing files

Everything you've stored in a variable so far disappears the moment your program ends. Working with files is how a program keeps something around after it's finished running.

### Writing to a file

````python
file = open("notes.txt", "w")
file.write("First line\n")
file.write("Second line\n")
file.close()
````

`open(..., "w")` opens a file for **writing** — `"w"` specifically means "start from empty," so if `notes.txt` already had content, it's gone the moment this line runs. `.write()` adds text to it (notice the explicit `\n` — unlike `print()`, `.write()` does not add a newline for you, so you have to ask for one yourself if you want each call on its own line). `.close()` finishes the operation and releases the file.

### Reading a file back

````python
file = open("notes.txt", "r")
contents = file.read()
file.close()

print(contents)
````

`"r"` means read (and is actually the default mode, if you leave the argument out entirely). `.read()` pulls in the file's entire contents as one string.

You can also read a file line by line, which is usually the more natural approach once a file has any real size:

````python
file = open("notes.txt", "r")
for line in file:
    print("Line:", line.strip())
file.close()
````

A file, it turns out, is itself iterable — `for line in file` works for exactly the same underlying reason `for letter in "cat"` did back in Chapter 14: something that produces a sequence of values, one at a time, can be handed directly to `for`. The `.strip()` call removes the trailing newline character each line comes with, since `print()` would otherwise add a second one on top of it.

### Why closing matters, and a cleaner way to guarantee it

Forgetting `.close()` is a real, common bug — it can leave data unwritten (the operating system may buffer what you've written and only actually commit it to disk when the file is closed) or leave the file locked against other programs. Manually calling `.close()` also means that if an error happens *between* `open()` and `.close()`, the file never gets closed at all, because the crash skips right past that line.

Python provides a cleaner pattern for exactly this problem — the `with` statement: 

````python
with open("notes.txt", "r") as file:
    contents = file.read()

print(contents)
````

This does the same thing as the manual open-and-close version, but it **guarantees** the file gets closed once the indented block finishes — even if an error occurs partway through. You don't need to understand exactly *how* `with` arranges that guarantee yet (the general mechanism, called a context manager, is a topic for later) — for now, treat `with open(...) as file:` as the standard, safer way to work with a file, and prefer it over calling `.close()` by hand.

### Appending instead of overwriting

`"a"` opens a file for appending — writing new content onto the end, without erasing what's already there, unlike `"w"`.

````python
with open("notes.txt", "a") as file:
    file.write("Third line\n")

with open("notes.txt", "r") as file:
    print(file.read())
````

## 27. Organizing code into multiple files

Every function and variable in this notebook has lived in one place. As a program grows, keeping everything in a single file stops being practical — related pieces of code get scattered among unrelated ones, and reusing something in a different project means copying and pasting it. A **module** is simply a `.py` file that other files can borrow code from.

Since a notebook can't easily demonstrate two separate files side by side, this chapter writes one to disk directly, then imports it — the exact same mechanism you'd use with two ordinary files sitting next to each other in a folder.

````python
with open("greetings.py", "w") as file:
    file.write(
        "def greet(name):\n"
        "    return \"Hello, \" + name + \"!\"\n"
        "\n"
        "default_greeting = \"Hi there\"\n"
    )
````

That created `greetings.py`, a file containing one function and one variable — an entirely ordinary Python file, no different in kind from any file you'd create yourself in an editor. Now, from a different piece of code, that file can be imported: 

````python
import greetings

print(greetings.greet("Ali"))
print(greetings.default_greeting)
````

`import greetings` runs `greetings.py` once, and makes everything it defined available through the name `greetings` — using `greetings.greet(...)` the way you'd use `.append()` on a list: a dot, followed by the specific thing you want from inside it. This is the exact same dotted syntax you've already been using for methods; a module is just another thing you can put a dot after.

### Importing specific names directly

If you don't want to keep typing `greetings.` in front of everything, you can import specific names directly into your own file's namespace: 

````python
from greetings import greet

print(greet("Sara"))   # no "greetings." prefix needed now
````

This is a genuine trade-off, not a strictly better option: `import greetings` keeps every borrowed name clearly marked with where it came from (`greetings.greet`), which makes large programs easier to read, at the cost of a little extra typing. `from greetings import greet` is shorter, but if two different modules both had a function called `greet`, importing both this way would silently collide, and it wouldn't be obvious from the call site which one you meant. Neither style is a beginner's mistake to grow out of — professional code uses both, chosen deliberately per situation.

### What a module actually is

There's nothing magical about a module — it's simply a `.py` file, and `import` is simply a way of running that file and keeping hold of the names it created, the same names that would exist if you'd typed the same code directly into your own file. This is why the function inside `greetings.py` didn't need any special "make this importable" declaration — every `.py` file is automatically importable, by its filename, from any other file in the same folder.

## 28. Formatting output properly — f-strings

Every `print()` call so far that combined text and values used `+` or commas — workable, but clumsy the moment more than one or two values are involved, and outright awkward once you need to control *how* a number is displayed (how many decimal places, for instance). An **f-string** solves both problems at once.

````python
name = "Usman"
age = 19

print(f"My name is {name} and I am {age}.")
````

The `f` immediately before the opening quote is what makes this an f-string rather than an ordinary one. Inside it, anything between `{` and `}` is not literal text — it's a Python expression, evaluated on the spot and inserted into the string at that position. Compare it to the equivalent written the old way: 

````python
print("My name is " + name + " and I am " + str(age) + ".")
````

Notice the f-string version didn't need `str(age)` — inside `{}`, Python converts the value to text automatically, regardless of its type, which is one of the concrete conveniences f-strings offer beyond just being shorter to type.

You aren't limited to a single variable name inside the braces — any expression is valid: 

````python
a = 5
b = 3
print(f"{a} + {b} = {a + b}")
````

### Controlling how a number is displayed

This is where f-strings earn the "formatting" half of their name. A colon inside the braces introduces a **format spec** — instructions for exactly how the value should be rendered.

````python
price = 19.9
print(f"{price:.2f}")
````

`.2f` means "fixed-point notation, with exactly two digits after the decimal point." Run it and compare against plain `price`, and the difference is immediate — `19.9` becomes `19.90`, padded to the requested precision.

````python
price = 19.9
print(price)
print(f"{price:.2f}")

pi_ish = 3.14159265
print(f"{pi_ish:.2f}")
print(f"{pi_ish:.4f}")
````

This matters for more than cosmetics: prices, percentages, and measurements almost always need a fixed, predictable number of decimal places, and relying on however many digits a float happens to produce naturally (which can be a long, ugly, imprecise-looking string of digits) looks unprofessional and can be genuinely misleading about how precise a value actually is.

A few other format specs worth knowing exist, without needing to memorize the full list right now: 

````python
large_number = 1234567
print(f"{large_number:,}")     # comma as a thousands separator

percentage = 0.4567
print(f"{percentage:.1%}")     # format as a percentage, one decimal place
````

## 29. Finally: type hints

Every function so far has looked like this:

```python
def add(a, b):
    return a + b
```

Nothing in that definition says what `a` and `b` are supposed to be — Python doesn't require it, and the function will happily run with numbers, strings (producing concatenation instead of addition), or anything else `+` is defined for. That flexibility is genuinely part of the language's design, not an oversight, but it comes at a cost: reading `add`'s definition alone doesn't tell you what it expects.

**Type hints** let you write that expectation down, without changing how the function behaves at all.

````python
def add(a: int, b: int) -> int:
    return a + b

print(add(3, 4))
````

`a: int` means "`a` is intended to be an `int`." `-> int` means "this function is intended to return an `int`." Read the whole signature as a sentence: "`add` takes two integers and returns an integer."

Variables can be annotated the same way: 

````python
age: int = 19
price: float = 19.99
name: str = "Ali" 
````

### What type hints do — and, just as important, what they don't do

Here is the detail worth being completely unambiguous about: **type hints are not enforced by Python at runtime.** Writing `a: int` does not make Python check, reject, or convert anything. 

````python
def add(a: int, b: int) -> int:
    return a + b

print(add("3", "4"))   # runs anyway -- Python never checked the hint
````

That returned `"34"` (string concatenation), despite the hint clearly saying `a` and `b` should be `int`. Python read the hint, and then completely ignored it while actually running the function. This surprises people who assume type hints work like they might in a language that enforces types strictly — Python's don't, by design.

### So what are they actually for?

Type hints exist for **people and tools**, not for the interpreter. Specifically:

- **Readability.** A function signature with hints tells the next reader what's expected without them needing to read the function body or guess from its name.
- **External tools.** Editors and separate programs called type checkers (not something this book covers) can read your hints and warn you — *before* you ever run the code — if you're about to call a function with the wrong kind of value. That checking happens entirely outside the Python interpreter itself, as a separate step you choose to run.
- **Documentation that can't silently go stale as easily as a comment.** A comment describing a parameter can drift out of sync with the code over time, because nothing connects the comment's text to the code's behavior. A type hint sits directly in the signature and gets read by tooling, which gives it at least a chance of being checked and kept honest, even though Python itself still won't enforce it.

None of this changes anything about how your programs behave once you add hints. It's a way of communicating intent — to your future self, to other people reading the code, and to tools that can catch certain classes of mistakes for you before runtime, if you choose to use them.

### One more shape worth knowing: hinting a collection

You can also hint what a list or dictionary is expected to contain: 

````python
def average(numbers: list[float]) -> float:
    return sum(numbers) / len(numbers)

print(average([1.0, 2.0, 3.0]))
````

`list[float]` means "a list, expected to contain floats." Again — purely informational; Python won't stop you from passing a list of strings instead.

That's genuinely as far as type hints need to go for now. The full `typing` module, with more elaborate hint shapes, belongs to the Intermediate phase of this curriculum, once you have more code worth annotating precisely.

### One possible solution

There is no single correct design for this project — what follows is one reasonable structure, not the only valid one. If your program satisfies the requirements and you understand every line of it, it's a success regardless of how different it looks from this.

````python
def load_expenses(filename: str) -> list[dict]:
    expenses = []
    try:
        with open(filename, "r") as file:
            for line in file:
                line = line.strip()
                if not line:
                    continue
                category, amount_text = line.split(",")
                expenses.append({"category": category, "amount": float(amount_text)})
    except FileNotFoundError:
        # No expenses recorded yet -- that's a normal starting state, not an error.
        pass
    return expenses


def save_expenses(filename: str, expenses: list[dict]) -> None:
    with open(filename, "w") as file:
        for expense in expenses:
            file.write(f"{expense['category']},{expense['amount']}\n")


def add_expense(expenses: list[dict]) -> None:
    category = input("Category: ").strip()
    while True:
        amount_text = input("Amount: ").strip()
        try:
            amount = float(amount_text)
            break
        except ValueError:
            print("That's not a valid amount -- try again.")
    expenses.append({"category": category, "amount": amount})
    print("Added.")


def print_all(expenses: list[dict]) -> None:
    if not expenses:
        print("No expenses recorded yet.")
        return
    for expense in expenses:
        print(f"{expense['category']}: {expense['amount']:.2f}")


def print_summary(expenses: list[dict]) -> None:
    if not expenses:
        print("No expenses recorded yet.")
        return

    total = 0.0
    by_category = {}

    for expense in expenses:
        total += expense["amount"]
        category = expense["category"]
        by_category[category] = by_category.get(category, 0.0) + expense["amount"]

    print(f"Total spent: {total:.2f}")
    for category, amount in by_category.items():
        print(f"  {category}: {amount:.2f}")


def run() -> None:
    filename = "expenses.txt"
    expenses = load_expenses(filename)

    while True:
        print()
        print("1. Add expense")
        print("2. View all expenses")
        print("3. View summary")
        print("4. Quit")
        choice = input("Choose an option: ").strip()

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            print_all(expenses)
        elif choice == "3":
            print_summary(expenses)
        elif choice == "4":
            save_expenses(filename, expenses)
            print("Saved. Goodbye.")
            break
        else:
            print("Not a valid option.")


# Uncomment to actually run the interactive program:
# run()
````

A few choices worth calling out, since a solution without commentary teaches less than one with it:

`load_expenses` catches `FileNotFoundError` specifically, rather than a bare `except:` — the same principle from Chapter 25: catch the specific failure you expect, so an unrelated bug elsewhere doesn't get silently swallowed along with it.

`add_expense` uses a `while True` loop around the amount input, looping until a valid number is actually given, rather than accepting one attempt and giving up. This is the same "repeat until valid" shape from the `while` loop chapter, just applied inside a function this time.

`by_category.get(category, 0.0)` reuses the safe-lookup-with-a-default idiom from the dictionaries chapter, avoiding a separate check for "have I seen this category before."

`expenses` is passed *into* `add_expense`, `print_all`, and `print_summary` as a parameter, rather than those functions reaching out to a variable defined outside them — which is exactly the scoping discipline Chapter 24 argued for: a function's effects should travel through what it's given and what it returns (or, since a list is mutable, what it modifies in place through a reference it was explicitly handed), not through quietly reading or writing something sitting in an outer scope.
