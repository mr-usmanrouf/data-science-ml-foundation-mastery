### Topic 1.1 — What Python Actually Is

#### Concept Overview
Python is technically just a language specification—a set of rules defining syntax and behavior. CPython is the reference implementation of this specification, written in C. When developers say "Python," they almost always mean CPython. Understanding this distinction clarifies why certain optimizations exist and how code executes.

When you execute a `.py` source file, CPython does not run it directly. First, a compiler translates the source code into bytecode, which is a lower-level representation. This bytecode is cached in `.pyc` files within the `__pycache__` directory to speed up future executions. The Python Virtual Machine (PVM), specifically the loop inside `ceval.c`, then interprets this bytecode instruction by instruction.

The execution model shifts depending on the environment. Running a `.py` file executes the script top-to-bottom and tears down the process. A REPL (Read-Eval-Print Loop) compiles and executes statements interactively. Jupyter notebooks extend the REPL model: each cell compiles and runs independently, but all cells share the same persistent namespace. Variables and functions defined in cell 1 remain resident in memory for cell 2.

#### The Mind Picture
```
[ Source Code (.py) ]
         |
    (Compiler)
         v
[ Bytecode (.pyc) ]
         |
    (PVM / ceval.c) ---> [ Output ]
         |
[ Object Memory / State ]
```

#### Syntax Anatomy
```python
import dis

def add(a, b):
    return a + b

# The dis module exposes the exact bytecode instructions generated by the compiler.
# LOAD_FAST pushes variables onto the evaluation stack.
# BINARY_OP pops them, adds them, and pushes the result.
# RETURN_VALUE pops the result and returns it to the caller.
```

#### Practical Examples
Here we inspect the bytecode generated by a simple function to see the compiler in action.


````python
import dis
def greet(name: str) -> str:
    return f"Hello, {name}"

print("Bytecode for greet:")
dis.dis(greet)
````

Next, we demonstrate that Jupyter's persistent namespace allows cross-cell memory access. We define a variable here.


````python
shared_namespace_variable: int = 42
print(f"Stored {shared_namespace_variable} in memory.")
````

And we retrieve it in the next cell, proving the namespace persists between independent executions.


````python
assert shared_namespace_variable == 42
print(f"Retrieved {shared_namespace_variable} successfully.")
````

#### Common Mistakes and Pitfalls
Assuming script state resets in Jupyter. Developers often modify variables in later cells, re-run earlier cells, and get unexpected results due to the persistent namespace.


````python
try:
    # Simulating out-of-order cell execution issues
    counter += 1
except NameError as e:
    print(f"Error caught: {e} - counter was not defined in a previous cell execution.")
    counter = 1
````

#### Best Practices
1. Restart your Jupyter kernel frequently to clear the persistent namespace and verify top-to-bottom execution.
2. Exclude `__pycache__` directories from version control using `.gitignore`.
3. Use type hints even in Jupyter cells to maintain clarity as state grows.
4. Keep cell logic small and modular to minimize namespace pollution.

#### Real-World Use Case
In heavy data science workflows, loading a 5GB dataset in cell 1 takes minutes. Jupyter's persistent namespace allows subsequent cells to process that dataset instantly without reloading it from disk.

#### Performance Notes
Bytecode interpretation in CPython (via `ceval.c`) introduces overhead compared to compiled languages like C or Rust. Performance critical sections are usually offloaded to C extensions (like NumPy) rather than executed purely in the PVM.

#### Quick Recap
* Python is a spec; CPython is the implementation.
* Source compiles to bytecode, cached in `.pyc` files.
* PVM (`ceval.c`) executes bytecode instructions.
* Scripts run linearly and terminate.
* Jupyter cells share a persistent memory namespace.
* `dis` module exposes bytecode instructions.

* **Mid**: How does Jupyter's execution model differ from running a standard Python script?

* **Senior**: Describe the role of `ceval.c` in CPython.
  * *Answer*: It implements the Python Virtual Machine evaluation loop. It fetches bytecode instructions one by one, manages the execution stack, handles object references, and delegates operations to underlying C functions.

### Topic 1.2 — Installing Python and Virtual Environments

#### Concept Overview
System Python is the Python installation owned and used by your operating system for system tools (like `apt` on Linux). Modifying it by installing user packages globally can break core OS functionalities. User installs provide a safer space, but true project isolation requires virtual environments.

A virtual environment is essentially an isolated directory structure containing a dedicated Python binary and its own `site-packages` directory. It prevents dependency conflicts between projects. When you run `python -m venv myenv`, CPython creates this directory.

Activating a virtual environment on Linux/Mac (`source myenv/bin/activate`) or Windows (`myenv\Scripts\activate`) prepends the virtual environment's binary directory to your shell's `PATH`. Running `pip install` inside an active environment places the package directly into its isolated `site-packages` folder. Running `sudo pip` bypasses this, polluting the system installation, which is a critical anti-pattern.

#### The Mind Picture
```
System PATH: /usr/bin/python -> [ OS System site-packages ]

Active Venv PATH: /myenv/bin/python -> [ /myenv/lib/python3.X/site-packages ]
                      |
                      v
             Fallbacks to system standard library
```

#### Syntax Anatomy
```bash
# Standard workflow for environment creation
python -m venv .venv       # Creates the directory structure
source .venv/bin/activate  # Alters $PATH to prioritize .venv/bin
pip install requests       # Installs into .venv/lib/.../site-packages
pip freeze > reqs.txt      # Dumps exact installed versions
```

### Topic 1.3 — Variables and the Object Model

#### Concept Overview
Variables in Python are not physical memory boxes where values are placed. They are name bindings—labels securely tied to objects living elsewhere in memory. When you assign a variable, you are pointing a label at an object. 

The `id()` function returns the physical memory address of the object in CPython. To manage memory, CPython uses reference counting. `sys.getrefcount()` reveals how many labels currently point to a specific object. When a reference count drops to zero, CPython immediately reclaims the memory.

Understanding the difference between identity and equality is mandatory. The `is` operator checks identity: do these two labels point to the exact same object in memory? The `==` operator checks equality: do these two objects contain the same value (evaluated via the `__eq__` method)? Because CPython optimizes memory by interning small integers (-5 to 256) and short strings, `is` might return True for equal small integers, but fail for larger ones. `None` is a strict singleton, so `is None` is the correct pattern.

#### The Mind Picture
```
Label 'a' ---\    
              v
           [ Object: int(256), refcount: 2 ]
              ^
Label 'b' ---/

Label 'c' ---> [ Object: int(1000), refcount: 1 ]
Label 'd' ---> [ Object: int(1000), refcount: 1 ]
```

#### Syntax Anatomy
```python
a = [1, 2, 3]
b = a         # b points to the SAME object as a
c = [1, 2, 3] # c points to a NEW object with the SAME value

a is b # True (identity)
a == c # True (equality)
a is c # False (identity)
```

#### Practical Examples
Variables are labels. Let's inspect object identity and reference counts.


````python
import sys

# Variables are labels pointing to objects
base_list: list[int] = [1, 2, 3]
alias_list: list[int] = base_list

print(f"base_list id: {id(base_list)}")
print(f"alias_list id: {id(alias_list)}")
print(f"Are they the same object? {base_list is alias_list}")
# getrefcount returns 1 higher than expected due to passing the variable to the function
print(f"Reference count: {sys.getrefcount(base_list) - 1}")
````

Let's demonstrate CPython's small integer interning. Numbers -5 to 256 are cached.


````python
small_x: int = 256
small_y: int = 256
print(f"256 is 256: {small_x is small_y}")

large_x: int = 257
large_y: int = 257
print(f"257 is 257: {large_x is large_y}") # Behavior depends on REPL vs script
````

Using `==` evaluates the `__eq__` method to check value equality, regardless of memory location.


````python
list_one: list[int] = [10, 20]
list_two: list[int] = [10, 20]
print(f"list_one == list_two: {list_one == list_two}")
print(f"list_one is list_two: {list_one is list_two}")
````

#### Common Mistakes and Pitfalls
Using `is` to compare values like integers or strings instead of `==`. Relying on `is` for integer comparison leads to brittle code that breaks unpredictably outside the -5 to 256 range.


````python
value: int = 1000
try:
    # Simulating a logic error (Modern Python may raise SyntaxWarning)
    assert value is 1000, "Identity check failed!"
except AssertionError:
    print("AssertionError: value is not the exact same object as the literal 1000 in memory.")
except SyntaxWarning:
    print("SyntaxWarning: 'is' with a literal.")
````

#### Best Practices
1. Use `is None` to check for None, as it is a guaranteed singleton.
2. Use `==` for evaluating if two objects have the same value.
3. Be aware that assigning mutable objects (like lists) to a new variable does not copy the object, it creates a new reference.
4. Avoid relying on CPython implementation details like small int interning for application logic.

#### Real-World Use Case
Object identity is crucial in caching and memoization systems (like `functools.lru_cache`). Using identity checks instead of deep equality checks on large structures saves significant CPU cycles when verifying if an input object has changed.

#### Performance Notes
Reference counting has overhead because every assignment and reassignment updates integers in memory, potentially blocking CPU caches. However, it provides deterministic memory management. The `is` operator is an O(1) pointer comparison and extremely fast compared to `==`, which might trigger expensive recursive comparisons.

#### Quick Recap
* Variables are labels pointing to objects.
* `id()` provides the memory address.
* Objects are tracked via reference counting.
* `is` checks memory identity; `==` checks value equality.
* CPython interns small integers and short strings.

#### Interview Questions
* **Junior**: What is the difference between `is` and `==`?
  * *Answer*: `is` checks if two variables point to the exact same memory address. `==` checks if the objects contain the same value by calling `__eq__`.
* **Mid**: Why does `a = 256; b = 256; a is b` return True, but using 257 returns False?
  * *Answer*: CPython pre-allocates and caches integers from -5 to 256 as an optimization. Both variables point to the same cached object. 257 triggers a new object allocation.
* **Senior**: How does Python clean up memory when an object is no longer needed?
  * *Answer*: Python uses reference counting. When a variable goes out of scope or is reassigned, the refcount decrements. When it hits zero, the memory is deallocated. A secondary garbage collector handles cyclic references.

### Topic 1.4 — Core Data Types

#### Concept Overview
Python's core data types abstract away complex memory management. `int` in Python 3 handles arbitrary precision arithmetic natively. It will automatically scales up memory usage to prevent overflow, meaning limits are bound only by system RAM. `float` implements the IEEE 754 double-precision 64-bit standard, meaning it shares the same precision constraints and floating-point math quirks found in C.

`complex` numbers are a native type represented with a `j` suffix, making Python inherently suited for scientific computing. `bool` is simply a subclass of `int` where `True` evaluates to 1 and `False` to 0. `str` represents immutable sequences of Unicode codepoints, providing robust internationalization support out of the box. `NoneType` is the type of the `None` object, serving as the universal null indicator.

Every core type is an object backed by specific C structures in CPython, containing a reference count, a type pointer, and the actual data payload.

#### The Mind Picture
```
[ int object ]         [ str object ]         [ float object ]
- ob_refcnt: 1         - ob_refcnt: 3         - ob_refcnt: 1
- ob_type: &PyLong     - ob_type: &PyUnicode  - ob_type: &PyFloat
- ob_size: N           - length: 5            - value: 3.14159
- ob_digit: [...]      - data: "Hello"
```

#### Syntax Anatomy
```python
x: int = 10**100      # Arbitrary precision
y: float = 3.14       # IEEE 754 64-bit
z: complex = 2 + 3j   # Native complex support
b: bool = True        # Subclass of int
s: str = "Unicode✓"   # Immutable text
n: type(None) = None  # The sole instance of NoneType
```

#### Practical Examples
We demonstrate arbitrary precision integers natively handling huge numbers.

Booleans are integers under the hood, allowing mathematical operations.

#### Common Mistakes and Pitfalls
Comparing floating-point numbers with `==` instead of `math.isclose()`. Modifying strings as if they were mutable arrays, leading to excessive allocations.


````python
import math
try:
    assert 0.1 + 0.2 == 0.3, "Math is broken!"
except AssertionError:
    print("AssertionError: Float comparison failed. Use math.isclose()")
    print(f"math.isclose result: {math.isclose(0.1 + 0.2, 0.3)}")
````

#### Quick Recap
* `int` grows to fill memory to avoid overflow.
* `float` uses IEEE 754 double precision.
* `complex` handles imaginary numbers.
* `bool` subclasses `int` (`True` is 1).
* `str` is an immutable Unicode sequence.
* Types are objects with heavy C structures behind them.

# Topic 1.5 — Mutable vs Immutable Types

#### Concept Overview
In Python, every value is an object, and every object has a core characteristic: whether its state can change after creation. Immutable types (like integers, floats, booleans, strings, tuples, and frozensets) represent values that are fixed in memory. When you attempt to change an immutable object, Python actually creates a completely new object in memory and rebinds the variable name to this new object. Mutable types (like lists, dictionaries, sets, and bytearrays), on the other hand, can have their internal state altered in-place without changing their memory address. This distinction is not merely semantic; it dictates how objects are passed to functions, how memory is managed, and whether an object can be used as a dictionary key or placed in a set (which requires hashability, a property intrinsically linked to immutability).

#### The Mind Picture
```text
Variables are just labels (pointers) attached to objects in memory.

      Immutable Reassignment (a = 1, then a = 2):
      -----------------------------------------
      Label 'a' -----> [Object: int(1)]
      
      After a = 2:
      Label 'a' -/---> [Object: int(1)] (ready for Garbage Collection if refcount == 0)
                 \---> [Object: int(2)] (new object created)

      Mutable In-Place Update (lst = [1], lst.append(2)):
      -----------------------------------------
      Label 'lst' ---> [Object: list (id: 0x123)] -> [ 1 ]
      
      After lst.append(2):
      Label 'lst' ---> [Object: list (id: 0x123)] -> [ 1, 2 ]
      (The list object identity remains the same)
```

#### Syntax Anatomy
```python
# The built-in id() function returns the memory address (identity) of an object.
a = 10           # Creates an immutable int object
print(id(a))     # Returns something like 140732890532432

a = a + 1        # Creates a NEW int object (11) and points 'a' to it
print(id(a))     # Returns a different memory address

lst = [10]       # Creates a mutable list object
print(id(lst))   # Returns something like 140234567812304

lst.append(11)   # Modifies the list in-place
print(id(lst))   # Returns the EXACT SAME memory address
```


````python
a = 42
initial_id = id(a)
a = a + 1
new_id = id(a)
print(f"Immutable integer changed identity: {initial_id != new_id}")
````

#### Practical Examples


````python
def demonstrate_list_mutation():
    original = [1, 2, 3]
    reference = original
    
    original.append(4)
    # Both names point to the same modified list
    print(f"Original: {original}")
    print(f"Reference: {reference}")
    print(f"Same identity? {original is reference}")

demonstrate_list_mutation()
````

Now let us look at the tuple binding interaction.


````python
def tuple_with_mutable_contents():
    # The tuple is immutable, meaning its slots cannot point to different objects.
    # However, the object it points to can still mutate itself.
    mixed_tuple = (1, [2, 3])
    print(f"Before: {mixed_tuple}")
    
    mixed_tuple[1].append(4)
    print(f"After: {mixed_tuple}")
    
    try:
        # We cannot rebind the tuple's slots
        mixed_tuple[1] = [2, 3, 4]
    except TypeError as e:
        print(f"TypeError caught: {e}")

tuple_with_mutable_contents()
````

#### Common Mistakes and Pitfalls


````python
# BAD: The Default Mutable Argument Trap
# The list is evaluated ONCE when the function is defined, not every time it is called.
def bad_append(val: int, target: list = []) -> list:
    target.append(val)
    return target

print(bad_append(1))
print(bad_append(2)) # Returns [1, 2] instead of [2]!

# FIX: Use None as a sentinel value
def good_append(val: int, target: list | None = None) -> list:
    if target is None:
        target = []
    target.append(val)
    return target

print(good_append(1))
print(good_append(2)) # Returns [2] correctly
````

#### Best Practices
1. Never use mutable objects as default arguments in function definitions; use `None`.
2. When you need a constant collection, use a `tuple` or `frozenset` instead of a `list` or `set`.
3. If you want to use custom objects as dictionary keys, implement `__hash__` and `__eq__`, and ensure the fields used in hashing are immutable.
4. When passing a mutable object to a function that might alter it, consider passing a copy (`lst.copy()`) if you need to preserve the original state in the caller.

#### Real-World Use Case
In caching libraries (like `functools.lru_cache`), function arguments are used as keys in a hidden dictionary to store results. If you pass a list to a cached function, it will throw a `TypeError: unhashable type: 'list'`. You must convert the list to a tuple before calling the cached function, ensuring the cache key is immutable and hashable.

#### Performance Notes
Creating new objects takes time and memory. String concatenation in a loop (`s += "x"`) is an O(N^2) operation because it allocates a new string and copies data at every iteration (though CPython has some specific optimizations for this, it is not guaranteed). Appending to a list (`lst.append("x")`) is amortized O(1). Therefore, always collect string fragments in a list and use `"".join(lst)`.

#### Quick Recap
* Variables are labels pointing to objects, not memory containers themselves.
* Immutable types (`int`, `str`, `tuple`) cannot change state after creation.
* Reassigning an immutable variable creates a new object and updates the label.
* Mutable types (`list`, `dict`) allow in-place modification.
* Tuples are immutable but can contain mutable elements whose internal state can change.
* Hashability requires immutability; lists and dicts cannot be dict keys or set elements.

#### Interview Questions
1. **Junior**: Explain why `a = "hello"; a[0] = "H"` throws an error but `b = ["hello"]; b[0] = "H"` does not.
   *Answer*: Strings are immutable in Python; their contents cannot be changed after creation. The assignment `a[0] = "H"` attempts to mutate the string in-place, causing a `TypeError`. Lists are mutable, so `b[0] = "H"` simply replaces the reference at index 0 of the list with a new string.
2. **Mid**: What is the default mutable argument trap, and how do you fix it?
   *Answer*: If you write `def func(lst=[])`, the list object `[]` is created exactly once when the module loads and the function is defined. Every call to `func()` without a second argument uses the exact same list instance. The fix is to write `def func(lst=None)` and then inside the function check `if lst is None: lst = []`.
3. **Senior**: Can a tuple be used as a dictionary key? Does it depend on its contents?
   *Answer*: A tuple can only be used as a dictionary key if all its elements are hashable (which usually means immutable). A tuple containing integers `(1, 2)` is fine. A tuple containing a list `(1, [2, 3])` will raise a `TypeError` when you try to hash it to insert into the dictionary, because the list inside is unhashable.

# Topic 1.6 — Type Conversion and Casting

#### Concept Overview
Type conversion is the process of changing an object from one data type to another. Python is a strongly typed language, meaning it does not perform implicit type coercion that might lose information or cause ambiguous behavior (like JavaScript adding numbers and strings). It will implicitly widen numeric types (e.g., adding an integer and a float produces a float via the `__add__` dunder method), but generally requires explicit casting. You perform explicit casting by calling the constructor of the target type, such as `int()`, `float()`, or `str()`.

#### The Mind Picture
```text
Conversion via Constructor calls:

"123" (str)  ---> int("123") ---> 123 (int)
123 (int)    ---> float(123) ---> 123.0 (float)
0 (int)      ---> bool(0)    ---> False (bool)

Failed Conversion:
"abc" (str)  ---> int("abc") ---> ValueError (cannot parse base 10)
```

#### Syntax Anatomy
```python
# Constructor Functions act as casting operators
string_num = "42"
integer_num = int(string_num)       # str to int
float_num = float(string_num)       # str to float
boolean_val = bool(integer_num)     # int to bool (0 is False, all else True)

# Base conversions
binary_string = "1010"
decimal_val = int(binary_string, 2) # Parses string as base-2

# Collection conversions
string_word = "hello"
char_list = list(string_word)       # ['h', 'e', 'l', 'l', 'o']
```


````python
def type_casting_demo():
    x: int = 10
    y: float = 2.5
    
    # Implicit widening
    result = x + y
    print(f"Type of {x} + {y} is {type(result)}: {result}")
    
    # Explicit conversion
    text: str = "100"
    number: int = int(text)
    print(f"Parsed {text!r} to integer {number}")

type_casting_demo()
````

Now let us look at some data structure conversions.


````python
def container_casting():
    # Deduplication using sets
    duplicates: list[int] = [1, 2, 2, 3, 3, 3]
    unique: set[int] = set(duplicates)
    print(f"List {duplicates} to Set {unique}")
    
    # Dict requires pairs
    pairs: tuple[tuple[str, int], ...] = (('a', 1), ('b', 2))
    mapping: dict[str, int] = dict(pairs)
    print(f"Pairs {pairs} to Dict {mapping}")

container_casting()
````

# Topic 1.7 — Operators

#### Concept Overview
Operators are special symbols in Python that carry out arithmetic or logical computation. Under the hood, operators are syntactic sugar for dunder (magic) methods on objects. For example, `a + b` translates to `a.__add__(b)`. This mechanism allows custom classes to define their own behavior for standard operators (operator overloading). Python supports several categories: arithmetic, comparison, logical, bitwise, assignment, identity, and membership operators, as well as the assignment expression (walrus) operator `:=`.

#### The Mind Picture
```text
The Operator Dispatch Mechanism:

a + b
 |
 v
a.__add__(b)
 |
 +--> If NotImplemented, tries b.__radd__(a)
       |
       +--> If still NotImplemented, raises TypeError
```

#### Syntax Anatomy
```python
# Arithmetic
res = 10 // 3   # Floor division (3)
res = 10 % 3    # Modulo (1)
res = 2 ** 3    # Exponentiation (8)

# Comparison
is_eq = (10 == 10.0) # True, checks value equality

# Logical (Short-circuiting)
# 'and' stops at first False, 'or' stops at first True
res = False and some_expensive_function() 

# Identity vs Membership
is_same_object = a is b    # Checks memory address equality
is_in_container = x in lst # Checks membership (calls lst.__contains__(x))

# Walrus (Python 3.8+)
if (n := len(lst)) > 10:
    print(f"List is too long: {n} elements")
```


````python
def arithmetic_and_bitwise():
    a: int = 10  # 1010 in binary
    b: int = 4   # 0100 in binary
    
    print(f"Floor Division: {a} // {b} = {a // b}")
    print(f"Bitwise AND: {a} & {b} = {a & b}")
    print(f"Bitwise OR: {a} | {b} = {a | b}")
    
    # Left shift multiplies by 2^n
    print(f"Left shift: {b} << 1 = {b << 1}")

arithmetic_and_bitwise()
````

Let's look at the walrus operator and logical operators.


````python
import re

def walrus_example(data: list[str]) -> list[str]:
    # Without walrus, we'd need to call re.match twice or use a loop.
    # With walrus, we assign and evaluate in the comprehension.
    pattern = re.compile(r'^user_(\d+)$')
    
    # Extracts the ID if the match is successful
    return [
        match.group(1) 
        for item in data 
        if (match := pattern.match(item)) is not None
    ]

results = walrus_example(["user_123", "admin", "user_456"])
print(f"Extracted IDs: {results}")
````

#### Common Mistakes and Pitfalls


````python
# BAD: Using 'is' for value comparison
a = 1000
b = 1000
# Might print False because 'is' checks memory identity, and large ints aren't interned.
print(f"BAD: a is b = {a is b}")

# FIX: Always use == for value equality
print(f"FIX: a == b = {a == b}")

# BAD: Chaining assignments modifying mutables
x = y = []
x.append(1)
# y is also modified because they point to the exact same list!
print(f"BAD: y is {y}")

# FIX: Initialize separately
x, y = [], []
x.append(1)
print(f"FIX: y is {y}")
````

#### Best Practices
1. Never use `is` to compare numbers or strings. Only use `is` to check against singletons like `None`, `True`, or `False` (e.g., `if val is None:`).
2. Utilize short-circuit evaluation to protect against errors: `if obj is not None and obj.value > 0:`.
3. Use the walrus operator `:=` sparingly, primarily in `while` loop conditions (reading chunks from a file) and list comprehensions to avoid duplicate work.
4. Avoid complex chained comparisons like `a < b > c` as they hurt readability, even though Python supports them (`a < b and b > c`).

#### Real-World Use Case
In parsing files chunk by chunk, the walrus operator cleans up loop structures significantly. 
Instead of:
```python
chunk = f.read(8192)
while chunk:
    process(chunk)
    chunk = f.read(8192)
```
You write:
```python
while chunk := f.read(8192):
    process(chunk)
```

#### Performance Notes
`in` checks on a list are O(N) because they scan the list linearly. `in` checks on a set or dict are O(1) due to hash lookups. If checking membership in a loop, always cast the list to a set beforehand.

#### Quick Recap
* Operators translate to dunder methods (`+` to `__add__`).
* `==` checks value equality; `is` checks memory identity.
* Logical operators `and`/`or` short-circuit and return the actual object evaluated, not necessarily a boolean.
* The walrus operator `:=` assigns and evaluates in a single expression.
* Bitwise operators work directly on binary representations of integers.

#### Interview Questions
1. **Junior**: What is the difference between `/` and `//` in Python?
   *Answer*: `/` always performs true division and returns a float. `//` performs floor division and returns an integer (or a float with no fractional part if operands are floats), rounding down towards negative infinity.
2. **Mid**: Explain how `a = a + b` differs from `a += b` for lists.
   *Answer*: `a = a + b` creates a new list by concatenating `a` and `b`, rebinding the label `a`. `a += b` calls `a.__iadd__(b)` (in-place add) on the list, which extends the existing list in memory (equivalent to `a.extend(b)`), mutating it without creating a new object.
3. **Senior**: How does Python evaluate `1 in [1, 2, 3] == True`?
   *Answer*: Python chains comparison operators. It evaluates as `(1 in [1, 2, 3]) and ([1, 2, 3] == True)`. The first part is `True`, but the second part is `False`, so the entire expression returns `False`. You must use parentheses `(1 in [1, 2, 3]) == True` to get the expected precedence.

# Topic 1.8 — Strings in Depth

#### Concept Overview
In Python 3, all strings are sequences of Unicode code points (characters), entirely distinct from raw bytes. They are immutable, meaning any string operation (replacement, uppercase, slicing) creates and returns a brand new string object. Because text manipulation is ubiquitous, the `str` type comes packed with a massive arsenal of highly optimized methods. Python also provides multiple syntax forms for string interpolation, with f-strings (formatted string literals, introduced in 3.6) being the most powerful, readable, and performant.

```text
+---+---+---+---+---+

s[1:4] returns a NEW string "ell"
```

#### Syntax Anatomy
```python

s2 = "Double quotes"
s3 = '''Multi
line'''

 literal (slashes ignored)"
```

````python
def string_manipulation():
    raw_data: str = "   user1, USER2, User3   \n"
    
    # Chaining methods (each step creates a new string)
    cleaned = raw_data.strip().lower()
    
    # Splitting into a list
    users: list[str] = [u.strip() for u in cleaned.split(',')]
    print(f"Parsed users: {users}")
    
    # Joining back together
    csv_line = ";".join(users)
    print(f"Re-joined: {csv_line}")

string_manipulation()
````

````python
def string_formatting():
    name: str = "Alice"
    score: float = 95.678
    
    # 1. f-strings (Modern, preferred)
    print(f"1. Name: {name}, Score: {score:.1f}")
    
    # f-string debugging feature (Python 3.8+)
    print(f"   Debug: {name=} {score=}")
    
    # 2. .format() method
    print("2. Name: {}, Score: {:.1f}".format(name, score))
    
    # 3. % formatting (Legacy)
    print("3. Name: %s, Score: %.1f" % (name, score))

string_formatting()
````

In data extraction tasks, strings often contain inconsistent whitespace. You combine `.strip()` to clean ends, `.split()` to tokenize, and `.join()` to reconstruct. For example, cleaning up dirty CSV lines: `clean_line = ",".join(field.strip() for field in line.split(","))`.

Still, `"".join(list_of_strings)` is universally O(N) and guaranteed to be performant. String interning (Python automatically caching small strings) makes identity comparisons (`is`) fast for identifiers, but this is an implementation detail you should not rely on for general text.

# Topic 1.10 — Control Flow

#### Concept Overview
Control flow dictates the execution path of a program based on runtime conditions. Python evaluates expressions in conditional statements based on their "truthiness." Python 3.10 introduced structural pattern matching (`match/case`), bringing functional-style destructuring and branching to the language, replacing long chains of `if/elif/else` for complex data shapes.

#### The Mind Picture
```text
          [ Execution Thread ]
                   |
             if condition:
            /             \
       [True]            [False]
         |                 |
    Execute Block A      elif condition2:
                           /           \
                       [True]         [False]
                         |              |
                    Execute Block B   else:
                                        |
                                   Execute Block C
```

#### Syntax Anatomy
```python
if condition_1:
    # block 1
elif condition_2:
    # block 2
else:
    # fallback block

# Ternary
result = value_if_true if condition else value_if_false
```

#### Practical Examples
Evaluating truthiness and basic branching.

````python
def process_items(items: list[str] | None) -> None:
    # None and empty lists evaluate to False
    if not items:
        print("No items to process.")
        return
    
    print(f"Processing {len(items)} items.")

process_items([])
process_items(["apple"])
````

Using the inline ternary operator for concise assignment.

````python
def get_status(code: int) -> str:
    return "Success" if code == 200 else "Error"

print(get_status(200))
print(get_status(404))
````

Structural pattern matching (Python 3.10+) for complex data dispatch.

````python
def parse_command(command: str | tuple) -> str:
    match command:
        case "quit" | "exit":
            return "Exiting program."
        case ("move", int(x), int(y)):
            return f"Moving to coordinates {x}, {y}"
        case _:  # Default fallback
            return "Unknown command."

print(parse_command(("move", 10, 20)))
print(parse_command("jump"))
````

#### Common Mistakes and Pitfalls
**Using equality instead of identity for `None`:**
```python
if x == None:  # Bad practice, can be overridden by __eq__
    pass
```
*Fix:* Always use `if x is None:`. Identity checks the memory address directly.

**Chaining complex logic without parentheses:**
```python

    pass
```
*Fix:* Use parentheses to explicitly group logic: `if (a and b) or (c and not d):`.

#### Best Practices
1. Rely on truthiness for empty collections (`if not my_list:` instead of `if len(my_list) == 0:`).
2. Return early from functions to avoid deep nesting (Guard Clauses).
3. Use `match/case` for destructuring dictionaries or objects, but stick to `if/elif` for simple boolean checks.
4. Keep ternary expressions simple; do not nest them.

#### Real-World Use Case
Parsing varying JSON shapes from a third-party API. `match/case` allows declaring the expected dictionary structures and binding variables directly if the shape matches, vastly cleaning up validation logic.

#### Performance Notes
Python evaluates boolean expressions sequentially and short-circuits. In `if A and B:`, if `A` is false, `B` is never executed. Place the cheapest or most likely failing checks first in chained conditions. `match/case` is optimized at the bytecode level to jump directly to matching shapes, making it slightly faster than long `elif` chains for certain types checks.

#### Quick Recap
- Falsy values: `0`, `0.0`, `""`, `[]`, `{}`, `set()`, `None`, `False`.
- Everything else evaluates to `True`.
- Ternary syntax: `X if condition else Y`.
- `is` checks memory identity; `==` checks value equality.
- `match/case` matches structural shapes and binds variables simultaneously.

#### Interview Questions
**Junior**: How does Python decide if a custom object is True or False?
*Answer*: It first looks for the `__bool__()` method. If absent, it checks `__len__()` (non-zero is True). If neither exists, the object is inherently True.

**Mid**: Why is `if x is None:` preferred over `if x == None:`?
*Answer*: `is` checks pointer identity, which is extremely fast and guarantees correctness. `==` invokes the object's `__eq__` method, which might be overridden to return incorrect or expensive results (e.g., pandas DataFrames raise exceptions on `== None`).

**Senior**: Explain how short-circuiting applies to the `or` operator and how it can be used for variable assignment.
*Answer*: `A or B` evaluates and returns `A` if `A` is truthy, skipping `B`. If `A` is falsy, it evaluates and returns `B`. This allows default assignments like `config = user_config or default_config`, picking the first truthy value.

# Topic 1.11 — Loops

#### Concept Overview
Loops execute a block of code repeatedly. Python provides `while` loops for condition-based iteration and `for` loops for iterating over collections (iterables). Python's `for` loop is fundamentally a "foreach" construct—it requests an iterator from an object and calls `next()` until a `StopIteration` exception is raised.

#### The Mind Picture
```text
for item in [1, 2, 3]:
         |
  calls iter([1, 2, 3]) -> returns Iterator
         |
+-----> calls next(Iterator)
|        | -> returns 1
|        |
|      calls next(Iterator)
|        | -> returns 2
|        |
|      calls next(Iterator)
|        | -> raises StopIteration
+--------+ (Loop terminates)
```

#### Syntax Anatomy
```python
for element in iterable:
    if element == target:
        break    # Exit loop entirely
    if skip_condition:
        continue # Skip to next iteration
else:
    # Executes ONLY if the loop finishes without hitting a 'break'
    pass
```

#### Practical Examples
Using `enumerate` and `zip` to loop intelligently without managing index counters.

````python
names: list[str] = ["Alice", "Bob", "Charlie"]
scores: list[int] = [85, 92, 78]

# zip pairs up iterables; enumerate provides an index
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"Rank {idx}: {name} with {score} points")
````

The `while` loop with `break` and the rare `else` clause.

````python
def find_prime(limit: int) -> int | None:
    n = 14
    while n < limit:
        if n % 2 != 0 and n % 3 != 0:
            print(f"Found pseudo-prime: {n}")
            break
        n += 1
    else:
        # Runs if the while condition becomes false (no break was hit)
        print("No pseudo-prime found in range.")
        return None
    return n

find_prime(20)
````

#### Common Mistakes and Pitfalls
**Modifying a list while iterating over it:**
```python
data = [1, 2, 3, 4]
for item in data:
    if item % 2 == 0:
        data.remove(item) # Causes elements to be skipped
```
*Fix:* Iterate over a copy (`for item in data[:]`) or use a list comprehension to create a new list.

**Using `range(len())` instead of iteration:**
```python
for i in range(len(my_list)):
    print(my_list[i])
```
*Fix:* Just use `for item in my_list:`. If you need the index, use `for i, item in enumerate(my_list):`.

#### Best Practices
1. Never modify a collection while iterating over it directly.
2. Use `enumerate()` instead of maintaining a manual integer counter.
3. Use `zip()` to iterate through multiple sequences simultaneously; it stops when the shortest iterable is exhausted.
4. The `else` block on a loop is useful for "search" loops to handle the "not found" case elegantly without flags.

#### Real-World Use Case
Polling an external service for a status update. A `while` loop checks the API every few seconds. If the status hits "COMPLETE", it `break`s. The `else` clause fires if the maximum number of attempts is reached, throwing a TimeoutError.

#### Performance Notes
Iterating in Python has overhead due to the dynamic type dispatch of `__iter__` and `__next__`. Nested loops multiply this complexity (O(N^2)). For heavy numerical iterations, use vectorization via NumPy or list comprehensions, which execute loop logic in optimized C code rather than Python bytecode.

#### Quick Recap
- `for` iterates over items of an iterable directly.
- `range(start, stop, step)` generates integers up to (but not including) `stop`.
- `break` forcefully exits a loop; `continue` jumps to the next evaluation.
- `else` on a loop executes if the loop terminates normally (without a `break`).
- `zip` combines iterables pairwise.

#### Interview Questions
**Junior**: What does the `pass` statement do inside a loop?
*Answer*: It is a null operation. It does nothing and is used as a placeholder when a statement is syntactically required but no action is needed.

**Mid**: Under what circumstances does the `else` block of a `for` loop execute?
*Answer*: The `else` block executes when the iterable is completely exhausted. It is skipped if the loop is terminated early via a `break` statement.

**Senior**: How does a Python `for` loop actually work under the hood with respect to magic methods?
*Answer*: It first calls `iter()` on the object, which invokes `__iter__()` returning an iterator object. The loop then repeatedly calls `next()` on this iterator, which invokes `__next__()`, binding the returned value to the loop variable. When `__next__()` raises the `StopIteration` exception, the loop catches it silently and terminates.

# Topic 1.12 — Lists

#### Concept Overview
The Python `list` is a mutable, ordered sequence of elements. Under the hood, it is not a linked list, but a dynamic array of pointers to Python objects. It handles resizing automatically via over-allocation. Because it stores pointers, a single list can hold mixed data types, though homogeneous data is preferred for sanity.

#### The Mind Picture
```text
List Object (my_list)
+---------+--------+--------+
| Size: 3 | Cap: 4 | Array* | 
+---------+--------+--------+
                       |
   [Pointer Array] <---+
   [ 0 | 1 | 2 | _ ]
     |   |   |   \___ (Extra capacity allocated for O(1) appends)
     v   v   v
  "A"   42   True    (Actual objects scattered in memory)
```

#### Syntax Anatomy
```python
lst = [1, 2, 3]                  # Literal creation
lst[start:stop:step]             # Slicing syntax
[expr for item in iter if cond]  # List comprehension
```
Core Methods:
- `append(x)`: Adds item to end.
- `extend(iterable)`: Appends elements from iterable.
- `insert(i, x)`: Inserts item at index `i`.
- `pop(i)`: Removes and returns item at index `i` (default -1).
- `sort(key=..., reverse=...)`: In-place sorting.

#### Practical Examples
Slicing allows powerful substring/sublist extraction without mutating the original.

````python
data: list[int] = [10, 20, 30, 40, 50, 60]

print(data[1:4])     # [20, 30, 40]
print(data[::-1])    # Reversed copy: [60, 50, 40, 30, 20, 10]
print(data[-2:])     # Last two elements: [50, 60]
````

Sorting using the `key` parameter to apply custom logic.

````python
words: list[str] = ["apple", "banana", "cherry", "date"]

# Sort by length of string, descending
words.sort(key=len, reverse=True)
print(words)  # ['banana', 'cherry', 'apple', 'date']
````

List comprehensions for concise, optimized mapping and filtering.

````python
matrix: list[list[int]] = [[1, 2], [3, 4], [5, 6]]

# Flatten matrix, only keeping even numbers
evens: list[int] = [num for row in matrix for num in row if num % 2 == 0]
print(evens)  # [2, 4, 6]
````

#### Common Mistakes and Pitfalls
**Mutable Default Arguments:**
```python
def add_item(val: int, cache: list[int] = []):
    cache.append(val)
    return cache
print(add_item(1)) # [1]
print(add_item(2)) # [1, 2] - The list persists across calls!
```
*Fix:* Use `None` as default.
```python
def add_item(val: int, cache: list[int] | None = None):
    if cache is None: cache = []
    cache.append(val)
```

**Removing items while iterating:** As covered in Loops, mutating list length during iteration breaks internal indexing.

#### Best Practices
1. Prefer List Comprehensions over initializing an empty list and `append`ing in a `for` loop.
2. Use `list.extend(other_list)` rather than `for item in other_list: list.append(item)`.
3. Use `list[:]` or `list.copy()` for shallow copies when passing lists to functions that might mutate them.
4. Type hint lists with `list[Type]` (Python 3.9+) to catch heterogeneous data bugs early.

#### Real-World Use Case
Batching operations. A system collects incoming metrics into a list. When `len(metrics) >= 100`, it sends the batch to a database, then calls `metrics.clear()` to reset the buffer for the next batch.

#### Performance Notes
- `append()` and `pop()` (from the end) are O(1) amortized.
- `insert(0, x)` and `pop(0)` are O(N) because every subsequent element must be shifted in memory. If you need queue-like behavior, use `collections.deque`.
- Python's `sort()` uses Timsort, an O(N log N) algorithm highly optimized for partially sorted data.

#### Quick Recap
- Lists are dynamic arrays of pointers, not linked lists.
- Negative indexing starts at `-1` (the last element).
- Slicing `[start:stop:step]` returns a *new* list object.
- Modifying operations like `sort()` and `reverse()` happen in-place and return `None`.
- List comprehensions combine map and filter operations cleanly and execute at C speed.

#### Interview Questions
**Junior**: How do you reverse a list without modifying the original?
*Answer*: By using slicing: `new_list = original[::-1]`. Calling `original.reverse()` mutates it in place.

**Mid**: Explain why inserting an element at the beginning of a list is slower than appending it to the end.
*Answer*: Lists are arrays in contiguous memory. Appending simply places a value in pre-allocated empty space at the end. Inserting at index 0 requires Python to shift every existing pointer one slot to the right, which is an O(N) operation.

**Senior**: Write a nested list comprehension to transpose a 2D matrix.
*Answer*: `transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]`.

### Topic 1.13 — Tuples

#### Concept Overview
Tuples are immutable sequences in Python, designed to hold heterogeneous collections of data or fixed-size structures where the position of the element has semantic meaning. Python added tuples to provide a lightweight, immutable alternative to lists. By preventing modifications after creation, tuples offer stability for data that should remain constant throughout a program's execution, acting as reliable records.

Tuples solve the problem of data integrity when passing collections across boundaries. Because they cannot be altered (no appending, removing, or reassigning elements), you can safely pass a tuple to a function knowing the original data remains pristine. They also serve as hashable sequences, enabling them to be used as keys in dictionaries or elements in sets, which lists cannot do.

#### The Mind Picture
```text
+-------------------+      +-------------------+
| List (Mutable)    |      | Tuple (Immutable) |
+-------------------+      +-------------------+
| ptr -> [0] -> int |      | ptr -> [0] -> int |
| ptr -> [1] -> str |      | ptr -> [1] -> str |
| ptr -> [2] -> ... |      +-------------------+
| capacity: 8       |      | fixed length: 2   |
+-------------------+      +-------------------+
```
A list is a dynamic array with over-allocated capacity to accommodate future growth. A tuple is a fixed-size struct containing exact pointers to its elements. There is no extra capacity. Once memory is allocated for a tuple, its shape is frozen. 

#### Syntax Anatomy
```python
# 1. Empty tuple
empty = ()

# 2. Single-element tuple (requires trailing comma)
single = (42,)     # Without comma, (42) is just an integer in parentheses

# 3. Standard tuple creation (parentheses are optional but recommended)
coords = 10, 20    # Tuple packing
x, y = coords      # Tuple unpacking

# 4. Extended unpacking (Python 3+)
first, *rest = (1, 2, 3, 4)
```

#### Practical Examples
Tuple packing and basic unpacking:


````python
from typing import Tuple

def get_db_config() -> Tuple[str, int]:
    host = "localhost"
    port = 5432
    return host, port

config = get_db_config()
host, port = config
print(f"Host: {host}, Port: {port}")
````

Extended unpacking in action:


````python
record: tuple[str, str, int, int, int] = ("jane_doe", "jane@example.com", 95, 87, 92)
username, email, *scores = record
print(f"User: {username}")
print(f"Scores: {scores}")  # scores is a list: [95, 87, 92]
````

Swapping variables without a temporary variable:


````python
a: int = 5
b: int = 10
a, b = b, a
print(f"a={a}, b={b}")
````

Using tuples as dictionary keys:


````python
from typing import Dict, Tuple

# Mapping (x, y) coordinates to terrain types
terrain_map: Dict[Tuple[int, int], str] = {
    (0, 0): "grass",
    (1, 0): "water",
    (0, 1): "mountain"
}
print(terrain_map[(0, 1)])
````

#### Common Mistakes and Pitfalls
Missing the trailing comma for single-element tuples. Devs write `(5)` expecting a tuple, but get an integer, causing iteration errors later.



````python
# BAD
config = ("prod")
# type(config) is str, iterating over it iterates over characters

# GOOD
config = ("prod",)
# type(config) is tuple
````

Assuming tuple immutability applies to its elements. A tuple is immutable, but if it contains a mutable object (like a list), that object can still be changed.


````python
# BAD
fixed_data = (1, 2, [3, 4])
fixed_data[2].append(5)  # Modifies the list inside the tuple!
print(fixed_data)  # Outputs: (1, 2, [3, 4, 5])
````

#### Best Practices
1. Use tuples for heterogeneous data (different types) and lists for homogeneous data (same type).
2. Always include parentheses for clarity, even when tuple packing allows omitting them.
3. Use namedtuples (`collections.namedtuple` or `typing.NamedTuple`) when a tuple has many fields and accessing by index becomes unreadable.
4. Return tuples from functions when returning multiple related values.
5. Employ extended unpacking `*rest` to clearly signal which elements of a sequence you care about.

#### Real-World Use Case
In Django, URL routing configurations rely on tuples. The `urlpatterns` list often contains tuples defining routes. Django uses tuples here because routing definitions should remain strictly unmodified at runtime.
```python
from django.urls import path
from . import views

urlpatterns = (
    path('index/', views.index, name='main-view'),
    path('bio/<username>/', views.bio, name='bio'),
)
```

#### Performance Notes
Tuples are fundamentally faster and more memory-efficient than lists. CPython aggressively optimizes small tuples (sizes 1-20) using a free list, meaning memory allocation for new small tuples is practically instantaneous. Because tuples lack dynamic resizing overhead (no overallocation), they consume less RAM. Memory complexity: `O(N)`. Access time: `O(1)`.

#### Quick Recap
*   Tuples are immutable; they cannot be altered after creation.
*   A single-element tuple strictly requires a trailing comma: `(1,)`.
*   Tuples automatically pack and unpack: `a, b = 1, 2`.
*   Extended unpacking uses `*`: `first, *rest = items`.
*   Variable swapping `a, b = b, a` uses tuple packing/unpacking under the hood.
*   Tuples containing only hashable elements can be used as dict keys.
*   CPython caches small tuples, making them faster to allocate than lists.

#### Interview Questions
1. **Junior**: How do you create a tuple with exactly one element?
   *Answer*: You must include a trailing comma after the element, like `my_tuple = (42,)`. Without the comma, Python evaluates the expression inside parentheses as a scalar value.
2. **Mid**: Explain why a tuple can be used as a dictionary key but a list cannot.
   *Answer*: Dictionary keys must be hashable. Tuples are immutable and provide a `__hash__` method (as long as all their elements are hashable). Lists are mutable, so they lack a `__hash__` method because their contents can change, which would break the dictionary's internal hash table structure.
3. **Senior**: What happens if you modify a list that is an element of a tuple? Is the tuple still immutable?
   *Answer*: Yes, the tuple is still immutable. The tuple stores a reference to the list object. Modifying the list (e.g., appending an item) changes the list's internal state, but the reference stored in the tuple remains exactly the same. The tuple itself has not been changed.

### Topic 1.14 — Dictionaries

#### Concept Overview
Dictionaries are highly optimized key-value mappings in Python. They associate unique, hashable keys with arbitrary values. Python introduced dictionaries to provide constant-time data retrieval via hashing, making them the standard tool for fast lookups.

They solve the problem of needing to find data based on a specific identifier rather than a numeric index. Every namespace, class definition, and object instance in Python is fundamentally backed by a dictionary.

#### The Mind Picture
```text
Key -> Hash Function -> Index -> [ Hash | Key | Value ] Array
                                 +--------------------+
"name" -> hash() -> 3 ---------> | h1   | "name" | "Alice" |
                                 | h2   | "age"  | 30      |
                                 +--------------------+
```
A dict uses a sparse array of indices pointing to a dense array of entries. This memory layout, introduced in CPython 3.6, guarantees insertion order and uses less memory than older hash table designs.

#### Syntax Anatomy
```python
# 1. Literal creation
user = {"id": 1, "role": "admin"}

# 2. dict() constructor
config = dict(host="localhost", port=8080)

# 3. Comprehension
squares = {x: x**2 for x in range(5)}

# 4. Merging (Python 3.9+)
merged = dict1 | dict2
```

#### Practical Examples
Creating and accessing dictionaries:


````python
from typing import Dict, Any

user: Dict[str, Any] = {
    "username": "admin",
    "active": True
}
# Safe access with get() prevents KeyError
login_count = user.get("logins", 0)
print(f"Logins: {login_count}")
````

Iterating over views:


````python
inventory: Dict[str, int] = {"apples": 50, "oranges": 20}

# views reflect changes instantly
keys_view = inventory.keys()
inventory["bananas"] = 30

for key, val in inventory.items():
    print(f"{key}: {val}")
````

Using `setdefault` and `pop`:


````python
stats: Dict[str, int] = {"visits": 100}

# Set default if key missing, returns the value
bounces = stats.setdefault("bounces", 0)
print(f"Stats: {stats}")

# Remove and return
visits = stats.pop("visits")
print(f"Visits removed: {visits}")
````

Dictionary comprehensions and merging:


````python
names = ["alice", "bob", "charlie"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)

defaults = {"theme": "light", "retries": 3}
user_prefs = {"theme": "dark"}

# Python 3.9+ merge operator
final_config = defaults | user_prefs
print(final_config)  # {"theme": "dark", "retries": 3}
````

#### Common Mistakes and Pitfalls
Using mutable types as keys. Keys must be hashable.



````python
# BAD
# invalid_dict = {[1, 2]: "value"} # Raises TypeError: unhashable type: 'list'

# GOOD
valid_dict = {(1, 2): "value"}
````

Modifying a dictionary while iterating over its keys. This raises a `RuntimeError`.



````python
# BAD
data = {"a": 1, "b": 2}
# for k in data.keys():
#     if k == "a": del data[k] # RuntimeError

# GOOD
for k in list(data.keys()):
    if k == "a":
        del data[k]
````

#### Best Practices
1. Use `dict.get(key, default)` when a key might be absent, rather than a `try/except KeyError` block for simple cases.
2. Rely on insertion order guarantee (Python 3.7+) when building ordered data pipelines.
3. Use the merge operator `|` (Python 3.9+) instead of `dict.update()` when you need to create a new dictionary without mutating the original.
4. Dictionary comprehensions are strictly cleaner and faster than a `for` loop calling `.update()`.
5. For complex default structures, use `collections.defaultdict` instead of `dict.setdefault()`.

#### Real-World Use Case
JSON APIs heavily map to Python dictionaries. When parsing a payload from an HTTP request in frameworks like FastAPI, the request body is transformed into a dictionary for validation and processing.
```python
def process_webhook(payload: dict) -> None:
    event_type = payload.get("type", "unknown")
    event_data = payload.get("data", {})
    if event_type == "payment.succeeded":
        amount = event_data.get("amount", 0)
        # handle payment
```

#### Performance Notes
Dictionary lookups, insertions, and deletions have an average time complexity of `O(1)`. In worst-case scenarios involving massive hash collisions, complexity degrades to `O(N)`, but Python's hash randomization minimizes this risk. The CPython 3.6 compact dict implementation reduces memory overhead by 20-25% compared to older versions.

#### Quick Recap
*   Dictionaries store key-value pairs with `O(1)` average lookups.
*   Keys strictly require hashability (strings, integers, tuples).
*   Insertion order is explicitly guaranteed in Python 3.7+.
*   `.get()` safely retrieves values with an optional default.
*   `.items()`, `.keys()`, and `.values()` return live views, not static lists.
*   `|` merges dictionaries directly (Python 3.9+).

#### Interview Questions
1. **Junior**: How do you safely check if a key exists in a dictionary?
   *Answer*: Use the `in` operator: `if "key" in my_dict:`. You can also use `my_dict.get("key")` which returns `None` if the key is absent.
2. **Mid**: What does it mean that dictionary views are "live"?
   *Answer*: It means that objects returned by `.keys()`, `.values()`, and `.items()` dynamically reflect changes to the underlying dictionary. If you add a key to the dictionary, the view object immediately includes it, without needing to recreate the view.
3. **Senior**: Explain the difference between `.update()` and the `|` merge operator.
   *Answer*: `dict.update()` mutates the dictionary in place and returns `None`. The `|` operator creates and returns an entirely new dictionary containing the merged keys and values, leaving the original dictionaries completely untouched.

### Topic 1.15 — Sets and Frozensets

#### Concept Overview
Sets are unordered collections of unique, hashable elements. They are Python's implementation of mathematical sets. Python added sets to provide high-performance operations for membership testing, deduplication, and mathematical computations like intersections and unions.

Sets solve the problem of uniqueness and fast `O(1)` lookups. Searching for an item in a list takes `O(N)` time, whereas searching in a set is instantaneous regardless of size. `frozenset` is the immutable counterpart, allowing sets to be used as dictionary keys or nested inside other sets.

#### The Mind Picture
```text
Hash Function -> Index -> [ Hash | Element ] Array
                          +------------------+
"apple" -> hash() -> 2 -> | h1   | "apple"   |
"pear"  -> hash() -> 5 -> | h2   | "pear"    |
                          +------------------+
```
A set is essentially a dictionary with only keys and no values. It uses the same underlying hash table implementation to ensure rapid element access and enforce uniqueness.

#### Syntax Anatomy
```python
# 1. Literal creation
active_users = {"alice", "bob"}

# 2. Empty set (MUST use constructor, {} makes a dict)
empty_set = set()

# 3. Frozenset
frozen = frozenset(["x", "y"])

# 4. Set Operators
union = a | b
intersection = a & b
difference = a - b
sym_diff = a ^ b
```

#### Practical Examples
Deduplication and membership testing:


````python
from typing import List, Set

raw_ids: List[int] = [101, 102, 101, 103, 104, 102]
unique_ids: Set[int] = set(raw_ids)
print(f"Unique IDs: {unique_ids}")

# O(1) membership check
if 103 in unique_ids:
    print("User 103 found.")
````

Mathematical set operations:


````python
admins: Set[str] = {"alice", "bob", "charlie"}
devs: Set[str] = {"bob", "dave", "eve"}

# Union (all unique users)
all_staff = admins | devs
print(f"All staff: {all_staff}")

# Intersection (users who are both admins and devs)
admin_devs = admins & devs
print(f"Admin Devs: {admin_devs}")

# Difference (admins who are NOT devs)
pure_admins = admins - devs
print(f"Pure Admins: {pure_admins}")
````

Set modification methods:


````python
tags: Set[str] = {"python", "coding"}

tags.add("backend")
# discard does not error if element is missing
tags.discard("frontend") 

print(tags)
````

Using `frozenset`:


````python
from typing import FrozenSet, Dict

# frozenset can be a dictionary key
valid_transitions: Dict[FrozenSet[str], str] = {
    frozenset(["draft", "review"]): "published",
    frozenset(["review", "rejected"]): "draft"
}
state_pair = frozenset(["draft", "review"])
print(valid_transitions[state_pair])
````

#### Common Mistakes and Pitfalls
Creating an empty set using `{}`.



````python
# BAD
my_set = {}
# type(my_set) is dict!

# GOOD
my_set = set()
````

Using `remove` instead of `discard` when you aren't sure the element exists, leading to a `KeyError`.



````python
# BAD
data = {1, 2, 3}
# data.remove(4)  # Raises KeyError: 4

# GOOD
data.discard(4)   # Safely does nothing if 4 is absent
````

#### Best Practices
1. Instantiate empty sets via `set()`.
2. Prefer operators (`|`, `&`, `-`) over named methods (`.union()`, `.intersection()`) for readability, unless the input is an arbitrary iterable (operators require both sides to be sets).
3. Use sets strictly for deduplication before processing list data to avoid O(N^2) complexity patterns.
4. Choose `discard()` over `remove()` to avoid `try/except` boilerplate when strict absence checking isn't required.
5. Use `frozenset` when exposing an immutable set of constants at the module level.

#### Real-World Use Case
Role-based access control (RBAC). Checking if a user has permission to access an endpoint often uses set intersection.
```python
def has_access(user_roles: set, required_roles: set) -> bool:
    # Returns True if there is any overlap
    return not user_roles.isdisjoint(required_roles)
```

#### Performance Notes
Set operations are heavily optimized in C. Membership testing (`in`), adding, and removing are `O(1)`. Union is `O(len(s1) + len(s2))`. Intersection is `O(min(len(s1), len(s2)))`. Using sets for deduplication (`set(my_list)`) is `O(N)` and drastically faster than looping through a list and appending non-duplicates to a new list.

#### Quick Recap
*   Sets hold unique, hashable elements and offer `O(1)` membership lookups.
*   `{}` creates a dict; `set()` creates an empty set.
*   Operators `|`, `&`, `-`, `^` perform mathematical set logic.
*   `.discard()` quietly ignores missing elements; `.remove()` raises `KeyError`.
*   `frozenset` is the immutable, hashable variant of a set.

#### Interview Questions
1. **Junior**: How do you find the unique elements in a list of integers?
   *Answer*: Pass the list to the `set()` constructor: `unique_items = set(my_list)`.
2. **Mid**: What is the difference between `set.remove()` and `set.discard()`?
   *Answer*: Both remove an element from the set. However, if the element does not exist, `remove()` raises a `KeyError`, while `discard()` silently does nothing.
3. **Senior**: Can you use a list inside a set? Why or why not?
   *Answer*: No. Set elements must be hashable because sets use a hash table internally. Lists are mutable and therefore unhashable. If you need a sequence inside a set, use a tuple.

### Topic 1.16 — Functions

#### Concept Overview
Functions are first-class, reusable blocks of code that encapsulate logic, accept inputs, and return outputs. Python added them to support procedural programming, abstracting complex tasks into modular routines.

Functions solve the problem of code duplication and variable namespace pollution. By defining strict boundaries using the LEGB scope rule, functions ensure variables declared inside them do not overwrite variables outside. First-class status means functions can be passed as arguments, returned, or assigned to variables.

#### The Mind Picture

#### Common Mistakes and Pitfalls
The Mutable Default Argument Trap. Default arguments are evaluated strictly once, at function definition time.



````python
# BAD
def add_item(item: str, target: list = []) -> list:
    target.append(item)
    return target
# add_item("a") -> ["a"]
# add_item("b") -> ["a", "b"] (The list persists!)

# GOOD
def add_item_safe(item: str, target: list = None) -> list:
    if target is None:
        target = []
    target.append(item)
    return target
````

### Topic 1.17 — Exception Handling Basics

#### Concept Overview
Exception handling is Python's mechanism for dealing with runtime errors gracefully. Instead of crashing the program when something unexpected happens—like a missing file or a network timeout—Python raises an exception object. 

The `try/except` block lets you intercept these exceptions. The `else` clause executes only if no exception was raised in the `try` block. The `finally` clause always executes, regardless of whether an exception occurred, making it ideal for cleanup operations.

Python organizes exceptions in a hierarchy. `BaseException` is the root, but most standard errors inherit from `Exception`. You should almost always catch `Exception`, not `BaseException`, to avoid intercepting system-level events like `KeyboardInterrupt` or `SystemExit`.


#### The Mind Picture
```
Execution Flow in try/except/else/finally:

[ try block ]
      |
      +-- (Exception raised?) -- YES --> [ except block ] --+
      |                                                     |
      NO                                                    v
      |                                              (reraise?) --> [Exit]
      v                                                     |
[ else block ] <--------------------------------------------+
      |
      v
[ finally block ] (Always runs)
      |
      v
[ Program Continues ]
```


#### Syntax Anatomy
```python
try:
    # 1. Code that might fail
    risky_operation()
except ValueError as e:
    # 2. Catches specific exception, binds instance to 'e'
    handle_value_error(e)
except (TypeError, KeyError):
    # 3. Catches multiple exception types
    handle_other_errors()
except Exception:
    # 4. Catches any non-system exception
    handle_unknown_error()
else:
    # 5. Runs ONLY if try block completed without exceptions
    success_path()
finally:
    # 6. ALWAYS runs, even if returning or raising inside except
    cleanup_resources()
```


#### Practical Examples
Basic try/except to handle a conversion error.


````python
def parse_integer(value: str) -> int | None:
    try:
        return int(value)
    except ValueError:
        print(f"Cannot parse '{value}' as integer.")
        return None

print(parse_integer("42"))
print(parse_integer("forty-two"))
````

Handling multiple exception types with different fallback logic.


````python
def divide_elements(data: dict[str, int], key1: str, key2: str) -> float:
    try:
        result = data[key1] / data[key2]
    except KeyError as e:
        print(f"Missing key in data: {e}")
        return 0.0
    except ZeroDivisionError:
        print("Cannot divide by zero.")
        return 0.0
    else:
        print("Division successful.")
        return result
    finally:
        print("Division operation attempted.")

data = {"a": 10, "b": 2, "c": 0}
print(divide_elements(data, "a", "b"))
print(divide_elements(data, "a", "x"))
print(divide_elements(data, "a", "c"))
````

Using `raise` to re-raise an exception after logging it.


````python
def process_data(value: int) -> None:
    try:
        if value < 0:
            raise ValueError("Value cannot be negative.")
    except ValueError as e:
        print(f"Logging error: {e}")
        # Re-raise the exact same exception to be handled upstream
        raise

try:
    process_data(-5)
except ValueError:
    print("Caught re-raised exception upstream.")
````

#### Common Mistakes and Pitfalls
Mistake 1: The bare `except:` or catching `BaseException`. This prevents you from safely stopping a script with Ctrl+C.


````python
import time

# WRONG
def bad_loop() -> None:
    try:
        # Simulate work
        time.sleep(0.1)
        raise ValueError("Oops")
    except: # Catches KeyboardInterrupt too!
        print("Caught an error, continuing...")

# RIGHT
def good_loop() -> None:
    try:
        time.sleep(0.1)
        raise ValueError("Oops")
    except Exception: # Let KeyboardInterrupt pass through
        print("Caught an error, continuing...")
````

Mistake 2: Putting too much code in the `try` block. This masks bugs because you might catch an exception from a line you didn't intend to protect.


````python
# WRONG
def bad_fetch(data: dict[str, str]) -> str:
    try:
        val = data["key"]
        val = val.upper()
        # What if upper() fails? We catch it as KeyError? No, wait...
        return val
    except Exception as e:
        return "fallback"

# RIGHT
def good_fetch(data: dict[str, str]) -> str:
    try:
        val = data["key"]
    except KeyError:
        return "fallback"
    else:
        # Only runs if key was found
        return val.upper()
````

#### Best Practices
1. Be specific. Catch `ValueError`, not `Exception`, if you only expect a parsing failure.
2. Keep `try` blocks tiny. Only wrap the exact lines that raise the exception you are prepared to handle.
3. Use `else` for code that depends on the `try` succeeding. This clarifies intent and keeps the `try` block small.
4. Use `finally` for resource cleanup (closing files, releasing locks, closing connections).
5. Never silence exceptions blindly (`except Exception: pass`). Log them or handle them explicitly.

#### Real-World Use Case
In a web scraper or API client, you often deal with network timeouts. You want to retry a few times before giving up, and you need to log specific failures.


#### Performance Notes
Setting up a `try` block in Python 3.11+ is virtually zero-cost if no exception is raised (the "zero-cost exceptions" implementation). Actually raising an exception is still slow because it requires instantiating the exception object and unwinding the call stack. Do not use exceptions for standard control flow (like breaking out of a loop) in performance-critical code.

#### Quick Recap
- `try` executes code that might fail.
- `except` catches specific errors.
- `else` runs if no exception occurred.
- `finally` always runs, used for cleanup.
- Catch `Exception`, never catch `BaseException` unless writing a framework-level supervisor.
- Bare `except:` is a bug waiting to happen.
- `raise` without arguments re-raises the active exception.

#### Interview Questions
**Junior**: What is the difference between catching `Exception` and a bare `except:`?
**Answer**: A bare `except:` catches `BaseException`, which includes `SystemExit` and `KeyboardInterrupt`. Catching `Exception` ignores these, allowing the script to be gracefully terminated by the user (Ctrl+C).

**Mid**: When should you use the `else` clause in a try/except block?
**Answer**: When you have code that should only execute if the `try` block succeeded, but that code itself shouldn't be caught by the `except` block. It keeps the `try` block small and prevents accidental catching of unrelated exceptions.

**Senior**: Explain the cost of exceptions in Python 3.11+.
**Answer**: Python 3.11 introduced "zero-cost" exceptions. Entering a `try` block no longer pushes state onto a runtime stack. Instead, compiler-generated exception tables determine where to jump if an error occurs. Thus, `try` is free if no error happens, but raising an exception remains slow due to traceback generation.

### Topic 1.18 — File I/O

#### Concept Overview
File I/O (Input/Output) is how Python reads from and writes to the filesystem. The primary tool is the built-in `open()` function. 

Files can be opened in text mode (handling strings and character encodings) or binary mode (handling raw bytes). Because opening a file consumes operating system resources (file descriptors), it is critical to close them when done. Python's `with` statement acts as a context manager, guaranteeing that `close()` is called automatically, even if the block exits via an exception.

#### The Mind Picture
```
File I/O with Context Manager:

OS Filesystem
     ^
     | (1. open file descriptor)
[ with open('data.txt', 'w') as f: ]
     |
     | (2. f.write() buffers data)
     v
[ internal memory buffer ]
     |
     | (3. block ends -> f.close() -> flush buffer to OS)
     v
OS Filesystem (file saved)
```

#### Syntax Anatomy
```python
# 1. Context manager    # 2. File path          # 3. Mode   # 4. Encoding
with                    open("data.txt",        "w",        encoding="utf-8") as f:
                                                                              # 5. File object
    # 6. Operations inside block
    f.write("Hello")
# 7. f.close() called implicitly here
```

#### Practical Examples
Writing and reading text files line by line.


````python
import os

def write_and_read(filepath: str) -> None:
    # Write to file (truncates if exists)
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write("Line 1\n")
        f.writelines(["Line 2\n", "Line 3\n"])
        
    # Read from file line by line
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            # strip() removes the trailing newline character
            print(line.strip())

write_and_read("example_io.txt")
os.remove("example_io.txt")
````

Appending to an existing file.


````python
def append_log(filepath: str, message: str) -> None:
    with open(filepath, 'a', encoding='utf-8') as f:
        f.write(f"LOG: {message}\n")

append_log("app.log", "System started.")
append_log("app.log", "User logged in.")

with open("app.log", "r", encoding="utf-8") as f:
    print(f.read())

import os; os.remove("app.log")
````

Using `pathlib` for cleaner path handling.


````python
from pathlib import Path

def process_paths() -> None:
    # Create path object
    file_path = Path("config.txt")
    
    # Path objects have built-in read/write methods for simple cases
    file_path.write_text("setting=enabled", encoding="utf-8")
    
    print(f"Content: {file_path.read_text(encoding='utf-8')}")
    print(f"Exists: {file_path.exists()}")
    print(f"Name: {file_path.name}")
    print(f"Suffix: {file_path.suffix}")
    
    file_path.unlink() # Delete file

process_paths()
````

#### Common Mistakes and Pitfalls
Mistake 1: Relying on the platform's default encoding. On Windows, this is often cp1252, which will crash when trying to write Unicode characters.


````python
# WRONG
def bad_write() -> None:
    # Works on Linux, fails on Windows if chars aren't in cp1252
    with open("temp.txt", "w") as f:
        f.write("Hello Unicode")

# RIGHT
def good_write() -> None:
    with open("temp.txt", "w", encoding="utf-8") as f:
        f.write("Hello Unicode")
````

Mistake 2: Loading an entire large file into memory with `f.read()` or `f.readlines()`.


````python
# WRONG: Reads whole 5GB file into RAM
def bad_read(filepath: str) -> None:
    with open(filepath, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        for line in lines:
            pass # process line

# RIGHT: Iterates one line at a time, keeping RAM usage low
def good_read(filepath: str) -> None:
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            pass # process line
````

#### Best Practices
1. Always use `with open(...)` to manage files.
2. Always specify `encoding="utf-8"` when opening text files.
3. Iterate over the file object directly (`for line in f:`) rather than using `.readlines()` for line-by-line reading.
4. Prefer `pathlib.Path` over `os.path` for path manipulation; it provides an object-oriented, cross-platform interface.
5. Use binary mode (`'rb'`, `'wb'`) only for non-text data like images, zip files, or raw byte streams.

#### Real-World Use Case
Parsing a huge CSV file. Using the `csv` module along with file iteration allows processing gigabytes of data on a machine with limited RAM.

#### Performance Notes
Disk I/O is orders of magnitude slower than CPU operations. Python buffers file operations by default to reduce system calls. Reading line-by-line uses constant memory (O(1)) regardless of file size, whereas `.read()` uses memory proportional to file size (O(N)).

#### Quick Recap
- `open(path, mode, encoding)` returns a file object.
- Modes: `'r'` (read), `'w'` (write/truncate), `'a'` (append), `'b'` (binary).
- The `with` statement guarantees `close()` is called.
- Explicitly define `encoding="utf-8"` for cross-platform safety.
- Iterate the file object to read large files efficiently.
- `pathlib` is the modern approach to paths.

#### Interview Questions
**Junior**: Why do we use the `with` statement when opening files?
**Answer**: It acts as a context manager that ensures the file is properly closed when the block exits, even if an exception occurs. This prevents resource leaks (running out of file descriptors).

**Mid**: What happens if you call `open('file.txt', 'w')` on an existing file?
**Answer**: The `'w'` mode truncates the file immediately to zero bytes before you write anything. If you want to keep existing data, you should use `'a'` (append) or `'r+'` (read/write without truncation, overwriting from the start).

**Senior**: Explain the difference between `os.path` and `pathlib`.
**Answer**: `os.path` processes paths as string literals. `pathlib` represents paths as objects, offering methods like `.read_text()`, `/` for joining paths, and `.parent`, abstracting away OS-specific path separators and reducing boilerplate.

### Topic 1.19 — Modules and Packages

#### Concept Overview
A module is simply a file containing Python code. A package is a directory containing modules and an `__init__.py` file. Python's import system allows you to organize code into logical units and reuse them across projects.

When you import a module, Python searches `sys.path` (a list of directories) to find the corresponding file. The `__name__ == '__main__'` guard prevents code from running automatically when the file is imported by another script, allowing files to act as both executable scripts and importable modules.

#### The Mind Picture
```
Python Import Search Path (sys.path order):
1. Current working directory (or script directory)
2. PYTHONPATH environment variable
3. Standard Library directories
4. Site-packages (where pip installs things)

Package Structure:
my_project/
├── my_package/
│   ├── __init__.py     (Makes directory a package)
│   ├── module_a.py     (import my_package.module_a)
│   └── module_b.py
└── main.py
```

#### Syntax Anatomy
```python
# 1. Imports entire module
import math

# 2. Imports specific function, binds to local name
from os import getenv

# 3. Imports and aliases to avoid namespace collisions
import datetime as dt

# 4. Imports from a sub-package/module
from urllib.request import urlopen
```

#### Common Mistakes and Pitfalls
Mistake 1: Shadowing standard library modules. If you name your file `math.py` and write `import math`, Python imports your local file instead of the standard library math module, causing mysterious `AttributeError`s.


````python
# WRONG: Naming a script 'email.py'
# file: email.py
import email # This will import itself!

# RIGHT: Use a distinct name
# file: send_email_script.py
import email
````

1. Group imports at the top of the file. Order them: Standard Library, Third-Party, Local Application.
2. Avoid wildcard imports (`from module import *`). They pollute the namespace and obscure where names come from.

5. Always pin versions in `requirements.txt` to guarantee reproducible environments.

#### Performance Notes
Importing a module executes the entire file top-to-bottom. Python caches the resulting module object in `sys.modules`. Subsequent imports of the same module simply return the cached object, making them O(1) and virtually free.

#### Quick Recap
- A module is a `.py` file; a package is a directory with `__init__.py`.

### Topic 1.20 — Comprehensions Unified

#### Concept Overview
Comprehensions provide a concise, readable syntax for creating new collections (lists, dictionaries, sets) or generators based on existing iterables. They replace the pattern of creating an empty collection and running a `for` loop with `.append()`. 

Beyond brevity, comprehensions execute at C-speed inside the Python interpreter, making them faster than explicit `for` loops. Generator expressions look like list comprehensions but use parentheses; instead of building the entire collection in memory, they yield items one by one, saving RAM.

#### The Mind Picture
```
List Comprehension vs Generator Expression Memory Footprint:

Data: range(1, 1000000)

List Comprehension [x*2 for x in data]
RAM: [==================================================] (Allocates all 1M items)

Generator Expression (x*2 for x in data)
RAM: [=] (Stores only state object, yields 1 item at a time)
```

#### Syntax Anatomy
```python
# 1. Output Expression  # 2. Iteration          # 3. Optional Condition
[    x.upper()          for x in words          if x.startswith('a')    ]

# Dictionary Comprehension (requires key: value)
{    x: len(x)          for x in words          if x.startswith('a')    }

# Set Comprehension (uses braces, unique values)
{    x.upper()          for x in words          if x.startswith('a')    }

# Generator Expression (uses parentheses)
(    x.upper()          for x in words          if x.startswith('a')    )
```

#### Practical Examples
All four comprehension types side-by-side.


````python
def demonstrate_comprehensions() -> None:
    nums = [1, 2, 2, 3, 4, 5, 5]
    
    # List: keeps duplicates and order
    evens_list = [x * 2 for x in nums if x % 2 == 0]
    
    # Set: removes duplicates
    evens_set = {x * 2 for x in nums if x % 2 == 0}
    
    # Dict: maps item to computed value
    squares_dict = {x: x**2 for x in set(nums)}
    
    # Generator: lazy evaluation
    evens_gen = (x * 2 for x in nums if x % 2 == 0)
    
    print(f"List: {evens_list}")
    print(f"Set:  {evens_set}")
    print(f"Dict: {squares_dict}")
    print(f"Gen:  {list(evens_gen)}") # Consume generator

demonstrate_comprehensions()
````

Nested comprehensions for flattening a matrix.


````python
def flatten_matrix(matrix: list[list[int]]) -> list[int]:
    # The outer loop comes first, then the inner loop
    return [item for row in matrix for item in row]

matrix = [[1, 2], [3, 4], [5, 6]]
print(flatten_matrix(matrix))
````

#### Common Mistakes and Pitfalls
Mistake 1: Using comprehensions just for their side effects (like printing), rather than capturing the resulting collection. This builds a useless list in memory.


````python
items = [1, 2, 3]

# WRONG
[print(x) for x in items] # Creates [None, None, None] internally

# RIGHT
for x in items:
    print(x)
````

Mistake 2: Nesting comprehensions too deeply, destroying readability.


````python
data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

# WRONG: Unreadable
flat = [x for sub1 in data for sub2 in sub1 for x in sub2 if x % 2 == 0]

# RIGHT: Use a loop or helper generator
flat_good = []
for block in data:
    for row in block:
        for x in row:
            if x % 2 == 0:
                flat_good.append(x)
````

#### Best Practices
1. Use comprehensions to transform or filter data into a new collection.
2. Limit comprehensions to one or two levels of `for` loops. If it takes multiple lines to understand, use an explicit `for` loop.
3. Use generator expressions when iterating over large datasets where you don't need all items in memory simultaneously (e.g., passing to `sum()` or `any()`).
4. You can format comprehensions across multiple lines for better readability.

#### Real-World Use Case
Data cleaning. Taking a list of raw dictionary records from an API and building a cleaned mapping of IDs to names, filtering out inactive users, all in one fast dictionary comprehension.

#### Performance Notes
List comprehensions are faster than explicit `for` loops with `.append()` because the `append` lookup and function call are bypassed; the loop runs in C code. However, `map()` with a built-in function (like `str`) can sometimes be slightly faster than a comprehension. For memory limits, generator expressions are O(1) space, while list comprehensions are O(N) space.

#### Quick Recap
- Four types: List `[]`, Dict `{k: v}`, Set `{}`, Generator `()`.
- Syntax: `[expression for item in iterable if condition]`.
- List stores all data in memory; Generator evaluates lazily.
- Comprehensions are faster than `append()` loops.
- Avoid nesting beyond 2 levels; readability comes first.

#### Interview Questions
**Junior**: What is a dictionary comprehension and how does it differ syntactically from a set comprehension?
**Answer**: Both use curly braces `{}`, but a dictionary comprehension requires a key-value pair separated by a colon (e.g., `{k: v for ...}`), whereas a set comprehension just has single expressions (e.g., `{expr for ...}`).

**Mid**: When would you choose a generator expression over a list comprehension?
**Answer**: When working with a very large iterable where storing the complete generated list in memory is unnecessary or would cause a MemoryError. Generator expressions yield one item at a time (lazy evaluation) resulting in O(1) memory usage.

**Senior**: Explain why `[x for row in matrix for x in row]` flattens a matrix. Why is the order `for row` then `for x`?
**Answer**: In a list comprehension, the `for` clauses are evaluated in the same order they would be written as nested `for` loops. The outermost loop (`for row in matrix`) must come first, followed by the inner loop (`for x in row`), even though the output expression `x` sits at the very beginning.

### Topic 2.1 — OOP Fundamentals

#### Concept Overview
Python's Object-Oriented Programming model provides a structured way to bundle data and behavior. Unlike languages that enforce strict access modifiers, Python relies on convention and dynamic typing. When you instantiate a class, Python allocates memory for a new object and calls `__new__` to create it, followed by `__init__` to initialize its state.

Classes define both shared behavior (methods) and shared state (class attributes). Instances carry their own unique state (instance attributes). The parameter `self` is explicitly required in method definitions but implicitly passed during method calls. It is not a reserved keyword, merely a strong convention. The class mechanism solves the problem of state management by encapsulating related variables and functions into single, coherent units.

#### The Mind Picture

Instances store their attributes in a dynamic dictionary called `__dict__`. The class itself is also an object, with its own `__dict__` storing methods and class attributes.

```
+-----------------------+      __class__     +-----------------------+
|  Instance (obj)       | -----------------> |  Class (MyClass)      |
+-----------------------+                    +-----------------------+
| __dict__:             |                    | __dict__:             |
|   'name': "Alice"     |                    |   'shared': 42        |
|   'id': 105           |                    |   '__init__': <func>  |
+-----------------------+                    |   'method': <func>    |
                                             +-----------------------+
```

#### Syntax Anatomy

```python
class Dog:
    species: str = "Canis familiaris"  # Class attribute (shared)

    def __init__(self, name: str) -> None:
        self.name = name  # Instance attribute (unique per instance)
```

- `class Dog:` - Defines the blueprint.
- `species` - Evaluated at class definition time.
- `def __init__(self, name: str)` - The initializer. `self` binds to the newly created instance.
- `self.name` - Attaches data to the specific instance's `__dict__`.

#### Practical Examples

First, defining a basic class and observing the `self` convention and instantiation process.

````python
class User:
    def __init__(this_can_be_anything: object, username: str) -> None:
        this_can_be_anything.username = username # type: ignore

u = User("admin")
print(u.username)
````

Next, analyzing the difference between class attributes and instance attributes. Modifying a class attribute via the class affects all instances, while modifying it via an instance creates a shadowing instance attribute.

````python
class Configuration:
    timeout: int = 30  # Class attribute

c1 = Configuration()
c2 = Configuration()

print(f"Initial: c1={c1.timeout}, c2={c2.timeout}")
Configuration.timeout = 60
print(f"Class update: c1={c1.timeout}, c2={c2.timeout}")

c1.timeout = 100  # Creates an instance attribute shadowing the class attribute
print(f"Instance update: c1={c1.timeout}, c2={c2.timeout}")
print(f"Class attribute remains: {Configuration.timeout}")
````

Finally, inspecting the memory addresses using `id()` and the underlying `__dict__` to prove how attributes are stored.

````python
class Connection:
    active_connections: list[str] = []  # Mutable class attribute
    
    def __init__(self, name: str) -> None:
        self.name = name

conn1 = Connection("db1")
conn2 = Connection("db2")

print(f"conn1 dict: {conn1.__dict__}")
print(f"conn2 dict: {conn2.__dict__}")

print(f"Class list id: {id(Connection.active_connections)}")
print(f"conn1 list id: {id(conn1.active_connections)}")
print(f"Same object? {id(Connection.active_connections) == id(conn1.active_connections)}")

# Mutating the shared mutable object
conn1.active_connections.append(conn1.name)
print(f"conn2 sees the change: {conn2.active_connections}")
````

#### Common Mistakes and Pitfalls

Mistake: Using mutable default values as class attributes when you intend them to be instance-specific.

````python
class BadRegistry:
    # BUG: This list is shared across all instances
    items: list[str] = []
    
    def add(self, item: str) -> None:
        self.items.append(item)

# Fix: Initialize mutable state in __init__
class GoodRegistry:
    def __init__(self) -> None:
        self.items: list[str] = []
        
    def add(self, item: str) -> None:
        self.items.append(item)
````

#### Best Practices

1. Always name the first parameter `self` for instance methods. Reviewers will reject anything else.
2. Do not use mutable class attributes unless implementing a specific pattern like a global cache or registry.
3. Keep `__init__` minimal. Avoid heavy computation or network requests during initialization.
4. Type hint all attributes directly in the class body if using dataclasses, or in `__init__` for standard classes.
5. Prefer `__init__` for setting instance variables. Do not dynamically add attributes outside the class definition.

#### Real-World Use Case

The `urllib3.PoolManager` uses instance attributes to track configuration like retries and headers specific to one client session, while managing an internal connection pool. You instantiate it once per application lifecycle, maintaining state across multiple HTTP requests without polluting the global namespace.

#### Performance Notes

Attribute access in Python requires dictionary lookups. Accessing `self.name` looks in the instance `__dict__`. If not found, it falls back to the class `__dict__`, and then traverses the method resolution order. For extreme performance scenarios, using `__slots__` bypasses the instance dictionary and allocates a static array, saving memory and speeding up attribute access.

#### Quick Recap

- Classes act as factories for creating instances.
- `__new__` allocates memory; `__init__` populates the initial state.
- `self` binds to the current instance; it is required but is just a naming convention.
- Class attributes live in the class `__dict__` and are shared.
- Instance attributes live in the instance `__dict__` and are unique.
- Modifying a class attribute on an instance creates a local copy, masking the class attribute.

#### Interview Questions

**Junior**: What is `self` in Python and why is it necessary?
**Answer**: `self` refers to the instance calling the method. Python passes the instance as the first argument automatically when a method is called. It is necessary because Python does not use special syntax (like `this.` in Java) to resolve instance variables; it requires explicit scoping.

**Mid**: Explain what happens if you mutate a class attribute list from an instance.
**Answer**: Because lists are mutable, calling `.append()` on a class attribute via an instance mutates the original list stored on the class object. All instances will see the modification. If you use assignment (`self.items = []`), you create a new instance attribute that shadows the class attribute.

**Senior**: Describe the sequence of execution when you call `MyClass()`.
**Answer**: First, `MyClass.__call__` (inherited from the metaclass `type`) is invoked. This calls `MyClass.__new__(MyClass)` to allocate and return an uninitialized object. Then, if the returned object is an instance of `MyClass`, it calls `MyClass.__init__(self)` on that new object to populate its state. Finally, the initialized object is returned to the caller.

### Topic 2.2 — Methods: instance, @classmethod, @staticmethod

#### Concept Overview
Python provides three primary method types to organize logic within a class. Instance methods require an instance and operate on its specific state. Class methods receive the class object explicitly and operate on class-level state. Static methods receive neither; they are standard functions nested in a class namespace for organizational grouping. Choosing the right method type signals intent to other developers and enforces appropriate data access boundaries.

#### The Mind Picture

```
Method Type     | Implicit First Arg | Access Level
----------------|--------------------|-------------------------
Instance Method | self (instance)    | Instance + Class state
@classmethod    | cls (class)        | Class state only
@staticmethod   | None               | Neither
```

#### Syntax Anatomy

```python
class Worker:
    def process(self) -> None:          # Instance method
        pass

    @classmethod
    def create(cls) -> "Worker":        # Class method
        return cls()

    @staticmethod
    def validate(data: str) -> bool:    # Static method
        return len(data) > 0
```

#### Practical Examples

A full class demonstrating all three method types working together.

````python
import json
from typing import Self

class AppConfig:
    _default_retries: int = 3

    def __init__(self, host: str, retries: int) -> None:
        self.host = host
        self.retries = retries

    def get_connection_string(self) -> str:
        # Instance method: needs self.host
        return f"tcp://{self.host}?retries={self.retries}"

    @classmethod
    def from_json(cls, payload: str) -> Self:
        # Class method: acts as an alternative constructor
        data = json.loads(payload)
        return cls(host=data['host'], retries=data.get('retries', cls._default_retries))

    @staticmethod
    def is_valid_host(host: str) -> bool:
        # Static method: pure utility logic
        return "." in host and not host.startswith("http")

# Usage
config = AppConfig.from_json('{"host": "db.local"}')
print(config.get_connection_string())
print(AppConfig.is_valid_host("db.local"))
````

#### Common Mistakes and Pitfalls

Mistake: Using `@staticmethod` when the method needs to instantiate the class. If the class name changes, or if it is subclassed, the static method returns the hardcoded parent class.

````python
class BaseParser:
    @staticmethod
    def create_bad() -> "BaseParser":
        return BaseParser()  # Hardcoded class name

    @classmethod
    def create_good(cls) -> "BaseParser":
        return cls()  # Returns the actual subclass if called via subclass

class JsonParser(BaseParser):
    pass

bad_parser = JsonParser.create_bad()
print(type(bad_parser))  # BaseParser (unexpected)

good_parser = JsonParser.create_good()
print(type(good_parser))  # JsonParser (expected)
````

#### Best Practices

1. Use `@classmethod` for alternative constructors (e.g., `from_string`, `from_file`).
2. Name the first argument of a class method `cls`.
3. Avoid `@staticmethod` if the method depends on class variables; use `@classmethod` instead.
4. If a method does not use `self` or `cls`, consider extracting it to a module-level function unless it strongly logically belongs to the class namespace.
5. Keep alternative constructors clean by parsing data and passing it directly to `__init__`.

#### Real-World Use Case

The standard library `datetime.datetime` class relies heavily on class methods for alternative constructors. Methods like `datetime.now()`, `datetime.fromtimestamp()`, and `datetime.strptime()` all use the class method pattern to parse various inputs and return a newly allocated instance of the class.

#### Performance Notes

Method dispatch incurs slight overhead. Instance methods bind `self` automatically, which takes a few CPU cycles. Static methods avoid this binding step, making them infinitesimally faster, though this difference is entirely negligible in IO-bound or typical applications. Optimizing method types for performance is a micro-optimization to avoid.

#### Quick Recap

- Instance methods take `self` and modify/read object state.
- Class methods take `cls`, decorated with `@classmethod`, and modify/read class state or instantiate objects.
- Static methods take neither, decorated with `@staticmethod`, acting as namespaced utility functions.
- Alternative constructors are the dominant use case for class methods.
- Subclasses inherit class methods and `cls` binds to the subclass, enabling polymorphic instantiation.

#### Interview Questions

**Junior**: Why does `fromtimestamp` in `datetime` need to be a class method rather than an instance method?
**Answer**: Because its purpose is to create and return a new instance based on the timestamp. If it were an instance method, you would already need a `datetime` instance to call it, defeating the purpose of a constructor.

**Mid**: How does `@classmethod` interact with inheritance compared to `@staticmethod`?
**Answer**: When a subclass calls an inherited `@classmethod`, the `cls` argument evaluates to the subclass, allowing the method to create subclass instances or access subclass attributes. An inherited `@staticmethod` has no reference to the class, so it behaves exactly as it did in the parent, with no awareness of the subclass.

**Senior**: How are class methods implemented under the hood in Python?
**Answer**: Python uses the descriptor protocol. The `@classmethod` decorator transforms the function into a class method object. When accessed via the class or an instance, its `__get__` method is invoked, which returns a bound method object enclosing the class itself as the first implicit argument.

### Topic 2.3 — Inheritance: single, multiple, MRO intro

#### Concept Overview
Inheritance allows a new class to inherit attributes and methods from existing classes, promoting code reuse. Python supports both single and multiple inheritance. When a class inherits from multiple parents, conflicts can arise if parent classes share method names. Python resolves this deterministically using the Method Resolution Order (MRO), specifically the C3 linearization algorithm, which guarantees a consistent, monotonic hierarchy sequence.

#### The Mind Picture

```
   A
  / \    Diamond Problem: If D calls a method defined in A, B, and C,
 B   C   which one executes?
  \ /    MRO ensures the order is D -> B -> C -> A -> object.
   D
```

#### Syntax Anatomy

```python
class Parent:
    pass

class Mixin:
    pass

class Child(Parent, Mixin):
    def __init__(self) -> None:
        super().__init__()  # Dispatches to the next class in the MRO
```

#### Practical Examples

Starting with standard single inheritance and using `super()` to extend parent logic.

````python
class Employee:
    def __init__(self, name: str) -> None:
        self.name = name

    def get_role(self) -> str:
        return "Staff"

class Manager(Employee):
    def __init__(self, name: str, department: str) -> None:
        super().__init__(name)
        self.department = department

    def get_role(self) -> str:
        return f"Manager of {self.department}"

m = Manager("Bob", "Engineering")
print(m.name, m.get_role())
````

Demonstrating multiple inheritance and observing the C3 Method Resolution Order in action.

````python
class BaseLogger:
    def log(self) -> str:
        return "Base log"

class ConsoleLogger(BaseLogger):
    def log(self) -> str:
        return "Console: " + super().log()

class FileLogger(BaseLogger):
    def log(self) -> str:
        return "File: " + super().log()

class AppLogger(ConsoleLogger, FileLogger):
    pass

logger = AppLogger()
print(logger.log())
print([cls.__name__ for cls in AppLogger.mro()])
````

#### Common Mistakes and Pitfalls

Mistake: Calling parent methods directly by class name instead of using `super()`. This breaks the MRO in multiple inheritance setups, leading to duplicate method calls or skipped classes.

````python
class A:
    def ping(self) -> None: print("A ping")

class B(A):
    def ping(self) -> None:
        A.ping(self)  # BUG: Hardcoding parent bypasses MRO
        print("B ping")

class C(A):
    def ping(self) -> None:
        super().ping()  # Correct
        print("C ping")
````

#### Best Practices

1. Always use `super()` instead of hardcoding parent class names.
2. In multiple inheritance, design parent classes cooperatively. All methods sharing a name across the hierarchy must accept compatible signatures (often using `**kwargs`).
3. Prefer composition over inheritance. Use multiple inheritance strictly for Mixins (classes containing small, isolated pieces of functionality without complex state).
4. Check the `__mro__` attribute when debugging unexpected method dispatch behavior.
5. Ensure all `__init__` methods in a multiple inheritance chain call `super().__init__(**kwargs)`.

#### Real-World Use Case

Django's Class-Based Views heavily utilize multiple inheritance. A view like `CreateView` inherits from `SingleObjectTemplateResponseMixin`, `BaseCreateView`, and others. The MRO guarantees that view rendering logic, form validation, and HTTP response generation execute in the exact intended sequence.

#### Performance Notes

Deep inheritance trees increase the time required for attribute and method resolution, as Python must traverse the MRO dictionary chain. The C3 algorithm is executed once at class creation time, so building the class is slightly slower, but method execution incurs only the runtime dictionary lookup cost along the cached MRO list.

#### Quick Recap

- Python supports single and multiple inheritance.
- `super()` does not just call the parent; it delegates to the next class in the computed MRO.
- The MRO uses C3 linearization to flatten the inheritance graph into a strict list.
- You can inspect the MRO using `ClassName.mro()` or `ClassName.__mro__`.
- The diamond problem is resolved smoothly because the shared base class appears only once, at the end of the MRO.

#### Interview Questions

**Junior**: What does `super()` do in Python?
**Answer**: It returns a proxy object that delegates method calls to a parent or sibling class. It is primarily used to invoke an overridden method from a superclass.

**Mid**: Explain the diamond problem and how Python solves it.
**Answer**: The diamond problem occurs in multiple inheritance when a class inherits from two classes that share a common base. If the child calls a method defined in the base class, the language must decide which path to follow. Python solves this using the C3 Method Resolution Order (MRO) algorithm, which flattens the hierarchy into a linear sequence ensuring the common base class is checked last, and no class is checked twice.

**Senior**: In multiple inheritance, why must sibling classes use `super()` cooperatively even if they don't directly inherit from each other?
**Answer**: Because `super()` delegates to the *next class in the MRO of the calling instance*, not necessarily the lexical parent of the class defining the method. If class `B` and `C` inherit from `A`, and `D` inherits from `B` and `C`, calling `super()` inside `B` will actually dispatch to `C` when executed on an instance of `D`. If `B` does not call `super()`, the chain breaks, and `C`'s method is never executed.

### Topic 2.4 — Polymorphism and Abstraction in Practice

#### Concept Overview
Polymorphism allows different objects to be treated through a uniform interface. Python implements this natively via "duck typing": if an object walks like a duck and quacks like a duck, it is treated as a duck. However, in large codebases, implicit duck typing causes bugs. Abstraction formalizes these interfaces. The `abc` (Abstract Base Classes) module enforces contracts by requiring subclasses to implement specific methods, preventing incomplete subclasses from being instantiated.

#### The Mind Picture

```
Duck Typing:                    Abstract Base Class (ABC):
+----------+   +----------+     +------------------------+
|   Bird   |   | Airplane |     | <<Abstract>>           |
+----------+   +----------+     | Flyer                  |
| fly()    |   | fly()    |     | @abstractmethod fly()  |
+----------+   +----------+     +------------------------+
      \             /                     ^      ^
       \           /                     /        \ 
      perform_flight(obj)          Bird        Airplane
                                  (must implement fly)
```

#### Syntax Anatomy

```python
from abc import ABC, abstractmethod

class Storage(ABC):
    @abstractmethod
    def save(self, data: bytes) -> None:
        pass  # Subclasses MUST override this
```

#### Practical Examples

Demonstrating classic duck typing where objects share no lineage but fulfill the same interface.

````python
class PDFFormatter:
    def format(self) -> str:
        return "PDF Content"

class CSVFormatter:
    def format(self) -> str:
        return "CSV Content"

def generate_report(formatter: PDFFormatter | CSVFormatter) -> None:
    # Duck typing in action: we just call .format()
    print(formatter.format())

generate_report(CSVFormatter())
````

Transitioning to strict abstraction using the `abc` module to enforce contracts.

````python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def process(self, amount: float) -> bool:
        pass

class StripeGateway(PaymentGateway):
    def process(self, amount: float) -> bool:
        print(f"Stripe processing ${amount}")
        return True

class IncompleteGateway(PaymentGateway):
    pass

# Works fine
gateway = StripeGateway()
gateway.process(100.0)

try:
    # Fails immediately at instantiation
    bad_gateway = IncompleteGateway()
except TypeError as e:
    print(f"Caught error: {e}")
````

#### Common Mistakes and Pitfalls

Mistake: Forgetting to inherit from `ABC` when using `@abstractmethod`. The decorator alone does not prevent instantiation; the metaclass `ABCMeta` (provided by `ABC`) enforces the rule.

````python
from abc import abstractmethod

class FakeAbstract:
    @abstractmethod
    def run(self) -> None:
        pass

# BUG: This instantiates perfectly fine because ABC was omitted
obj = FakeAbstract()
print("Object created despite abstractmethod!")
````

#### Best Practices

1. Use ABCs to define clear architectural boundaries, especially for plugins or backend strategies.
2. Inherit from `abc.ABC` rather than defining `metaclass=abc.ABCMeta` manually, as it is cleaner.
3. Combine `@abstractmethod` with `@property` to force subclasses to define specific class or instance properties.
4. Type hint abstract parameters explicitly to ensure subclasses match the expected signature.
5. Rely on duck typing for small scripts or simple utilities, but transition to ABCs or `typing.Protocol` in libraries and enterprise code.

#### Real-World Use Case

The `collections.abc` module defines the core interfaces of Python data structures. For example, `Sequence` is an ABC. If you write a custom list-like object and inherit from `Sequence`, you only need to implement `__getitem__` and `__len__`; the ABC automatically provides concrete implementations for `__iter__`, `__contains__`, `index`, and `count`.

#### Performance Notes

ABCs introduce a slight overhead during class definition and instantiation due to metaclass checks verifying that all abstract methods are implemented. However, method invocation on the instantiated object runs at identical speeds to standard classes. The safety guarantees vastly outweigh the negligible startup cost.

#### Quick Recap

- Duck typing assumes object compatibility based on method names, not inheritance.
- Abstract Base Classes (ABCs) enforce implementation contracts.
- You cannot instantiate an ABC if it has unimplemented abstract methods.
- Use `from abc import ABC, abstractmethod` to create an interface.
- Subclasses missing an `@abstractmethod` fail at instantiation time, not at call time.

#### Interview Questions

**Junior**: What is duck typing in Python?
**Answer**: Duck typing is the concept where an object's suitability is determined by the presence of certain methods and properties, rather than the type of the object itself. "If it walks and quacks like a duck, it's a duck."

**Mid**: What happens if you subclass an ABC but forget to implement one of its abstract methods?
**Answer**: Python will raise a `TypeError` when you attempt to instantiate the subclass. The error occurs at instantiation time, not when the class is defined or when the method is called.

**Senior**: How does `typing.Protocol` differ from `abc.ABC` for defining interfaces?
**Answer**: `abc.ABC` requires explicit inheritance (nominal subtyping). The subclass must explicitly inherit from the ABC. `typing.Protocol` supports structural subtyping. A class satisfies a Protocol simply by implementing the required methods, without any explicit inheritance or runtime dependency on the Protocol. Protocols are purely for static type checking, whereas ABCs enforce constraints at runtime.

# Topic 2.5 — Dunder Methods
#### Concept Overview
Dunder (double underscore) methods are Python's mechanism for operator overloading and class behavior customization. Also known as magic methods, they allow user-defined objects to interact natively with built-in functions and operators. Instead of writing `vector.add(other)`, dunders enable `vector + other`. This provides a consistent interface across the language, ensuring that custom objects feel like built-in types.

#### The Mind Picture
```text
+-------------------+
| Object Instance   |  <-- print(obj)  ==> calls __str__()
|                   |  <-- repr(obj)   ==> calls __repr__()
|  +-------------+  |  <-- len(obj)    ==> calls __len__()
|  | __str__     |  |  <-- obj + other ==> calls __add__()
|  | __repr__    |  |  <-- obj == other==> calls __eq__()
|  | __eq__      |  |
|  | __hash__    |  |  <-- set(obj)    ==> needs __hash__ & __eq__
|  +-------------+  |
+-------------------+
```

#### Syntax Anatomy
```python
class Anatomy:
    def __init__(self, value: int) -> None:
        self.value = value
        
    def __str__(self) -> str:
        # Human-readable string representation
        return f"Anatomy with {self.value}"
        
    def __repr__(self) -> str:
        # Unambiguous string, ideally evaluable: eval(repr(obj)) == obj
        return f"Anatomy({self.value})"
```

#### Practical Examples


````python
from typing import Any

class Vector2D:
    def __init__(self, x: float, y: float) -> None:
        self.x = x
        self.y = y

    def __str__(self) -> str:
        return f"({self.x}, {self.y})"

    def __repr__(self) -> str:
        return f"Vector2D({self.x}, {self.y})"
````

Next, we implement equality and hashing so Vectors can be compared and used in sets.


````python
class Vector2DEq(Vector2D):
    def __eq__(self, other: Any) -> bool:
        if not isinstance(other, type(self)):
            return NotImplemented
        return self.x == other.x and self.y == other.y

    def __hash__(self) -> int:
        return hash((self.x, self.y))
````

Finally, let's add operator overloading for arithmetic, length, and containment.


````python
import math

class Vector2DFull(Vector2DEq):
    def __add__(self, other: "Vector2DFull") -> "Vector2DFull":
        if not isinstance(other, Vector2DFull):
            return NotImplemented
        return Vector2DFull(self.x + other.x, self.y + other.y)

    def __len__(self) -> int:
        return 2

    def __contains__(self, item: float) -> bool:
        return item == self.x or item == self.y
````

#### Common Mistakes and Pitfalls
Forgetting that overriding `__eq__` makes `__hash__` None by default.


````python
class BadHash:
    def __init__(self, x: int) -> None:
        self.x = x
    def __eq__(self, other: Any) -> bool:
        return isinstance(other, BadHash) and self.x == other.x
    # FIX: Implement __hash__ if you need it in a set/dict
    # def __hash__(self) -> int: return hash(self.x)
    
# b = BadHash(1)
# s = {b}  # Raises TypeError: unhashable type: 'BadHash'
````

#### Best Practices
- Return `NotImplemented` from comparison and arithmetic dunders if the types are unsupported, rather than raising `TypeError`.
- Ensure `__eq__` checks `isinstance`.
- Keep `__repr__` unambiguous, ideally returning a string that reconstructs the object.
- Make `__hash__` rely on an immutable tuple of the object's defining attributes.
- Do not implement `__len__` if your object is not a container.

#### Real-World Use Case
The `pathlib.Path` class heavily uses dunder methods. It implements `__truediv__` to allow path joining via the `/` operator, creating a domain-specific language for paths: `Path('/tmp') / 'file.txt'`.

#### Performance Notes
Dunder method lookups are highly optimized in CPython, bypassing the standard `__getattribute__` machinery. They are resolved directly on the class's type object (the C struct `tp_str`, `tp_richcompare`, etc.), meaning instance-level dunders are ignored and class-level lookups are O(1) fast.

#### Quick Recap
- Dunder methods define object behavior for built-in operations.
- `__str__` is for users (printing); `__repr__` is for developers (debugging/REPL).
- Defining `__eq__` removes default hashability; you must implement `__hash__` to restore it.
- Return `NotImplemented` for unsupported operations to allow Python to try the reversed operator on the other operand.
- The `in` operator uses `__contains__`, while `+` uses `__add__`.

#### Interview Questions
- **Junior**: What is the difference between `__str__` and `__repr__`? 
  **Answer**: `__str__` is for readable string representations intended for end-users, called by `print()` and `str()`. `__repr__` is unambiguous and intended for developers, called by `repr()` and the REPL. It should ideally be valid Python code to recreate the object.
- **Mid**: Why does an object become unhashable when you implement `__eq__`?
  **Answer**: Python defaults to hashing by memory address (identity). If you define custom equality (`__eq__`), objects with different addresses might be equal. If they are equal, they must have the same hash to work in sets/dicts. Python deletes the default `__hash__` to prevent you from violating this invariant accidentally.
- **Senior**: How does Python handle `a + b` if `a.__add__(b)` returns `NotImplemented`?
  **Answer**: Python falls back to calling `b.__radd__(a)`. If `b` does not implement `__radd__` or it also returns `NotImplemented`, Python finally raises a `TypeError`. This allows right-hand operands to support operations with left-hand operands they don't explicitly know about.

# Topic 2.6 — Properties
#### Concept Overview
Properties provide a way to manage attribute access using getter, setter, and deleter methods, while exposing a clean attribute-style public API. Python rejects the Java-style `get_x()` and `set_x()` boilerplate. Instead, you start with simple public attributes. If you later need to add validation, logging, or computed values, you convert the attribute to a property without breaking the interface of the consumers.

#### The Mind Picture
```text
Client Code: obj.x = 5
       |
       v
+-----------------------+
| @x.setter             |
| def x(self, value):   |  <-- Intercepts assignment
|     if value < 0:     |  <-- Validation logic
|         raise Error   |
|     self._x = value   |  <-- Stores in private backing field
+-----------------------+
```

#### Syntax Anatomy
```python
class PropertyAnatomy:
    def __init__(self) -> None:
        self._attr = 0  # Backing field
        
    @property
    def attr(self) -> int:
        return self._attr
        
    @attr.setter
    def attr(self, value: int) -> None:
        self._attr = value
        
    @attr.deleter
    def attr(self) -> None:
        del self._attr
```

#### Practical Examples


````python
class Temperature:
    def __init__(self, celsius: float) -> None:
        self.celsius = celsius  # Uses the setter immediately!

    @property
    def celsius(self) -> float:
        return self._celsius

    @celsius.setter
    def celsius(self, value: float) -> None:
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is impossible")
        self._celsius = value
````

Properties are also excellent for computed attributes that don't need to be stored.


````python
class Circle:
    def __init__(self, radius: float) -> None:
        self.radius = radius

    @property
    def diameter(self) -> float:
        return self.radius * 2

    @diameter.setter
    def diameter(self, value: float) -> None:
        self.radius = value / 2
````

#### Common Mistakes and Pitfalls
Infinite recursion in getters or setters by accessing the property name instead of the backing field.


````python
class InfiniteRecursion:
    def __init__(self, val: int) -> None:
        self.val = val  # Setter called

    @property
    def val(self) -> int:
        # FIX: return self._val
        return self.val  # RecursionError!

    @val.setter
    def val(self, value: int) -> None:
        # FIX: self._val = value
        self.val = value  # RecursionError!
````

#### Best Practices
- Never use `get_x()` and `set_x()` methods; use plain attributes first, then properties if logic is needed.
- Name the backing attribute with a single leading underscore (e.g., `_x`).
- Keep property getters computationally cheap. If it takes significant time or I/O, make it an explicit method (e.g., `fetch_data()`) so the caller knows it's expensive.
- Properties should not have side effects beyond modifying the object's internal state.

#### Real-World Use Case
In SQLAlchemy, ORM models often use properties to expose compound fields or parse raw database columns. A `User` model might have a `@property def full_name(self)` that computes `f"{self.first_name} {self.last_name}"` dynamically, ensuring it is always in sync with the underlying columns.

#### Performance Notes
Attribute access via a property is slightly slower than direct attribute access because it involves a function call and the descriptor protocol overhead. However, the difference is negligible for all but the most extreme, tight-loop performance requirements.

#### Quick Recap
- Properties turn methods into attributes.
- Use `@property` for the getter.
- Use `@name.setter` and `@name.deleter` for setters and deleters.
- They allow you to add validation later without breaking the public API.
- Backing fields should be prefixed with an underscore.
- Under the hood, `@property` implements the descriptor protocol.

#### Interview Questions
- **Junior**: Why avoid Java-style getters and setters in Python?
  **Answer**: Python supports the `@property` decorator, which allows you to intercept attribute access natively. This means you can start with plain attributes and upgrade to properties later without forcing callers to change `obj.x` to `obj.get_x()`.
- **Mid**: How does `@property` avoid infinite recursion?
  **Answer**: By explicitly reading and writing to a different "private" backing attribute, conventionally prefixed with an underscore (like `_x`). If the setter writes to `self.x` instead of `self._x`, it recursively invokes itself.
- **Senior**: How is `@property` actually implemented under the hood in Python?
  **Answer**: `@property` is a built-in class that implements the descriptor protocol (`__get__`, `__set__`, `__delete__`). When you decorate a method with `@property`, you are creating a property descriptor object on the class. When an instance accesses that attribute, Python's `__getattribute__` machinery defers the lookup to the descriptor's `__get__` method.

# Topic 2.7 — Iterators and the Iterator Protocol
#### Concept Overview
Iteration in Python is driven by a formal protocol. An **iterable** is anything that can be looped over (like a list or string). An **iterator** is the engine that does the looping. The distinction is crucial: an iterable produces an iterator when asked. Iterators maintain state (where they are in the sequence) and generate values one at a time via `next()`. Once an iterator runs out of values, it signals exhaustion by raising `StopIteration`.

#### The Mind Picture
```text
 Iterable (e.g. List)          Iterator
+-------------------+        +-------------------+
| [10, 20, 30]      | iter() | pointer -> 0      | next() -> 10
| __iter__() --------======> | __iter__()        | next() -> 20
+-------------------+        | __next__()        | next() -> 30
                             +-------------------+ next() -> StopIteration
```

#### Syntax Anatomy
```python
# Iterable Protocol
class MyIterable:
    def __iter__(self):
        return MyIterator()

# Iterator Protocol
class MyIterator:
    def __iter__(self):
        return self
    
    def __next__(self):
        # return next item OR raise StopIteration
        pass
```

#### Practical Examples


````python
from typing import Iterator

class CountdownIterator:
    def __init__(self, start: int) -> None:
        self.current = start

    def __iter__(self) -> Iterator[int]:
        return self

    def __next__(self) -> int:
        if self.current <= 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value
````

Let's see how a `for` loop interacts with iterables manually.


````python
# A standard for loop:
# for x in [1, 2, 3]: print(x)
# Is syntactic sugar for:

my_list = [1, 2, 3]
iterator = iter(my_list)  # Calls my_list.__iter__()

while True:
    try:
        item = next(iterator)  # Calls iterator.__next__()
        # print(item)
    except StopIteration:
        break
````

#### Common Mistakes and Pitfalls
Attempting to reuse an exhausted iterator.


````python
nums = [1, 2, 3]
num_iter = iter(nums)

list1 = list(num_iter)  # [1, 2, 3]
list2 = list(num_iter)  # [] - The iterator is exhausted!
# FIX: Get a fresh iterator by calling iter(nums) again.
````

#### Best Practices
- Iterators must always return `self` from `__iter__`. This ensures that iterators are also iterables, meaning you can pass an iterator to anything that expects an iterable (like a `for` loop).
- Prefer generators (Topic 2.8) over writing manual iterator classes, as they handle the state and `StopIteration` boilerplate automatically.
- Never catch `StopIteration` manually unless you are explicitly building or unwrapping the protocol. Let `for` loops and comprehensions handle it.

#### Real-World Use Case
The `csv.reader` in the standard library returns an iterator. As you call `next()` on it, it reads the next line from the file, parses it, and returns a list of strings. It doesn't load the entire CSV into memory, making it capable of parsing multi-gigabyte files efficiently.

#### Performance Notes
Iterators are O(1) in memory space. They only hold the state necessary to produce the next element, not the entire collection. Time complexity for `next()` is typically O(1) depending on the generation logic.

#### Quick Recap
- Iterable: has `__iter__()` returning an iterator.
- Iterator: has `__iter__()` returning `self`, and `__next__()` returning the next value.
- `StopIteration` is raised to signal the end of the sequence.
- Iterators are single-use; once exhausted, they are permanently empty.
- A `for` loop is syntactic sugar over `iter()` and `next()`.

#### Interview Questions
- **Junior**: What is the difference between an iterable and an iterator?
  **Answer**: An iterable is a collection that can be looped over (like a list), and it implements `__iter__` to return a new iterator. An iterator is the object that remembers the state of iteration, implementing `__next__` to return items one by one and raising `StopIteration` when done.
- **Mid**: Why do iterators also implement `__iter__` returning `self`?
  **Answer**: So they can be used anywhere an iterable is expected. For example, if a function takes an iterable and does `for item in it:`, passing an iterator directly works because the `for` loop calls `iter()` on it, which returns the iterator itself.
- **Senior**: How can you check if an object is an iterator without using the `typing` module or `collections.abc`?
  **Answer**: You can check if the object has both `__iter__` and `__next__` methods using `hasattr(obj, '__iter__') and hasattr(obj, '__next__')`. Additionally, you could check if `iter(obj) is obj`.

# Topic 2.8 — Generators
#### Concept Overview
Generators are the pythonic, elegant way to create iterators. Instead of writing a class with `__iter__` and `__next__` methods to maintain state, you write a standard function and use the `yield` keyword. When Python sees `yield`, it marks the function as a generator. Calling the function doesn't execute it; it returns a lazy generator object. Each time `next()` is called on this object, execution resumes until it hits `yield`, returning the yielded value and pausing the execution frame, keeping all local variables perfectly intact.

#### The Mind Picture
```text
  Generator Function         Generator Object
+--------------------+      +--------------------+
| def count():       |      | frame state        |
|   yield 1          | ===> | locals: {}         | -> next() -> 1
|   yield 2          |      | instruction ptr    | -> next() -> 2
+--------------------+      +--------------------+ -> next() -> StopIteration
```

#### Syntax Anatomy
```python
def simple_generator():
    yield 1
    yield 2
    # Implicit return None raises StopIteration automatically

gen = simple_generator() # Returns generator object, code hasn't run
val = next(gen)          # Runs to first yield
```

#### Practical Examples


````python
from typing import Generator

def fibonacci(limit: int) -> Generator[int, None, None]:
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b

# Using the generator
for num in fibonacci(100):
    pass # 0, 1, 1, 2, 3, 5...
````

A highly practical example: a lazy file reader that processes a file line-by-line without loading it all into RAM.


````python
import os
from typing import Generator

def read_large_file(file_path: str) -> Generator[str, None, None]:
    # Simulating a file read for the example
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            yield line.strip()
````

Generator expressions are the lazy equivalent of list comprehensions.


````python
# List comprehension (eager, allocates memory for all items)
squares_list = [x * x for x in range(1000)]

# Generator expression (lazy, O(1) memory)
squares_gen = (x * x for x in range(1000))
````

#### Common Mistakes and Pitfalls
Assuming generator expressions can be indexed or have a length.


````python
gen = (x for x in range(5))
# print(len(gen))  # TypeError: object of type 'generator' has no len()
# print(gen[0])    # TypeError: 'generator' object is not subscriptable

# FIX: You must iterate through it, or convert to a list if you need indexing (sacrificing memory efficiency).
````

#### Best Practices
- Always use generators for data pipelines, especially when processing large files, streams, or infinite sequences.
- Use generator expressions `(x for x in data)` instead of list comprehensions `[x for x in data]` if you are only going to iterate over the result once.
- Use `yield from iterable` to yield all items from a sub-iterable without a manual `for` loop.

#### Real-World Use Case
In web scraping using Scrapy, spiders are written as generator functions. A spider parses a webpage and `yield`s either a dictionary representing the extracted item or a `Request` object for the next page to scrape. The framework iterates over these yielded objects asynchronously.

#### Performance Notes
Generators shine in space complexity. Processing a 10GB log file with a generator requires O(1) memory because only one line is held in memory at a time. A list comprehension would attempt to load all 10GB into RAM, likely causing an OutOfMemory kill. Time complexity remains O(N), but avoiding memory allocation overhead often makes generators faster in practice.

#### Quick Recap
- A generator is a function containing the `yield` keyword.
- Calling a generator function returns a lazy generator object; it doesn't execute the function body.
- `yield` pauses execution and saves the local state (the frame).
- Generator expressions use parentheses `()` instead of brackets `[]` and are lazy.
- They provide massive memory efficiency for large datasets.

#### Interview Questions
- **Junior**: What is the difference between `yield` and `return`?
  **Answer**: `return` terminates a function entirely and destroys its local state. `yield` pauses the function, saves its local state (variables), and returns a value to the caller. The next time the generator is iterated, execution resumes immediately after the `yield`.
- **Mid**: Can a generator function contain both `yield` and `return`?
  **Answer**: Yes. A `return` statement inside a generator causes the generator to terminate immediately by raising `StopIteration`. If you `return value`, the value is attached to the `StopIteration` exception, which is how `yield from` captures return values from subgenerators.
- **Senior**: How does Python handle the execution frame of a generator under the hood?
  **Answer**: CPython represents execution frames as heap-allocated C structs. Normally, a frame is destroyed when a function returns. When a function has `yield`, CPython detaches the frame from the call stack and attaches it to the generator object. When `next()` is called, CPython links the frame back onto the stack, resumes execution, and detaches it again upon the next `yield`.


### Topic 2.9 — Decorators

#### Concept Overview
A decorator is a callable that takes a function as input and returns a function as output. Python added this syntax to enable clean, reusable modification of function behavior without altering the original function's internal logic. It solves the problem of cross-cutting concerns—such as logging, authentication, and caching—where the same wrapping logic applies to many different functions.

The `@` syntax is syntactic sugar. Writing `@decorator` above a function definition is exactly equivalent to defining the function and then reassigning its name: `func = decorator(func)`. By abstracting this, Python allows developers to wrap functions succinctly and clearly.

#### The Mind Picture
```text
+-------------------+
| Original Function |
|   func(args)      |
+-------------------+
         |
         v
+-------------------+
|   Decorator       |
|  decorator(func)  |
+-------------------+
         |
         v
+-------------------+
|  Wrapper Function |
|  wrapper(*args)   |
+-------------------+
```

#### Syntax Anatomy
```python
# The Decorator Definition
def my_decorator(func: Callable) -> Callable:
    @functools.wraps(func)           # 1. Preserves metadata
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        # 2. Pre-execution logic
        result = func(*args, **kwargs) # 3. Original execution
        # 4. Post-execution logic
        return result                  # 5. Return value
    return wrapper                     # 6. Return wrapper
```

#### Practical Examples
A simple timer decorator to measure execution speed.


````python
import time
from typing import Callable, Any
import functools

def timer(func: Callable[..., Any]) -> Callable[..., Any]:
    @functools.wraps(func)
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        start: float = time.perf_counter()
        result: Any = func(*args, **kwargs)
        end: float = time.perf_counter()
        print(f"{func.__name__} took {end - start:.4f}s")
        return result
    return wrapper

@timer
def slow_task(n: int) -> int:
    time.sleep(0.1)
    return n * 2

slow_task(5)
````

Using `functools.lru_cache` for memoization.


````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n: int) -> int:
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

# This runs instantly thanks to caching.
print(fibonacci(50))
````

Decorators with arguments require an extra layer.


````python
def repeat(times: int) -> Callable[[Callable[..., Any]], Callable[..., Any]]:
    def decorator(func: Callable[..., Any]) -> Callable[..., Any]:
        @functools.wraps(func)
        def wrapper(*args: Any, **kwargs: Any) -> Any:
            for _ in range(times - 1):
                func(*args, **kwargs)
            return func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(times=3)
def greet(name: str) -> None:
    print(f"Hello, {name}!")

greet("Alice")
````

#### Common Mistakes and Pitfalls
Forgetting `functools.wraps`.


````python
def bad_decorator(func: Callable[..., Any]) -> Callable[..., Any]:
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        return func(*args, **kwargs)
    return wrapper

@bad_decorator
def calculate_tax(amount: float) -> float:
    """Calculates the standard tax."""
    return amount * 0.2

# The name and docstring are lost, pointing to the wrapper instead.
print(calculate_tax.__name__)  # Prints 'wrapper'
print(calculate_tax.__doc__)   # Prints None
````

Stacking multiple decorators in the wrong order. Decorators apply bottom-up.


````python
def make_bold(func: Callable[..., str]) -> Callable[..., str]:
    @functools.wraps(func)
    def wrapper(*args: Any, **kwargs: Any) -> str:
        return f"<b>{func(*args, **kwargs)}</b>"
    return wrapper

def make_italic(func: Callable[..., str]) -> Callable[..., str]:
    @functools.wraps(func)
    def wrapper(*args: Any, **kwargs: Any) -> str:
        return f"<i>{func(*args, **kwargs)}</i>"
    return wrapper

@make_bold
@make_italic
def format_text(text: str) -> str:
    return text

# The bottom decorator runs first: italic, then bold.
print(format_text("Hello"))  # <b><i>Hello</i></b>
````

#### Best Practices
1. Always use `@functools.wraps` to preserve `__name__`, `__doc__`, and `__wrapped__`.
2. Keep decorator logic orthogonal to the business logic of the wrapped function.
3. Accept `*args` and `**kwargs` in the wrapper to guarantee signature compatibility.
4. Return the exact value produced by the wrapped function unless modifying it is the core goal.
5. Use class-based decorators for heavy state management to avoid deep nesting of closures.

#### Real-World Use Case
Flask uses decorators heavily to map URLs to view functions: `@app.route('/index')`. This registers the function in a routing table without changing the function's internal behavior.

#### Performance Notes
Applying a decorator adds a constant-time overhead for the wrapper function call. This cost is negligible for I/O bound operations but can accumulate in tightly bound CPU loops.

#### Quick Recap
* Decorators modify or wrap functions.
* Syntax `@decorator` equals `func = decorator(func)`.
* Argument-taking decorators need three levels of functions (factory, decorator, wrapper).
* Use `functools.wraps` unconditionally.
* Stacked decorators run bottom-to-top.

#### Interview Questions
* **Junior**: What does the `@` symbol do in Python?
  * *Answer*: It is syntactic sugar for passing a function into another callable and reassigning the name.
* **Mid**: How do you write a decorator that takes arguments?
  * *Answer*: You write a factory function that takes the arguments and returns the actual decorator function.
* **Senior**: Why does stacking decorators bottom-up make sense functionally?
  * *Answer*: Because `func = a(b(func))` evaluates the inner function call `b(func)` first, passing the result to `a()`.

### Topic 2.10 — Context Managers

#### Concept Overview
Context managers handle the setup and teardown of resources automatically. Python introduced the `with` statement to guarantee that specific cleanup operations execute even if an exception occurs inside the block. This solves the problem of dangling file handles, uncommitted database transactions, and unreleased locks caused by manual `try...finally` boilerplate.

#### The Mind Picture
```text
       with manager as var:
       |
       +--> __enter__() executes, returns value bound to 'var'
       |
       +--> [ Code block inside 'with' executes ]
       |
       +--> __exit__(exc_type, exc_val, exc_tb) executes
            (Handles exception or cleans up resource unconditionally)
```

#### Syntax Anatomy
```python
class MyContext:
    def __enter__(self) -> 'MyContext':
        # Resource acquisition
        return self # Bound to 'as' name

    def __exit__(self, exc_type: Any, exc_val: Any, exc_tb: Any) -> bool:
        # Resource cleanup
        if exc_type is not None:
            # Handle exception if desired
            return False # False propagates the exception
```

#### Practical Examples
Managing file I/O safely.


````python
from typing import Optional, Type, Any
from types import TracebackType

class ManagedFile:
    def __init__(self, filename: str) -> None:
        self.filename: str = filename
        self.file: Any = None
        
    def __enter__(self) -> Any:
        self.file = open(self.filename, 'w')
        return self.file
        
    def __exit__(self, exc_type: Optional[Type[BaseException]], 
                 exc_val: Optional[BaseException], 
                 exc_tb: Optional[TracebackType]) -> bool:
        if self.file:
            self.file.close()
        return False

with ManagedFile('test.txt') as f:
    f.write('Hello, Context Managers!')
````

Using `contextlib.contextmanager`.


````python
import contextlib
from typing import Iterator

@contextlib.contextmanager
def temporary_mode(mode: str) -> Iterator[str]:
    print(f"Setting mode to {mode}")
    yield mode
    print(f"Restoring original mode")

with temporary_mode("admin") as m:
    print(f"Doing work in {m} mode")
````

#### Common Mistakes and Pitfalls
Returning `True` from `__exit__` by mistake.


````python
class BadContext:
    def __enter__(self) -> None:
        pass
    def __exit__(self, exc_type: Any, exc_val: Any, exc_tb: Any) -> bool:
        return True # Suppresses all exceptions!

with BadContext():
    raise ValueError("This will be silently swallowed!")
````

#### Best Practices
1. Use `contextlib.contextmanager` for simple resource handling instead of writing a full class.
2. Never return `True` from `__exit__` unless you are explicitly building an exception-suppressing utility.
3. Keep `__enter__` logic minimal to avoid exceptions before the context is fully established.
4. Name your context variables clearly (`with open() as file:`).

#### Real-World Use Case
SQLAlchemy uses context managers for database sessions. `with session.begin():` automatically commits on success and rolls back on failure, abstracting away complex transaction boundaries.

#### Performance Notes
Generator-based context managers (`contextlib.contextmanager`) incur slight overhead due to generator initialization, but for I/O operations (like file handling or DB connections), this overhead is invisible.

#### Quick Recap
* Context managers guarantee cleanup.
* `__enter__` sets up and returns the variable bound by `as`.
* `__exit__` handles cleanup and receives exception details.
* `contextlib.contextmanager` splits enter and exit around a `yield` statement.

#### Interview Questions
* **Junior**: What is the purpose of the `with` keyword?
  * *Answer*: It ensures that resources are properly cleaned up, acting as a cleaner `try...finally` block.
* **Mid**: What does `yield` do in a `@contextmanager` function?
  * *Answer*: It pauses execution to let the `with` block run, yielding the value bound to the `as` variable, then resumes for cleanup.
* **Senior**: How do you suppress an exception using `__exit__`?
  * *Answer*: By returning `True`. If you return `False` or `None`, the exception continues to propagate.

### Topic 2.11 — Closures and nonlocal

#### Concept Overview
A closure is a function that retains access to variables from its enclosing lexical scope even after that scope has finished execution. Python implements this by creating cell objects that point to the bindings. `nonlocal` explicitly declares that a variable assignment should target an enclosing scope rather than creating a new local variable, which is critical for stateful closures.

#### The Mind Picture
```text
Outer Function Execution Scope
+------------------------------------+
| count = 0   <-- Cell Object        |
|                                    |
| Inner Function (Closure)           |
| +--------------------------------+ |
| | nonlocal count                 | |
| | count += 1                     | |
| | references Outer's Cell Object | |
| +--------------------------------+ |
+------------------------------------+
```

#### Syntax Anatomy
```python
def outer() -> Callable[[], int]:
    x = 10 # Enclosing variable
    def inner() -> int:
        nonlocal x # Binds 'x' to the outer scope
        x += 1
        return x
    return inner
```

#### Practical Examples
A stateful counter factory using closures and `nonlocal`.


````python
from typing import Callable

def make_counter() -> Callable[[], int]:
    count: int = 0
    def counter() -> int:
        nonlocal count
        count += 1
        return count
    return counter

c1 = make_counter()
print(c1()) # 1
print(c1()) # 2
````

#### Common Mistakes and Pitfalls
The classic loop-closure late-binding bug.


````python
from typing import List, Callable

# BUG: All lambdas capture the same loop variable binding, which ends up as 2.
funcs: List[Callable[[], int]] = [lambda: i for i in range(3)]
print([f() for f in funcs]) # [2, 2, 2]

# FIX: Force early binding using default arguments.
fixed_funcs: List[Callable[[], int]] = [lambda i=i: i for i in range(3)]
print([f() for f in fixed_funcs]) # [0, 1, 2]
````

Forgetting `nonlocal` causes `UnboundLocalError`.


````python
from typing import Callable
def broken_counter() -> Callable[[], int]:
    count = 0
    def counter() -> int:
        # count += 1 raises UnboundLocalError because it tries to read 'count' as a local variable before assignment.
        pass
    return counter
````

#### Best Practices
1. Use closures for simple state retention instead of defining a single-method class.
2. Use `nonlocal` carefully to avoid hidden side effects; if state becomes complex, use a class.
3. Be explicit about late vs early binding in loops by using default arguments.

#### Real-World Use Case
Callback systems often use closures to maintain state across asynchronous events without forcing the user to implement class interfaces.

#### Performance Notes
Closures are fast and have minimal memory footprint. However, each call to the outer function creates a new inner function object and new cell objects, which requires allocation.

#### Quick Recap
* Closures capture free variables from enclosing scopes.
* Variables are late-bound, capturing the cell, not the value at creation time.
* `nonlocal` permits assignment to enclosing variables.
* Default arguments solve the loop-closure binding issue.

#### Interview Questions
* **Junior**: What error do you get if you assign to an enclosing variable without `nonlocal`?
  * *Answer*: `UnboundLocalError`.
* **Mid**: Explain the loop-closure late-binding problem.
  * *Answer*: Closures capture the variable binding, not the value. In a loop, all closures reference the same variable, which holds the final loop value when executed.
* **Senior**: How does Python store closed-over variables internally?
  * *Answer*: Python stores them as cell objects in the function's `__closure__` tuple.

### Topic 2.12 — Custom Exceptions and Exception Chaining

#### Concept Overview
Custom exceptions allow developers to represent specific error states unique to their domain. By subclassing `Exception`, you can attach additional metadata to errors. Exception chaining explicitly connects a new exception to a previous one (`__cause__`) or implicitly connects them when an exception happens inside an `except` block (`__context__`), preserving the full traceback.

#### The Mind Picture
```text
try block fails -> OriginalException (A)
      |
except OriginalException as e:
      |
      +--> raise CustomException(B) from e
      
Traceback Shows:
OriginalException (A)
The above exception was the direct cause of the following exception:
CustomException (B)
```

#### Syntax Anatomy
```python
class MyError(Exception):
```

2. Create a base exception for your library or module to allow users to catch all related errors easily.

## Topic 2.13 — Regular Expressions

#### Concept Overview
Regular expressions (regex) are a micro-language for text pattern matching and extraction. Python implements them via the `re` module. Instead of chaining dozens of `.split()`, `.startswith()`, and `.replace()` string methods, regex allows you to define a declarative pattern that describes the exact shape of the data you want. This solves the problem of parsing semi-structured text like log files, messy user input, or serialized payloads where the format is predictable but not strictly enforced. The `re` module compiles patterns into state machines for execution, meaning complex string parsing is pushed down to highly optimized C code rather than looping in Python.

#### The Mind Picture
```text
State Machine Execution of pattern: r"A\d+B" against target "xA12B"

String:   x  A  1  2  B
Index:    0  1  2  3  4
          |  |  |  |  |
Start --->[FAIL]      |
             |        |
             +-> [Match A] -> [Match \d] -> [Match \d] -> [Match B] -> ACCEPT
```

#### Syntax Anatomy
```python
import re

# Raw string prevents Python from escaping the backslashes before regex engine sees them
pattern = re.compile(
    r"^(?P<year>\d{4})"  # ^ start anchor, (?P<name>...) named group, \d digit, {4} exact count
    r"-"                 # literal dash
    r"(?P<month>\d{2})"  # named group
    r"$",                # end anchor
    flags=re.IGNORECASE | re.MULTILINE  # bitwise OR flags
)
```


````python
import re
from typing import Match, Optional

def extract_domain(email: str) -> Optional[str]:
    # search finds the first match anywhere in the string
    match: Optional[Match[str]] = re.search(r"@([\w.-]+)", email)
    if match:
        return match.group(1)
    return None

print(extract_domain("user@example.com"))
````

#### Practical Examples

Finding all occurrences is a common task. `findall` returns strings, `finditer` returns Match objects which contain positional data.


````python
import re
from typing import Iterator, Match, List

def find_hex_colors(text: str) -> List[str]:
    # \b is a word boundary, (?:...) is a non-capturing group
    return re.findall(r"#(?:[0-9a-fA-F]{3}){1,2}\b", text)

def iterate_prices(text: str) -> None:
    # finditer is better when you need the exact string indices
    iterator: Iterator[Match[str]] = re.finditer(r"\$\d+\.\d{2}", text)
    for match in iterator:
        print(f"Found {match.group(0)} at chars {match.start()}-{match.end()}")

print(find_hex_colors("Colors: #F00, #ff00ff, and #12345Z"))
iterate_prices("Total: $14.99 and tax is $1.50.")
````

Named groups create a dictionary mapping group names to matched substrings.


````python
import re
from typing import Dict, Optional

def parse_log_line(line: str) -> Optional[Dict[str, str]]:
    pattern = re.compile(
        r"\[(?P<timestamp>.*?)\] "
        r"(?P<level>INFO|WARN|ERROR): "
        r"(?P<message>.*)"
    )
    match = pattern.match(line)
    if match:
        return match.groupdict()
    return None

print(parse_log_line("[2023-10-01 10:00:00] ERROR: Disk full"))
````

Replacement via `re.sub` is excellent for data sanitization.


````python
import re

def redact_phone_numbers(text: str) -> str:
    # Replace anything looking like a US phone number with [REDACTED]
    return re.sub(r"\b\d{3}[-.]\d{3}[-.]\d{4}\b", "[REDACTED]", text)

print(redact_phone_numbers("Call me at 555-123-4567 tomorrow."))
````

#### Common Mistakes and Pitfalls
Mistake: Forgetting to use raw strings for regex patterns.
```python
# Bad: \b is the backspace character in normal strings
re.findall("\bword\b", text) 
```
Fix: Always prefix regex patterns with `r`.
```python
# Good: \b is passed literally to the regex engine
re.findall(r"\bword\b", text)
```

Mistake: Using greedy quantifiers when non-greedy is needed.
```python
# Bad: .* matches everything up to the LAST quote
html = 'id="header" class="main"'
re.findall(r'id="(.*)"', html) # Returns ['header" class="main']
```
Fix: Add `?` to make the quantifier lazy (non-greedy).
```python
re.findall(r'id="(.*?)"', html) # Returns ['header']
```

#### Best Practices
1. Always use raw strings (`r"..."`) for regex patterns to prevent Python from consuming escape sequences.
2. Compile patterns with `re.compile()` if they are used inside loops or called frequently, to avoid redundant parsing overhead.
3. Prefer named capture groups `(?P<name>...)` over positional groups for complex patterns; they make the code self-documenting.
4. Use `re.VERBOSE` flag to split long patterns across multiple lines and add comments.
5. Bound patterns explicitly with `^` and `$` or `\A` and `\Z` when you expect a complete string match.

#### Real-World Use Case
Django's URL routing system originally used compiled regular expressions to map HTTP request paths to view functions. The named capture groups in the regex mapped directly to the keyword arguments passed to the Python view function. For example, `r"^articles/(?P<year>\d{4})/$"` would route paths and inject `year="2023"` into the handler.

#### Performance Notes
Python's `re` module uses a backtracking engine. It operates in O(N) time for simple patterns, but poorly constructed patterns (like `(a+)+b`) evaluated against pathological input can trigger catastrophic backtracking, leading to O(2^N) exponential time complexity. If processing untrusted input where ReDoS (Regular Expression Denial of Service) is a concern, avoid nested quantifiers or consider a third-party non-backtracking engine like `re2`.

#### Quick Recap
- `re.match` anchors at the start, `re.search` scans the whole string, `re.fullmatch` requires an exact match.
- `re.findall` returns a list of strings/tuples; `re.finditer` yields `Match` objects.
- Raw strings `r""` are mandatory to preserve backslashes.
- `*` and `+` are greedy; `*?` and `+?` are lazy.
- Named groups `(?P<name>pattern)` enable extracting sub-components into dictionaries.

#### Interview Questions
- **Junior**: What is the difference between `re.match` and `re.search`?
  *Answer*: `re.match` only attempts to match the pattern at the very beginning of the string. `re.search` scans through the string looking for the first location where the pattern produces a match.
- **Mid**: Why do we prefix regular expression strings with `r` in Python?
  *Answer*: The `r` prefix marks a raw string, telling Python not to interpret backslash escape sequences (like `\n` for newline or `\b` for backspace). Since regular expressions heavily use backslashes for their own syntax (like `\d` for digits), using raw strings prevents collisions and eliminates the need for double-escaping `\\d`.
- **Senior**: How would you optimize a regex pattern that is causing catastrophic backtracking on long inputs?
  *Answer*: I would analyze the pattern for nested quantifiers (e.g., `(.*)*` or `(a+)+`) or overlapping alternations. The fix usually involves replacing greedy quantifiers with lazy ones `.*?`, making quantifiers mutually exclusive, replacing `.*` with more restrictive character classes (like `[^"]*`), or using atomic groups (which Python's `re` module lacks natively, but can be simulated or accessed via the third-party `regex` module).


## Topic 2.14 — Dates and Times

#### Concept Overview
Managing time in software is inherently complex due to leap years, time zones, and daylight saving time transitions. Python's `datetime` module provides the foundational objects for dates, times, and intervals. The system differentiates between "naive" objects (which carry no timezone information and assume local time) and "aware" objects (which explicitly define their timezone). The introduction of `zoneinfo` in Python 3.9 standardized IANA timezone support, removing the need for third-party libraries like `pytz` for basic timezone operations and solving ambiguities related to UTC offsets.

#### The Mind Picture
```text
datetime.datetime
┌───────────────┬───────────────┬────────────────┐
│     Date      │     Time      │    Timezone    │
│ (Year/Mo/Day) │ (Hr/Min/Sec)  │ (tzinfo object)│
├───────────────┼───────────────┼────────────────┤
│  2023-10-31   │  14:30:00.123 │ America/New_York
└───────────────┴───────────────┴────────────────┘
     naive vs aware boundary ---> [If None, Naive]
```

#### Syntax Anatomy
```python
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

# Constructing an aware datetime
dt = datetime(
    year=2023, month=10, day=31, 
    hour=14, minute=30, 
    tzinfo=ZoneInfo("America/New_York")
)
```


````python
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

def get_current_time_utc() -> datetime:
    # now(timezone.utc) is the correct modern approach.
    # utcnow() is deprecated because it returns a naive datetime!
    return datetime.now(timezone.utc)

def convert_to_local(dt: datetime, tz_name: str) -> datetime:
    tz = ZoneInfo(tz_name)
    # astimezone converts aware dt to another timezone
    return dt.astimezone(tz)

utc_now = get_current_time_utc()
tokyo_time = convert_to_local(utc_now, "Asia/Tokyo")
print(f"UTC: {utc_now.isoformat()}")
print(f"Tokyo: {tokyo_time.isoformat()}")
````

#### Practical Examples

Parsing strings into datetimes and formatting datetimes back to strings is done using `strptime` and `strftime`.


````python
from datetime import datetime

def parse_and_format(date_str: str) -> str:
    # strptime: String Parse Time
    dt_obj: datetime = datetime.strptime(date_str, "%d/%m/%Y %H:%M:%S")
    
    # strftime: String Format Time
    return dt_obj.strftime("%Y-%m-%d %I:%M %p")

formatted = parse_and_format("31/10/2023 14:30:00")
print(f"Formatted output: {formatted}")
````

Calculating differences between dates uses `timedelta`.


````python
from datetime import datetime, timedelta, date

def calculate_deadline(start_date: date, days_to_add: int) -> date:
    # timedelta represents a duration
    duration = timedelta(days=days_to_add)
    return start_date + duration

today: date = date.today()
deadline: date = calculate_deadline(today, 14)
print(f"Deadline in 14 days: {deadline}")
````

#### Common Mistakes and Pitfalls
Mistake: Using `datetime.utcnow()` and assuming it's timezone-aware.
```python
# Bad: Returns a naive datetime representing UTC time.
# If you pass this to astimezone(), Python assumes it is currently LOCAL time, shifting it incorrectly.
now = datetime.utcnow()
```
Fix: Use `datetime.now(timezone.utc)`.
```python
from datetime import timezone
now = datetime.now(timezone.utc) # Returns an aware datetime
```

Mistake: Storing local time in a database without timezone information.
If you save `2023-11-05 01:30:00` during the Daylight Saving Time fallback in the US, that time happens twice. The exact instant is unrecoverable.
Fix: Always convert to UTC before storing.

#### Best Practices
1. Store all timestamps in databases as UTC. Only convert to local timezones when displaying to the user.
2. Never use `datetime.utcnow()`. Use `datetime.now(timezone.utc)`.
3. Use `zoneinfo` (Python 3.9+) instead of `pytz`. It is built-in and adheres to standard IANA timezone rules.
4. When comparing two datetime objects, ensure both are either naive or aware. Mixing them raises a `TypeError`.
5. For serialization, ISO 8601 string format (`.isoformat()`) is the universal standard.

#### Real-World Use Case
Celery, a distributed task queue for Python, relies heavily on aware datetimes for scheduling future tasks via the `eta` parameter. By ensuring all timestamps passed to the broker are UTC aware, Celery guarantees that a task scheduled by a worker in London to run at a specific UTC time executes correctly, even if processed by a worker in AWS us-east-1.

#### Performance Notes
`datetime` operations are very fast, primarily implemented in C. However, instantiating `ZoneInfo` parses timezone files from the OS. While caching mitigates this, parsing complex strings with `strptime` in tight loops can be a bottleneck. For heavy time series data parsing (millions of rows), use `pandas.to_datetime()` which vectorizes the operation.

#### Quick Recap
- `datetime` combines `date` (year, month, day) and `time` (hour, minute, second).
- `timedelta` represents duration and supports basic arithmetic with `datetime` objects.
- Naive datetimes lack timezone context; aware datetimes include it.
- `strftime` formats to string; `strptime` parses from string.
- `zoneinfo.ZoneInfo` manages IANA standard timezones securely.

#### Interview Questions
- **Junior**: What is a `timedelta` and how is it used?
  *Answer*: A `timedelta` represents a duration or difference between two dates or times. It is used to perform arithmetic, such as adding 5 days to a given `datetime` or calculating the time elapsed between two timestamps.
- **Mid**: Why should you never use `datetime.utcnow()`?
  *Answer*: `datetime.utcnow()` returns a naive datetime object representing the UTC time. Because it lacks tzinfo, if you attempt to convert it to another timezone using `.astimezone()`, Python treats the naive object as if it represents local system time, resulting in an incorrect offset. `datetime.now(timezone.utc)` returns a proper timezone-aware object.
- **Senior**: Explain the ambiguity that occurs during a Daylight Saving Time fallback and how an aware datetime handles it.
  *Answer*: During the DST fallback (e.g., clocks rolling back from 2 AM to 1 AM), the local hour between 1 AM and 2 AM occurs twice. A naive datetime of "1:30 AM" is strictly ambiguous. Aware datetimes using `ZoneInfo` resolve this by inspecting the `fold` attribute. A `fold` of 0 represents the first occurrence (pre-transition), and a `fold` of 1 represents the second occurrence (post-transition), allowing precise resolution to UTC.


## Topic 2.15 — Serialization: JSON, CSV, pickle

#### Concept Overview
Serialization is the process of translating data structures or object state into a format that can be stored or transmitted and reconstructed later. JSON (JavaScript Object Notation) is the ubiquitous text format for web APIs, standardizing strings, numbers, booleans, and nested dictionaries. CSV (Comma-Separated Values) remains the standard for flat, tabular data exchange. Python's `pickle` is a binary protocol specific to Python that can serialize nearly any arbitrary Python object, including custom classes and functions. Choosing the right format depends strictly on the data shape, language interoperability, and security context.

#### The Mind Picture
```text
Memory (Objects)            Serialization Engine           Disk/Network
[ {'id': 1} ]  ------------> json.dumps() -------------->  '[{"id": 1}]'
CustomClass()  ------------> pickle.dumps() ------------>  b'\x80\x04\x95...'
```

#### Syntax Anatomy
```python
import json

# dump/load operations interact with file objects
with open("data.json", "w") as f:
    json.dump({"key": "value"}, f, indent=4)

# dumps/loads (with an 's') operate directly on Python strings
json_string = json.dumps({"key": "value"})
```


````python
import json
from typing import Any, Dict

def serialize_payload(data: Dict[str, Any]) -> str:
    # dumps converts a Python dict to a JSON formatted string
    return json.dumps(data, separators=(',', ':'))

def deserialize_payload(payload: str) -> Dict[str, Any]:
    # loads parses the JSON string back into a Python dict
    return json.loads(payload)

raw_data = {"user": "alice", "admin": True, "roles": ["read", "write"]}
serialized = serialize_payload(raw_data)
print(f"Serialized: {serialized}")
print(f"Deserialized: {deserialize_payload(serialized)}")
````

#### Practical Examples

JSON does not natively support `datetime` objects. You must define a custom serializer using the `default` parameter.


````python
import json
from datetime import datetime, timezone
from typing import Any

def custom_serializer(obj: Any) -> Any:
    if isinstance(obj, datetime):
        return obj.isoformat()
    raise TypeError(f"Type {type(obj)} not serializable")

data = {"timestamp": datetime.now(timezone.utc), "event": "login"}
# Pass the function to the default argument
json_output = json.dumps(data, default=custom_serializer)
print(json_output)
````

CSV files are best handled using `DictReader` and `DictWriter` to map rows to dictionaries rather than relying on positional list indexes.


````python
import csv
import io
from typing import List, Dict

def write_and_read_csv(data: List[Dict[str, str]]) -> None:
    # Using StringIO to simulate a file for the example
    output = io.StringIO()
    
    # Write
    fieldnames = ["name", "age"]
    writer = csv.DictWriter(output, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)
    
    # Read
    output.seek(0)
    reader = csv.DictReader(output)
    for row in reader:
        print(f"Row read: {row}")

write_and_read_csv([{"name": "Alice", "age": "30"}, {"name": "Bob", "age": "25"}])
````

#### Common Mistakes and Pitfalls
Mistake: Unpickling data from an untrusted source.
`pickle` can execute arbitrary code during deserialization via the `__reduce__` method.
```python
# DANGEROUS: If untrusted_bytes contains a malicious payload, it will execute on your machine.
import pickle
data = pickle.loads(untrusted_bytes)
```
Fix: Never unpickle untrusted data. Use `json` for external data exchange.

Mistake: Attempting to serialize `bytes` or custom objects directly with `json`.
```python
import json
json.dumps({"data": b"raw_bytes"}) # TypeError: Object of type bytes is not JSON serializable
```
Fix: Encode bytes to Base64 strings or use a custom serializer.

#### Best Practices
1. Default to JSON for all external communications, APIs, and configuration files.
2. Use `csv.DictReader`/`DictWriter` instead of raw `csv.reader` to prevent breaking if column orders change.
3. Use `pickle` ONLY for caching complex Python objects locally or transmitting trusted data between internal nodes (e.g., multiprocessing).
4. When using `json.dump`, use `indent=4` for human-readable outputs and `separators=(',', ':')` for compact network payloads.

#### Real-World Use Case
Machine learning workflows frequently use `pickle` (or its faster drop-in replacement, `joblib`) to save trained model objects to disk. A trained scikit-learn model is a complex object hierarchy containing large NumPy arrays and state. Serializing this to JSON is virtually impossible without heavy custom logic, whereas `pickle.dump(model, file)` serializes the entire state tree perfectly for later inference loading.

#### Performance Notes
The standard library `json` module is fast enough for general use, but for high-throughput APIs dealing with massive payloads, third-party libraries like `orjson` or `ujson` drop down to C/Rust implementations offering 2x-5x speedups. `pickle` speed varies based on protocol level; always use the highest protocol available (`pickle.HIGHEST_PROTOCOL`) for optimal binary packing.

#### Quick Recap
- `json.dumps`/`loads` handles strings; `json.dump`/`load` handles files.
- JSON handles simple types but requires custom logic via `default` for types like `datetime`.
- `csv.DictReader` provides reliable tabular data parsing mapped to column headers.
- `pickle` serializes arbitrary Python objects to binary.
- Unpickling executes code; never use it on untrusted input.

#### Interview Questions
- **Junior**: What is the difference between `json.dump` and `json.dumps`?
  *Answer*: `json.dumps` (dump string) takes a Python object and returns a JSON-formatted string. `json.dump` takes a Python object and writes it directly to a file-like object.
- **Mid**: How do you serialize a Python `datetime` object to JSON?
  *Answer*: The standard JSON encoder does not know how to handle `datetime`. You must provide a custom function to the `default` keyword argument of `json.dumps` that checks if the object is a `datetime` instance and returns a string representation (usually `.isoformat()`).
- **Senior**: Why is `pickle` considered a massive security risk?
  *Answer*: The pickle protocol allows objects to define a `__reduce__` method that dictates how they should be reconstructed. A malicious payload can define `__reduce__` to return the `os.system` function along with a tuple of arguments (like a shell command). When `pickle.loads` evaluates the payload, it executes that command, resulting in Remote Code Execution (RCE).


## Topic 2.16 — Functional Tools

#### Concept Overview
Functional programming treats computation as the evaluation of mathematical functions, avoiding state changes and mutable data. Python is not a purely functional language, but it provides core functional primitives: `map`, `filter`, and `reduce`. These tools allow you to apply transformations and filters to iterables in a declarative style, abstracting away explicit `for` loops. While list comprehensions have largely superseded simple `map`/`filter` calls in modern Python, these functional tools remain highly effective for point-free programming and complex data pipelining when combined with the `operator` and `functools` modules.

#### The Mind Picture
```text
map:    [1, 2, 3] -> (f(x)) -> [f(1), f(2), f(3)]
filter: [1, 2, 3] -> (x>1)  -> [2, 3]
reduce: [1, 2, 3] -> (x+y)  -> (((1 + 2) + 3)) -> 6
```

#### Syntax Anatomy
```python
from functools import reduce

# map(function, iterable)
squares = map(lambda x: x**2, [1, 2, 3]) 

# filter(function, iterable)
evens = filter(lambda x: x % 2 == 0, [1, 2, 3])

# reduce(function, iterable, initializer)
total = reduce(lambda acc, x: acc + x, [1, 2, 3], 0)
```


````python
from typing import Iterator, List

def process_data(data: List[str]) -> Iterator[int]:
    # map returns a lazy iterator, not a list!
    lengths: Iterator[int] = map(len, data)
    return lengths

words = ["functional", "python", "mastery"]
print(list(process_data(words)))
````

#### Practical Examples

`filter` evaluates a predicate function against each item, yielding only items where the predicate returns True.


````python
from typing import Iterator, List

def get_positive_numbers(numbers: List[int]) -> Iterator[int]:
    # filter also returns a lazy iterator
    return filter(lambda x: x > 0, numbers)

nums = [-2, 5, -9, 10, 0]
print(list(get_positive_numbers(nums)))
````

The `operator` module provides fast C-implemented functions representing standard Python operators, avoiding the need for slow `lambda` definitions.


````python
from operator import itemgetter, attrgetter
from typing import List, Dict, Any

class User:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
    def __repr__(self) -> str:
        return f"User({self.name})"

users = [User("Alice", 30), User("Bob", 25)]
# attrgetter creates a fast callable for sorting
sorted_users = sorted(users, key=attrgetter('age'))

records: List[Dict[str, Any]] = [{"id": 2}, {"id": 1}]
# itemgetter works similarly for dictionary keys or tuple indexes
sorted_records = sorted(records, key=itemgetter('id'))

print(sorted_users)
print(sorted_records)
````

#### Common Mistakes and Pitfalls
Mistake: Assuming `map` or `filter` return lists in Python 3.
```python
# Bad: map returns an iterator, so printing it yields <map object at ...>
result = map(str.upper, ["a", "b"])
print(result) 
# Also, iterators exhaust. If you loop over it twice, the second loop does nothing.
```
Fix: explicitly cast to a list if you need to reuse the data or print it.
```python
result = list(map(str.upper, ["a", "b"]))
```

#### Best Practices
1. Prefer list comprehensions or generator expressions over `map` and `filter` when a `lambda` is required. `[x**2 for x in data if x > 0]` is more readable than `map(lambda x: x**2, filter(lambda x: x > 0, data))`.
2. Use `map` and `filter` when applying an already existing function (e.g., `map(int, string_list)`).
3. Use the `operator` module (`itemgetter`, `attrgetter`) instead of lambdas for sorting keys; it is executed in C and runs faster.
4. Always provide the `initializer` argument to `functools.reduce` to prevent `TypeError` on empty iterables.

#### Real-World Use Case
Data processing pipelines in frameworks like PySpark or Apache Beam heavily utilize the `map`/`reduce` paradigm. In standard Python, extracting a specific nested field from a list of deeply nested JSON API responses is efficiently handled using `map` paired with `operator.itemgetter`, cleanly isolating the extraction logic from the iteration.

#### Performance Notes
`map` and `filter` evaluate lazily, meaning they consume minimal memory regardless of the input iterable's size. They simply yield one item at a time. Applying a built-in function (like `len` or `int`) via `map` is actually slightly faster than a list comprehension because it avoids the bytecode execution overhead of the Python loop, but introducing a `lambda` destroys that performance advantage.

#### Quick Recap
- `map(func, iter)` applies a function to every item.
- `filter(func, iter)` yields items where `func(item)` is truthy.
- `functools.reduce(func, iter, init)` folds an iterable down to a single value.
- Both `map` and `filter` return lazy iterators in Python 3.
- `operator` functions (`itemgetter`, `attrgetter`) are faster, cleaner alternatives to lambdas for key extraction.

#### Interview Questions
- **Junior**: What is the difference between `map` and `filter`?
  *Answer*: `map` applies a specific transformation function to every item in an iterable, returning an iterator of the results. `filter` applies a boolean condition (predicate) to every item, returning an iterator of only the items that evaluated to True.
- **Mid**: Why do `map` and `filter` return iterators instead of lists in Python 3?
  *Answer*: Returning iterators allows for lazy evaluation. This means the transformations or filters are only computed one at a time as they are requested, drastically reducing memory usage when processing large datasets and allowing processing of infinite streams.
- **Senior**: When would you use `functools.reduce` over a simple `for` loop, and what is the importance of the initializer argument?
  *Answer*: `reduce` is useful for accumulating state across an iterable in a pure functional style, especially when passing the reduction logic as a first-class function to a higher-order pipeline. The initializer argument is critical because it defines the starting accumulator value. Without it, `reduce` uses the first element of the iterable as the starting point, which will raise a `TypeError` if the iterable is empty, and can cause logic errors if the accumulated type differs from the iterable's element type.


# Topic 2.17 — The collections Module

#### Concept Overview
Python's built-in `dict`, `list`, `set`, and `tuple` handle most basic data storage. However, they are general-purpose. The `collections` module provides specialized container datatypes. These specialized types solve specific problems like maintaining insertion order (historically), tracking element counts, grouping elements without missing-key errors, or managing memory efficiently for simple data structures. Python added these to prevent developers from writing slow, bug-prone custom classes for common data patterns.

#### The Mind Picture
```
Collections Memory Layout

list (O(N) pop(0))     deque (O(1) popleft)
[ A | B | C ]          ( A ) <-> ( B ) <-> ( C )
                      head                tail

dict (KeyError)        defaultdict (auto-factory)
{'a': 1}               {'a': 1} -> missing 'b'? 
                       call factory() -> insert 'b': factory()

tuple                  namedtuple
(1, 2)                 Point(x=1, y=2) -> accessible via .x and .y
```

#### Syntax Anatomy
```python
from collections import namedtuple, defaultdict, Counter, deque, OrderedDict, ChainMap

# namedtuple: typename, field_names
Point = namedtuple('Point', ['x', 'y'])

# defaultdict: default_factory
d = defaultdict(int) # int() returns 0

# Counter: iterable
c = Counter(['a', 'a', 'b'])

# deque: iterable, maxlen
q = deque([1, 2], maxlen=10)

# ChainMap: *maps
cm = ChainMap({'a': 1}, {'b': 2})
```


````python
from collections import namedtuple, defaultdict, Counter, deque, ChainMap
from typing import Any

# namedtuple for simple structs
Config = namedtuple('Config', ['host', 'port'])
cfg: Config = Config(host='localhost', port=8080)
print(f"Connecting to {cfg.host}:{cfg.port}")

# defaultdict for grouping
words: list[str] = ['apple', 'bat', 'cat', 'ant', 'bear']
grouped: defaultdict[str, list[str]] = defaultdict(list)
for word in words:
    grouped[word[0]].append(word)
print(dict(grouped))

# Counter for frequency and multiset operations
inventory: Counter[str] = Counter(['sword', 'shield', 'sword', 'potion'])
loot: Counter[str] = Counter(['sword', 'gold', 'gold'])
print(inventory + loot)  # Combine multisets
print(inventory.most_common(1))

# deque for sliding window
history: deque[int] = deque(maxlen=3)
for i in range(5):
    history.append(i)
    print(list(history))

# ChainMap for cascading lookups (e.g., config overrides)
default_config: dict[str, str] = {'theme': 'light', 'lang': 'en'}
user_config: dict[str, str] = {'theme': 'dark'}
active_config: ChainMap[str, str] = ChainMap(user_config, default_config)
print(active_config['theme']) # Looks in user_config first
````

#### Practical Examples
Moving from basic usage to a realistic problem: tracking active web requests per IP address over time. We use `defaultdict` to automatically initialize `deque` objects for each IP, keeping a fixed history of timestamps.


````python
import time
from collections import defaultdict, deque
from typing import DefaultDict, Deque

class RequestTracker:
    def __init__(self, history_size: int = 5) -> None:
        # Factory is a lambda returning a bound deque
        self.requests: DefaultDict[str, Deque[float]] = defaultdict(
            lambda: deque(maxlen=history_size)
        )
        
    def log_request(self, ip_address: str) -> None:
        self.requests[ip_address].append(time.time())
        
    def get_recent(self, ip_address: str) -> list[float]:
        return list(self.requests[ip_address])

tracker: RequestTracker = RequestTracker(history_size=2)
tracker.log_request("192.168.1.1")
tracker.log_request("192.168.1.1")
tracker.log_request("192.168.1.1")
# Only keeps the last 2 requests due to maxlen
print(tracker.get_recent("192.168.1.1"))
````

#### Common Mistakes and Pitfalls
Developers often misuse `defaultdict` by providing a value instead of a callable, or they mutate dictionaries while iterating when trying to manually count items instead of using `Counter`.


````python
from collections import defaultdict
from typing import DefaultDict

# BAD: Passing a value instead of a callable
# d = defaultdict(0)  # TypeError: first argument must be callable or None

# GOOD: Pass a type or a lambda
d_good: DefaultDict[str, int] = defaultdict(int)
d_good['missing'] += 1

# BAD: Using defaultdict for iteration if missing keys trigger creation unexpectedly
tracker: DefaultDict[str, int] = defaultdict(int)
tracker['a'] = 1
# Checking for 'b' creates it if you access it directly instead of using 'in'
if tracker['b'] > 0:
    pass
print(list(tracker.keys())) # ['a', 'b'] - 'b' was created by the check!

# FIX: Use 'in' for membership testing even on defaultdicts
if 'c' in tracker and tracker['c'] > 0:
    pass
print(list(tracker.keys())) # 'c' is not created
````

#### Best Practices
- Use `collections.abc` for type checking against interfaces (e.g., `isinstance(obj, collections.abc.Iterable)`) rather than concrete types.
- Prefer `dataclasses` over `namedtuple` for anything more complex than a simple data record, as dataclasses support mutability and custom methods easily.
- Use `Counter` for all counting logic instead of manual `dict.get(key, 0) + 1`.
- Specify `maxlen` on `deque` when you only care about the N most recent items to avoid memory leaks.
- Use `ChainMap` when layering configuration (CLI args > ENV vars > config file > defaults) to avoid copying and updating dictionaries.

#### Real-World Use Case
In web scraping or log parsing, `Counter` is heavily used to find the top occurring IP addresses, HTTP status codes, or error types. The `most_common()` method is highly optimized using heapq under the hood for finding the top N elements.


````python
from collections import Counter
import random

# Simulating 10,000 log entries
status_codes: list[int] = [random.choice([200, 404, 500, 403]) for _ in range(10000)]

# Analyze
counts: Counter[int] = Counter(status_codes)
top_two: list[tuple[int, int]] = counts.most_common(2)
print("Top 2 status codes:", top_two)
````

#### Performance Notes
- `deque.append()` and `deque.popleft()` are O(1).
- `list.insert(0, val)` and `list.pop(0)` are O(N) because all subsequent elements must be shifted in memory. Never use a `list` as a queue.
- `Counter` operations (addition, subtraction) are O(N) where N is the number of distinct keys.
- `ChainMap` lookups are O(K) where K is the number of mapped dictionaries, but updates are O(1) as they only affect the first dictionary.

#### Quick Recap
- `namedtuple`: Memory-efficient classes with named fields.
- `defaultdict`: Dictionary that calls a factory function for missing keys.
- `Counter`: Dictionary subclass for counting hashable objects, supports multiset math.
- `deque`: Double-ended queue, O(1) operations on both ends, optional fixed length.
- `OrderedDict`: Maintains insertion order. Mostly obsolete since Python 3.7 dictionaries do this natively, but retains `move_to_end()`.
- `ChainMap`: Logical view over multiple dictionaries, searched in order.

#### Interview Questions
- **Junior**: How do you sort a dictionary by its values in Python? 
  *Answer*: Use `sorted(d.items(), key=lambda item: item[1])`. Alternatively, load it into a `Counter` and use `most_common()`.
- **Mid**: Why use `deque` instead of a list for a queue?
  *Answer*: A list is a contiguous array. Removing from the front (`pop(0)`) requires shifting every other element in memory, making it O(N). A `deque` is implemented as a doubly-linked list of blocks, making operations at both ends O(1).
- **Senior**: Explain how `ChainMap` differs from `dict.update()`.
  *Answer*: `dict.update()` creates a single merged dictionary, destroying the separation of the originals and requiring O(N) time and memory to copy. `ChainMap` stores a list of references to the original dictionaries and routes lookups through them sequentially. It takes O(1) time to create and reflects underlying changes immediately.

# Topic 2.18 — Type Hints and the typing Module

#### Concept Overview
Python is fundamentally dynamically typed: variables point to objects, and objects have types, but variables do not enforce types at runtime. Type hints (PEP 484) introduce optional static typing. They serve as machine-readable documentation and allow tools like `mypy` or `pyright` to catch type errors before the code runs. They do not affect runtime execution speed or enforce runtime correctness.

#### The Mind Picture
```
Dynamic vs Static Analysis

Runtime (CPython): 
x: int = "hello"  --> Executes perfectly fine. x points to a string.

Static Analysis (mypy):
x: int = "hello"  --> ERROR: Incompatible types in assignment 
                    (expression has type "str", variable has type "int")
```

#### Syntax Anatomy
```python
from typing import Any, Callable, TypeVar

# Variable annotation
count: int = 10

# Function annotation
def process(data: str, flag: bool = False) -> list[str]:
    return data.split() if flag else []

# Built-in generics (Python 3.9+)
items: list[int] = [1, 2, 3]
mapping: dict[str, float] = {"pi": 3.14}

# Union (Python 3.10+)
result: int | None = None
```


````python
from typing import Callable, Any, TypeVar, Literal

T = TypeVar('T')

# Generic function
def get_first(items: list[T]) -> T | None:
    return items[0] if items else None

# Callable: takes two ints, returns an int
def apply_op(a: int, b: int, op: Callable[[int, int], int]) -> int:
    return op(a, b)

# Literal for specific restricted values
def set_mode(mode: Literal['read', 'write']) -> None:
    print(f"Mode set to {mode}")

print(get_first([1, 2, 3]))
print(apply_op(5, 3, lambda x, y: x + y))
set_mode('read')
````

#### Practical Examples
Type hints shine in complex domain modeling. Here we define a configuration parser that uses typed dictionaries and optionals to clearly declare what data is expected.


````python
from typing import TypedDict, Any
import json

class UserConfig(TypedDict, total=False):
    username: str
    email: str
    retries: int

def parse_config(raw_json: str) -> UserConfig | None:
    try:
        data: dict[str, Any] = json.loads(raw_json)
        # Type checking stops at the boundary; we assert the shape here
        config: UserConfig = {
            "username": str(data.get("username", "guest")),
            "email": str(data.get("email", ""))
        }
        if "retries" in data:
            config["retries"] = int(data["retries"])
        return config
    except (json.JSONDecodeError, ValueError):
        return None

result: UserConfig | None = parse_config('{"username": "admin", "retries": "5"}')
print(result)
````

#### Common Mistakes and Pitfalls
Developers often mix up `Any` and `object`, or use mutable default arguments incorrectly when trying to satisfy type checkers.


````python
from typing import Any

# BAD: Using Any defeats the purpose of type checking
def bad_process(data: Any) -> Any:
    return data.upper() # Static analyzer won't catch if data is an int

# GOOD: Use explicit types or TypeVar
def good_process(data: str) -> str:
    return data.upper()

# BAD: Confusing `object` with `Any`.
# `Any` turns off type checking. `object` is the base class of everything,
# but you can't call arbitrary methods on it safely.
def safe_print(item: object) -> None:
    print(item)
    # item.upper()  # Static analyzer correctly flags this as an error

safe_print(123)
````

#### Best Practices
- Annotate all function signatures (arguments and return types). This yields 90% of the value of type checking.
- Use `list[T]`, `dict[K, V]`, `set[T]` built-ins instead of importing `List`, `Dict`, `Set` from `typing` (Python 3.9+).
- Use `X | None` instead of `Optional[X]` (Python 3.10+).
- Avoid `Any` like the plague. If a function can take multiple types, use a Union (`int | str`) or generics (`TypeVar`).
- Use `mypy --strict` on new codebases to enforce rigorous typing from day one.

#### Real-World Use Case
Modern frameworks like FastAPI and Pydantic parse request bodies and validate JSON schemas entirely based on Python type hints. They inspect the type annotations at runtime using `typing.get_type_hints()` to coerce data (e.g., converting a JSON string to a Python `datetime` object).


````python
from dataclasses import dataclass
import typing
from typing import Any

@dataclass
class APIRequest:
    endpoint: str
    timeout: int

# Inspecting hints at runtime (framework behavior)
hints: dict[str, Any] = typing.get_type_hints(APIRequest)
print("Runtime hints:", hints)
````

#### Performance Notes
Type hints have almost zero runtime performance impact because the Python interpreter ignores them during execution. The only cost is a marginal memory overhead to store the annotations in the `__annotations__` dictionary of the function or class.

#### Quick Recap
- Python remains dynamically typed at runtime.
- Type hints enable static analysis, better IDE autocompletion, and machine-readable documentation.
- Built-in collections (`list`, `dict`) support generic indexing directly since Python 3.9.
- `X | Y` is the modern syntax for Unions (Python 3.10+).
- `Callable`, `TypeVar`, and `Literal` handle functions, generics, and specific values.

#### Interview Questions
- **Junior**: What does `def func(x: int) -> str:` mean, and what happens if you pass a float at runtime?
  *Answer*: It indicates the function expects an integer and returns a string. At runtime, passing a float executes normally without crashing (unless the code itself fails), because type hints are not enforced by the interpreter.
- **Mid**: What is the difference between `list` and `list[Any]` in type annotations?
  *Answer*: `list` (without brackets) is equivalent to `list[Any]` in a dynamically typed context, but strict type checkers often flag bare `list` as a missing generic type argument. `list[Any]` explicitly states "a list containing elements of any type," passing strict checks.
- **Senior**: How does `TypeVar` differ from `Union`?
  *Answer*: `Union[A, B]` means a variable can be exactly A or exactly B. `TypeVar('T')` binds a specific type across a function signature. For example, `def f(x: T) -> T` guarantees the return type exactly matches the argument type. `def f(x: int | str) -> int | str` allows passing an `int` and returning a `str`, which is a weaker constraint.

# Topic 2.19 — Testing Fundamentals

#### Concept Overview
Testing guarantees code behaves as expected and prevents regressions when modifying systems. Python has the built-in `unittest` framework (inspired by Java's JUnit) and a massive third-party ecosystem led by `pytest`. Both execute tests, but they differ radically in philosophy. `unittest` relies on class inheritance and explicit assertion methods. `pytest` uses plain functions and overrides the `assert` statement via AST manipulation to provide deep introspection on failure.

#### The Mind Picture
```
Test Execution Flow

      Setup (Database, mocks)
         |
      Execution (Call function)
         |
      Assertion (Verify result)
         |
      Teardown (Clean up)
```

#### Syntax Anatomy
```python
import unittest

class TestMath(unittest.TestCase):
    def setUp(self) -> None:
        self.value: int = 5

    def test_addition(self) -> None:
        self.assertEqual(self.value + 5, 10)
        self.assertTrue(True)
        with self.assertRaises(ZeroDivisionError):
            _ = 1 / 0
```


````python
import unittest

# Example of a function we want to test
def divide(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

# The test suite
class TestDivide(unittest.TestCase):
    def test_normal_division(self) -> None:
        self.assertEqual(divide(10, 2), 5.0)
        self.assertAlmostEqual(divide(10, 3), 3.3333333, places=5)

    def test_divide_by_zero(self) -> None:
        with self.assertRaises(ValueError) as context:
            divide(10, 0)
        self.assertIn("zero", str(context.exception))

# In a real file, we run this via: if __name__ == '__main__': unittest.main()
# For the notebook, we run it manually to show output without crashing the kernel:
suite = unittest.TestLoader().loadTestsFromTestCase(TestDivide)
unittest.TextTestRunner(verbosity=2).run(suite)
````

#### Practical Examples
While `unittest` is built-in, `pytest` is the industry standard due to its brevity. With `pytest`, you write simple functions and use the standard `assert` keyword. Pytest inspects the `assert` expression to provide detailed breakdown on failure.


````python
# Pytest syntax example (conceptual, as pytest is a CLI tool)

def calculate_discount(price: float, discount: float) -> float:
    return price * (1 - discount)

def test_calculate_discount() -> None:
    # No self.assertEqual, just pure assert
    assert calculate_discount(100.0, 0.2) == 80.0
    assert calculate_discount(50.0, 0.1) == 45.0

# Pytest fixtures provide dependency injection
# import pytest
# @pytest.fixture
# def db_connection() -> str:
#     return "connected"
#
# def test_db(db_connection: str) -> None:
#     assert db_connection == "connected"

# Running this cell just executes the function directly to prove it works
test_calculate_discount()
print("Pytest-style test passed manually.")
````

#### Common Mistakes and Pitfalls
Using standard `assert` for production data validation is a critical error. The `assert` statement can be globally disabled by running Python with the `-O` (optimize) flag. Tests sharing state is another major pitfall causing flaky test suites.


````python
# BAD: Using assert for business logic validation
def process_payment(amount: float) -> None:
    # If python is run with -O, this check is entirely skipped!
    assert amount > 0, "Amount must be positive"
    print(f"Processing ${amount}")

# GOOD: Explicitly raise exceptions for business logic
def process_payment_safe(amount: float) -> None:
    if amount <= 0:
        raise ValueError("Amount must be positive")
    print(f"Processing ${amount}")

process_payment_safe(50.0)
````

#### Best Practices
- **Test Isolation**: A test should never depend on the execution order of other tests. Clean up all state.
- **Arrange, Act, Assert**: Structure tests into three blocks: set up the data, call the function, assert the outcome.
- Never use `assert` in production code for data validation or control flow. Use it strictly for internal invariants and testing.
- Use `pytest.mark.parametrize` to run the same test logic against multiple sets of inputs.
- Aim to test behavior, not implementation details. Refactoring internal logic shouldn't break tests if the public API behaves the same.

#### Real-World Use Case
In continuous integration (CI) pipelines (like GitHub Actions), `pytest` is executed on every commit. Projects use coverage tools (`pytest-cov`) to ensure that at least 80-90% of the codebase is executed by tests. Parametrized testing is widely used to ensure edge cases (nulls, empty strings, negative numbers) are verified without duplicating test code.


````python
# Demonstrating the logic of parameterized tests manually

def is_even(n: int) -> bool:
    return n % 2 == 0

test_cases: list[tuple[int, bool]] = [
    (2, True),
    (3, False),
    (0, True),
    (-4, True)
]

for input_val, expected in test_cases:
    assert is_even(input_val) == expected, f"Failed on {input_val}"
print("All parameterized cases passed.")
````

#### Performance Notes
Test execution time matters heavily in large projects. Slow tests disrupt developer workflow. I/O operations (network, disk, database) make tests slow. Mocking (`unittest.mock.patch`) is used to intercept I/O calls and return static data instantly, keeping unit tests executing in milliseconds.

#### Quick Recap
- `unittest` is built-in, uses classes, and requires specific `self.assertX` methods.
- `pytest` is third-party, uses plain functions, pure `assert` statements, and fixtures.
- `assert` statements are ignored when Python runs with `-O`. Never use them for app security.
- Good tests are isolated, reproducible, and run quickly.
- Use parameterization to test multiple inputs on the same logic path.

#### Interview Questions
- **Junior**: Why shouldn't you use `assert` statements to validate user input in a web application?
  *Answer*: Because running the Python interpreter with the `-O` optimization flag strips all `assert` statements from the bytecode, entirely removing the validation and exposing the app to bad data or security flaws.
- **Mid**: How do `unittest.setUp` and `pytest` fixtures differ?
  *Answer*: `setUp` is an object-oriented approach running before every test method in a class, often creating state attached to `self`. `pytest` fixtures use dependency injection: you define a fixture function and request it as an argument in your test function. Fixtures are more modular, reusable across different files, and support sophisticated scoping (function, module, session).

  *Answer*: `pytest` hooks into Python's import mechanism and rewrites the Abstract Syntax Tree (AST) of test files before they are compiled to bytecode. It intercepts `assert` statements and replaces them with a custom function call that stores the intermediate values of the expression, allowing it to print the exact state of variables upon failure.

# Topic 2.20 — Packages in Depth

#### Concept Overview
A Python module is a single `.py` file. A Python package is a directory containing modules and a special `__init__.py` file. Packages allow hierarchical structuring of the module namespace. As projects grow, dropping everything into one directory causes name collisions and massive files. Packages solve this by providing namespaces (e.g., `app.database.models`).

#### The Mind Picture
```
Directory Structure           Python Import Path

my_project/                   
├── main.py                   import my_package
└── my_package/               from my_package import utils
    ├── __init__.py           (Executes __init__.py on first import)
    ├── utils.py              
    └── data/                 
        ├── __init__.py       from my_package.data import loader
        └── loader.py         
```

#### Syntax Anatomy
```python
# Absolute import (resolves from project root / sys.path)
from my_package.data.loader import load_csv
import my_package.utils

# Relative import (resolves relative to current module's position)
# ONLY works inside a package, NEVER if the script is run directly as __main__
from . import utils          # Import sibling module
from ..data import loader    # Import from parent package

# In __init__.py, controlling the public API:
__all__ = ['load_csv', 'utils']
```

#### Practical Examples
The `__init__.py` file can be completely empty, serving only to mark the directory as a package. Alternatively, it can be used to lift classes or functions up the namespace hierarchy, making the public API cleaner for users.

#### Common Mistakes and Pitfalls
The most common error in Python packages is `ImportError: attempted relative import with no known parent package`. This happens when you run a submodule directly (e.g., `python my_package/utils.py`). Python sets the module name to `__main__` and forgets its place in the package hierarchy, breaking any relative imports (like `from . import core`).


````python
# BAD: Running a file with relative imports directly
# If this file contains `from . import config` and you run `python this_file.py`
# It crashes with ImportError.

# GOOD: Run the file as a module from the project root
# `python -m my_package.utils`
# This tells Python to resolve the package hierarchy properly before execution.

def demonstrate_execution_context() -> None:
    print(f"My context is: {__name__}")

demonstrate_execution_context()
# When run directly, __name__ is '__main__'.
# When imported, __name__ is 'my_package.module_name'.
````

#### Best Practices
- Keep `__init__.py` files mostly empty. Use them strictly for lifting imports to simplify APIs, not for heavy logic or initialization code.
- Use absolute imports over relative imports. Absolute imports (`from my_package.data import loader`) are easier to read, grep, and refactor than relative ones (`from ..data import loader`).
- Define `__all__` in modules to explicitly document the public API and control `import *` behavior (though `import *` should be avoided regardless).
- Run internal package scripts using `python -m package.module` from the project root to ensure `sys.path` is correct.

- A package is a directory with an `__init__.py` file, creating a namespace.

### Topic 3.1 — Memory Management

#### Concept Overview
CPython manages memory primarily through reference counting. Every object allocated in CPython has an internal counter, `ob_refcnt`, that tracks how many references point to it. When a new reference to an object is created, the count increments. When a reference goes out of scope or is reassigned, the count decrements. Once the reference count hits exactly zero, CPython deallocates the object's memory immediately. This deterministic deallocation differs drastically from tracing garbage collectors in runtimes like the JVM or V8, where memory is reclaimed in unpredictable batches.

Reference counting provides immediate reclamation of memory but has a critical limitation: reference cycles. If object A references object B, and object B references object A, their reference counts will never hit zero, even if neither is accessible from the rest of the program. To solve this, CPython includes a secondary mechanism: the cyclic garbage collector (the `gc` module). This collector runs periodically to detect and clean up isolated cyclic graphs. Additionally, CPython applies specific optimizations to reduce memory overhead and allocation time, such as interning small integers and strings, and using `__slots__` to prevent the creation of instance dictionaries.


#### The Mind Picture
```text
[ Stack Frame ]               [ Heap Memory ]
    a ---------------------> +-------------------+
                             | ob_refcnt: 2      |
    b ---------------------> | ob_type: int      |
                             | ob_ival: 42       |
                             +-------------------+
                                      
                             +-------------------+
    cycle_a ---------------> | ob_refcnt: 1      | <---+
                             | next: cycle_b     | ----+
                             +-------------------+     |
                                                       |
                             +-------------------+     |
    cycle_b ---------------> | ob_refcnt: 1      | ----+
                             | next: cycle_a     | <---+
                             +-------------------+
```


#### Syntax Anatomy
```python
import sys
import gc

# 1. Inspecting reference counts
count: int = sys.getrefcount(obj)  # Returns actual references + 1 (the argument itself)

# 2. Forcing cyclic garbage collection
collected: int = gc.collect()      # Returns the number of unreachable objects found

# 3. Suppressing instance dictionaries for memory savings
class CompactClass:
    __slots__: list[str] = ['attr1', 'attr2']
```


#### Practical Examples

First, we observe basic reference counting and immediate deallocation.


````python
import sys

class Tracker:
    def __del__(self) -> None:
        print("Tracker instance destroyed")

def basic_refcount() -> None:
    t = Tracker()
    # getrefcount returns count + 1
    print(f"Initial count: {sys.getrefcount(t)}")
    
    t_ref = t
    print(f"Count after second ref: {sys.getrefcount(t)}")
    
    del t_ref
    print(f"Count after deleting second ref: {sys.getrefcount(t)}")

basic_refcount()
# Tracker instance destroyed prints immediately after function exit
````

Next, we intentionally create a reference cycle and use the `gc` module to clean it up.


````python
import gc

class Node:
    def __init__(self, name: str) -> None:
        self.name: str = name
        self.child: 'Node' | None = None
    
    def __del__(self) -> None:
        print(f"Node {self.name} destroyed")

def reference_cycle() -> None:
    gc.disable()  # Disable automatic collection for demonstration
    a = Node("A")
    b = Node("B")
    
    a.child = b
    b.child = a
    
    # Removing variables from local scope
    del a
    del b
    print("Variables deleted, but nodes not destroyed yet due to cycle.")
    
    collected: int = gc.collect()
    print(f"Garbage collector freed {collected} objects.")
    gc.enable()

reference_cycle()
````

We can use `__slots__` to eliminate the overhead of the per-instance `__dict__`.


````python
import sys

class DictPoint:
    def __init__(self, x: float, y: float) -> None:
        self.x: float = x
        self.y: float = y

class SlotPoint:
    __slots__: list[str] = ['x', 'y']
    def __init__(self, x: float, y: float) -> None:
        self.x: float = x
        self.y: float = y

p1 = DictPoint(1.0, 2.0)
p2 = SlotPoint(1.0, 2.0)

# sys.getsizeof only measures the instance object structure, not the dict contents
# but the difference is stark because SlotPoint has no __dict__ pointer at all.
print(f"Size of DictPoint: {sys.getsizeof(p1)} bytes")
print(f"Size of SlotPoint: {sys.getsizeof(p2)} bytes")
````

Finally, observe CPython's small integer cache (numbers -5 to 256) and string interning.


````python
def memory_optimizations() -> None:
    # Small integers are cached
    a: int = 100
    b: int = 100
    print(f"100 is 100: {a is b}")
    
    # Large integers are not cached (usually)
    x: int = 1000
    y: int = 1000
    print(f"1000 is 1000: {x is y}")
    
    # Small literal strings are interned
    s1: str = "hello"
    s2: str = "hello"
    print(f"'hello' is 'hello': {s1 is s2}")

memory_optimizations()
````

#### Common Mistakes and Pitfalls

Mistake 1: Misinterpreting `sys.getrefcount()`. Developers often assume it returns the exact number of external references.


````python
import sys

def check_count() -> None:
    my_list: list[int] = [1, 2, 3]
    # Mistake: Expecting 1 here
    print(f"Count: {sys.getrefcount(my_list)}")  # Prints 2

check_count()
````

Mistake 2: Accidentally pinning memory globally. Storing objects in a class-level dictionary or global list prevents their reference count from ever reaching zero.


````python
# BROKEN: Unbounded cache
class DataProcessor:
    _cache: dict[str, dict] = {}
    
    def process(self, key: str, payload: dict) -> None:
        # We process the payload, but inadvertently keep a reference forever
        self._cache[key] = payload

# FIX: Use LRU cache or explicit eviction
from functools import lru_cache

@lru_cache(maxsize=1000)
def process_data(key: str) -> dict:
    return {"data": "processed"}
````

#### Best Practices
- Define `__slots__` strictly for classes instantiated thousands or millions of times (e.g., ORM models, data transfer objects).
- Do not manually invoke `gc.collect()` in typical application code. Let the runtime handle it unless clearing massive reference cycles after a distinct phase of processing.
- Avoid module-level lists or dicts acting as unbounded caches. Use `functools.lru_cache` which caps size and allows old items to be deallocated.
- Depend on local variable scoping. When a function returns, its stack frame is destroyed, automatically decrementing the reference counts of all local variables.

#### Real-World Use Case
In data engineering pipelines processing large CSVs, representing each row as an object can consume gigabytes of RAM. Libraries like SQLAlchemy implement `__slots__` on core structural objects to prevent massive memory overhead from `__dict__` pointers when materializing hundreds of thousands of query results.

#### Performance Notes
Reference counting imposes an overhead on every assignment, function call, and return. The CPU must constantly increment and decrement counters. However, `__slots__` provides a tangible performance benefit by storing instance attributes in a C array rather than a hash map, resulting in faster attribute access and substantially lower memory footprint.

#### Quick Recap
- Every CPython object has an `ob_refcnt`. Zero references means instant deallocation.
- `sys.getrefcount()` includes the temporary reference passed to the function itself.
- Reference cycles prevent refcounts from reaching zero. The `gc` module periodically sweeps these.
- `__slots__` prevents the creation of `__dict__`, saving significant memory per instance.
- CPython caches integers from -5 to 256 and small string literals to avoid re-allocation overhead.

#### Interview Questions
1. **Junior**: Why does `sys.getrefcount(obj)` return a number higher than the expected references?
   *Answer*: Passing the object as an argument into the `getrefcount` function creates an additional local reference on the C call stack. The function counts this temporary reference.
2. **Mid**: When would you explicitly call `gc.collect()`?
   *Answer*: Almost never in normal execution, as the GC triggers automatically based on allocation thresholds. However, in a batch-processing script where you create massive cyclic graphs, process them, and move on, calling `gc.collect()` manually after each batch can prevent the GC from interrupting during a time-sensitive phase.
3. **Senior**: Explain the memory difference between a standard Python class and one with `__slots__`.
   *Answer*: A standard class instance maintains a dynamic dictionary (`__dict__`) to map attribute names to values, allowing arbitrary attributes to be added at runtime. This dictionary has base overhead and hash-table sparseness. A class using `__slots__` allocates a fixed-size C array for exactly the specified attributes, eliminating the dictionary pointer and the hash table entirely.


### Topic 3.2 — The GIL

#### Concept Overview
The Global Interpreter Lock (GIL) is a mutex native to CPython that mandates only one thread can execute Python bytecode at any given time. CPython's memory management relies entirely on reference counting. If two threads incremented an object's reference count simultaneously without a lock, a race condition could leave the count permanently incorrect, causing memory leaks or premature deallocations (segmentation faults). To secure the reference counters, the core developers introduced a single, global lock.

The GIL means CPython threads cannot execute CPU-bound work in parallel on multiple cores. However, the GIL is released during I/O operations (like writing to a socket or reading from a file) and during certain underlying C library calls. Therefore, multithreading is still highly effective for I/O-bound tasks. Python 3.13 introduces an experimental free-threaded build (`--disable-gil`) that removes the GIL, utilizing fine-grained locking and specialized memory allocation to achieve true multithreading.


#### The Mind Picture
```text
[ Core 1 ]           [ Core 2 ]           [ Core 3 ]
Thread 1 (Has GIL)   Thread 2 (Waiting)   Thread 3 (I/O Wait, No GIL needed)
  |                    .                    |
[Executes Bytecode]  [Blocks on Lock]     [Waits for Network]
  |                    .                    |
[Releases GIL] ----> [Acquires GIL]       |
  .                  [Executes Bytecode]  |
```


#### Syntax Anatomy
```python
import sys

# Checking the GIL switch interval
# How long (in seconds) a thread runs before the interpreter forces it to drop the GIL
interval: float = sys.getswitchinterval()  # default is 0.005 (5ms)

# Changing the switch interval (rarely needed)
sys.setswitchinterval(0.010)
```


#### Practical Examples

First, we observe the GIL's impact on CPU-bound work.


````python
import time
import threading
import sys

def cpu_bound_task(n: int) -> int:
    count: int = 0
    for i in range(n):
        count += i
    return count

def run_single() -> None:
    start = time.time()
    cpu_bound_task(20_000_000)
    cpu_bound_task(20_000_000)
    print(f"Sequential: {time.time() - start:.2f}s")

def run_threaded() -> None:
    start = time.time()
    t1 = threading.Thread(target=cpu_bound_task, args=(20_000_000,))
    t2 = threading.Thread(target=cpu_bound_task, args=(20_000_000,))
    
    t1.start()
    t2.start()
    t1.join()
    t2.join()
    print(f"Threaded: {time.time() - start:.2f}s")

# In CPython, threaded is often slightly slower due to GIL contention overhead
run_single()
run_threaded()
````

Next, we observe how threads release the GIL during I/O, allowing concurrency.


````python
import time
import threading

def io_bound_task() -> None:
    # time.sleep releases the GIL, simulating network/disk I/O
    time.sleep(1)

def io_single() -> None:
    start = time.time()
    io_bound_task()
    io_bound_task()
    print(f"Sequential I/O: {time.time() - start:.2f}s")

def io_threaded() -> None:
    start = time.time()
    t1 = threading.Thread(target=io_bound_task)
    t2 = threading.Thread(target=io_bound_task)
    
    t1.start()
    t2.start()
    t1.join()
    t2.join()
    print(f"Threaded I/O: {time.time() - start:.2f}s")

# Threaded will be almost exactly 2x faster
io_single()
io_threaded()
````

#### Common Mistakes and Pitfalls

Mistake 1: Spawning threads for data processing pipelines and wondering why throughput doesn't improve.


````python
import threading
import hashlib

data_chunks: list[bytes] = [b"data" * 1000] * 4

# BROKEN EXPECTATION: Thinking threading speeds up hashing
def hash_data(chunk: bytes) -> None:
    hashlib.sha256(chunk).hexdigest()

threads: list[threading.Thread] = []
for chunk in data_chunks:
    t = threading.Thread(target=hash_data, args=(chunk,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
````

Mistake 2: Assuming that because the GIL exists, you don't need locks for shared state. The GIL protects the interpreter's internal state, not your application's state. Two threads can still preempt each other halfway through an application-level transaction. (More on this in Topic 3.3).


#### Best Practices
- Never use threading to attempt to parallelize CPU-bound calculations (number crunching, image processing, complex parsing) in standard CPython.
- Use threading explicitly for I/O bound bottlenecks: hitting external APIs, querying databases, and reading from files.
- Remember that C extensions (like NumPy) can release the GIL internally when doing heavy computations. In those specific cases, multithreading can speed up CPU work because the GIL is dropped.
- Adjust `sys.setswitchinterval()` only if you strictly understand the latency requirements of your threads, though the default of 5 milliseconds is highly tuned for general workloads.

#### Real-World Use Case
In a web scraping application, an architecture using 100 threads can drastically improve scraping speed. Because `requests.get()` spends 99% of its time waiting for the remote server's TCP response, it drops the GIL. The CPU then freely processes responses from other threads.

#### Performance Notes
Because the GIL necessitates thread switching, CPU-bound multithreaded code often performs worse than sequential code due to context-switching overhead. The OS rapidly suspends and resumes threads competing for a single lock, degrading CPU cache performance.

#### Quick Recap
- The GIL is a single mutex protecting CPython's reference counts from race conditions.
- Only one thread executes Python bytecode at a time.
- I/O operations inherently drop the GIL.
- CPU-bound multithreading in CPython is anti-pattern; it adds overhead without parallel execution.
- Python 3.13 introduces a compile-time option to remove the GIL entirely.

#### Interview Questions
1. **Junior**: What is the GIL and what problem does it solve?
   *Answer*: The Global Interpreter Lock is a lock in CPython that ensures only one thread executes Python bytecode at once. It solves the problem of thread-safe memory management, preventing race conditions on object reference counts.
2. **Mid**: Can multiple threads ever execute at the same time in CPython?
   *Answer*: Python bytecode cannot be executed simultaneously. However, if a thread is waiting on I/O (like a network request) or executing a C extension that manually releases the GIL (like NumPy matrix multiplication), other Python threads can execute concurrently.
3. **Senior**: How does the GIL relate to the `sys.getswitchinterval()`?
   *Answer*: The switch interval dictates how long a CPU-bound thread is permitted to hold the GIL before the interpreter forces it to drop the lock, giving other threads a chance to run. If set too high, other threads starve. If set too low, context switching overhead degrades overall performance.


### Topic 3.3 — Concurrency with threading

#### Concept Overview
The `threading` module is CPython's primary interface for OS-level threads. Threads operate concurrently within the same memory space, making it easy to share state but hazardous if not synchronized properly. You spawn a thread by pointing it at a target function and starting it. It runs independently until completion.

When threads modify shared mutable state, race conditions emerge because operations like `count += 1` are not atomic (they require reading the value, modifying it, and writing it back). The interpreter can interrupt a thread between these bytecode steps. To prevent corruption, `threading` provides synchronization primitives: `Lock` for mutual exclusion, `RLock` for reentrant locks, `Event` for boolean signaling between threads, and `Semaphore` for bounding concurrency. `concurrent.futures.ThreadPoolExecutor` provides a higher-level abstraction for managing worker pools.


#### The Mind Picture
```text
[ Main Thread ]
      |
      |--- thread.start() ---> [ Worker Thread ]
      |                              |
      |   [ Shared Memory Box ]      |
      |   | state_var = 1     | <----+ (Lock acquired)
      |                              |
      |<------ thread.join() --------+ (Thread terminates)
      |
```


#### Syntax Anatomy
```python
import threading

# 1. Spawning a thread
t = threading.Thread(target=worker_func, args=(arg1,), daemon=True)
t.start()  # Initiates execution
t.join()   # Blocks caller until thread terminates

# 2. Mutex Lock
lock = threading.Lock()
with lock:
    # Critical section

# 3. Thread Pool
from concurrent.futures import ThreadPoolExecutor
with ThreadPoolExecutor(max_workers=5) as executor:
    future = executor.submit(func, arg)
```


#### Practical Examples

First, thread creation and execution lifecycle.


````python
import threading
import time

def worker(worker_id: int) -> None:
    print(f"Worker {worker_id} starting")
    time.sleep(0.1)
    print(f"Worker {worker_id} done")

def main_lifecycle() -> None:
    t: threading.Thread = threading.Thread(target=worker, args=(1,))
    t.start()
    print("Main thread waiting for worker")
    t.join()
    print("Main thread complete")

main_lifecycle()
````

Next, a demonstration of a race condition and how to fix it using `threading.Lock`.


````python
import threading

class Counter:
    def __init__(self) -> None:
        self.value: int = 0
        self.lock: threading.Lock = threading.Lock()
        
    def unsafe_increment(self) -> None:
        # Prone to race conditions
        temp = self.value
        temp += 1
        self.value = temp
        
    def safe_increment(self) -> None:
        # Atomic with respect to other threads using this lock
        with self.lock:
            temp = self.value
            temp += 1
            self.value = temp

def run_race_condition() -> None:
    counter = Counter()
    
    def worker_unsafe() -> None:
        for _ in range(100_000):
            counter.unsafe_increment()
            
    def worker_safe() -> None:
        for _ in range(100_000):
            counter.safe_increment()

    # Using unsafe increment
    threads = [threading.Thread(target=worker_unsafe) for _ in range(5)]
    for t in threads: t.start()
    for t in threads: t.join()
    print(f"Unsafe count (expected 500000): {counter.value}")

    # Reset and use safe increment
    counter.value = 0
    threads = [threading.Thread(target=worker_safe) for _ in range(5)]
    for t in threads: t.start()
    for t in threads: t.join()
    print(f"Safe count (expected 500000): {counter.value}")

run_race_condition()
````

Using `ThreadPoolExecutor` for cleaner orchestration of multiple tasks.


````python
from concurrent.futures import ThreadPoolExecutor, as_completed
import time

def fetch_data(item_id: int) -> str:
    time.sleep(0.1)  # Simulate network call
    return f"Data for {item_id}"

def run_pool() -> None:
    results: list[str] = []
    # Context manager handles pool shutdown
    with ThreadPoolExecutor(max_workers=3) as executor:
        # Submit tasks
        futures = [executor.submit(fetch_data, i) for i in range(5)]
        
        # Process as they complete
        for future in as_completed(futures):
            results.append(future.result())
            
    print(f"Collected {len(results)} items.")

run_pool()
````

#### Common Mistakes and Pitfalls

Mistake 1: Calling `run()` instead of `start()`. Calling `run()` executes the target function directly in the current main thread, providing no concurrency.


````python
import threading

def task() -> None:
    print("Running in thread")

t = threading.Thread(target=task)
# BROKEN: This runs sequentially on the main thread
t.run() 
# FIX: Use t.start() to spawn a new OS thread
````

Mistake 2: Forgetting to `join()` threads. If you spawn threads and don't join them, the main program might exit prematurely, or attempt to evaluate state that the worker threads haven't finished mutating yet.


#### Best Practices
- Always use `ThreadPoolExecutor` instead of manually appending threads to a list and joining them in a loop. It handles thread lifecycle, error propagation, and result aggregation cleanly.
- Use `with lock:` (context manager) rather than `lock.acquire()` and `lock.release()`. If an exception occurs between acquire and release, the context manager guarantees the lock is freed.
- Use `daemon=True` for background polling threads that should not block the application from exiting.
- Pass required state into the thread target explicitly via arguments, rather than relying on global mutable state, to minimize the need for synchronization.

#### Real-World Use Case
A log aggregator daemon. The main thread accepts HTTP requests and writes log strings to a queue. A daemon thread loops indefinitely, pulling batches of logs from the thread-safe `queue.Queue`, sending them to AWS CloudWatch over the network. When the web server shuts down, the daemon thread safely dies because it was marked `daemon=True`.

#### Performance Notes
Thread creation involves asking the OS to allocate stack space and kernel resources. It is not free. Creating 10,000 threads simultaneously will likely crash or stall the OS. Concurrency should be bounded using thread pools or semaphores.

#### Quick Recap
- Threads share memory; race conditions occur without synchronization.
- `threading.Lock` enforces mutual exclusion on critical sections.
- `start()` spawns the thread; `join()` waits for its completion.
- Daemon threads exit immediately when the main program finishes.
- `ThreadPoolExecutor` is the modern, preferred API for worker pools.

#### Interview Questions
1. **Junior**: What happens if an exception is raised inside a thread? Does it crash the main program?
   *Answer*: No. By default, the exception traceback is printed to stderr, and that specific thread terminates. The main thread and other threads continue executing normally.
2. **Mid**: When would you use an `RLock` (Reentrant Lock) over a standard `Lock`?
   *Answer*: A standard `Lock` deadlocks if the thread holding it attempts to acquire it again (e.g., in a recursive function). An `RLock` tracks the owning thread and a recursion count, allowing the same thread to acquire it multiple times without blocking.
3. **Senior**: Explain the difference between `Event` and `Condition` in the `threading` module.
   *Answer*: An `Event` is a simple boolean flag. Threads can wait for the flag to become true, and one thread can set it to true. It alerts all waiting threads simultaneously. A `Condition` is bound to a specific lock and allows threads to wait for complex state changes, waking up one or all threads (`notify()` vs `notify_all()`). Conditions are used to build thread-safe queues.


### Topic 3.4 — Parallelism with multiprocessing

#### Concept Overview
Because the GIL prevents multithreading from achieving true CPU parallelism in CPython, the `multiprocessing` module bypasses the GIL entirely by spawning completely separate OS processes. Each process boots its own independent Python interpreter, has its own GIL, and allocates its own distinct memory space. This allows Python code to utilize all CPU cores simultaneously.

Because memory is isolated, sharing state between processes is complex. You cannot simply read and write to a global variable. Data must be serialized (using the `pickle` module) and transmitted over inter-process communication (IPC) channels like Pipes or Queues, or placed in specialized shared memory structures (`Value`, `Array`, `Manager`).


#### The Mind Picture
```text
[ Process 1 (Main) ]         [ Process 2 (Worker) ]
 Interpreter A                Interpreter B
 memory space A               memory space B
 GIL A                        GIL B
       |                            |
       +------ IPC (Pickle) ------->|
```


#### Syntax Anatomy
```python
import multiprocessing as mp

# 1. Spawning a process
p = mp.Process(target=func, args=(arg,))
p.start()
p.join()

# 2. Worker Pool
from multiprocessing import Pool
with Pool(processes=4) as pool:
    # Distributes iterable across processes
    results = pool.map(func, iterable)

# 3. Inter-Process Queue
queue = mp.Queue()
queue.put(item)
item = queue.get()
```


#### Practical Examples

First, achieving true parallelism for CPU-bound tasks using `ProcessPoolExecutor`.


````python
from concurrent.futures import ProcessPoolExecutor
import time
import math

def heavy_computation(x: int) -> float:
    # A CPU bound task
    return sum(math.sqrt(i) for i in range(x, x + 5_000_000))

def run_parallel() -> None:
    inputs = [10, 20, 30, 40]
    start = time.time()
    
    # max_workers defaults to number of logical CPU cores
    with ProcessPoolExecutor() as executor:
        results = list(executor.map(heavy_computation, inputs))
        
    print(f"Parallel time: {time.time() - start:.2f}s")

# Note: In a script, multiprocessing requires the __name__ == '__main__' guard.
# Since this is a notebook cell, the environment handles it, but in raw files:
# if __name__ == '__main__':
#     run_parallel()
````

Next, sharing simple state using `multiprocessing.Value`.


````python
import multiprocessing as mp
import time

# 'i' indicates an integer type in C
def increment_shared(val: mp.Value, lock: mp.Lock) -> None:
    for _ in range(100):
        # We must lock the shared memory just like threading
        with lock:
            val.value += 1

def run_shared_state() -> None:
    shared_counter = mp.Value('i', 0)
    lock = mp.Lock()
    
    processes = [
        mp.Process(target=increment_shared, args=(shared_counter, lock))
        for _ in range(4)
    ]
    
    for p in processes: p.start()
    for p in processes: p.join()
    
    print(f"Final shared value: {shared_counter.value}")

run_shared_state()
````

Using `Pool.starmap` to pass multiple arguments to a target function.


````python
from multiprocessing import Pool

def multiply(x: int, y: int) -> int:
    return x * y

def run_starmap() -> None:
    args_list = [(2, 3), (4, 5), (6, 7)]
    
    with Pool(processes=3) as pool:
        # map() only takes one argument per function call
        # starmap() unpacks the tuples into multiple arguments
        results = pool.starmap(multiply, args_list)
        
    print(f"Starmap results: {results}")

run_starmap()
````

#### Common Mistakes and Pitfalls

Mistake 1: Attempting to pass unpicklable objects (like active network sockets, database connections, or threading Locks) between processes. Inter-process communication strictly requires data to be serializable by `pickle`.


````python
import multiprocessing as mp
import socket

def worker(sock: socket.socket) -> None:
    pass

# BROKEN: Sockets cannot be pickled
def run_broken() -> None:
    s = socket.socket()
    p = mp.Process(target=worker, args=(s,))
    # p.start() # This raises TypeError/PicklingError
````

Mistake 2: Missing the `if __name__ == '__main__':` block on Windows. Because Windows lacks the `fork()` system call, CPython must import the main module again to initialize the new process. If the process spawning code is unguarded, it will execute recursively, crashing the system.


#### Best Practices
- Always guard multiprocessing entry points with `if __name__ == '__main__':`.
- Prefer `concurrent.futures.ProcessPoolExecutor` for functional, input-output mapping workloads. Use the `multiprocessing` module directly only when you need fine-grained control over queues, events, and shared memory.
- Pass primitives or simple data classes between processes. Re-establish database or network connections independently inside the worker process function.
- Be aware of IPC overhead. If a function executes in 0.1ms, sending data over a multiprocessing queue to a worker will take longer than executing the function locally.

#### Real-World Use Case
Image and video processing. A Python application processing 10,000 high-resolution images via `Pillow` or `OpenCV`. A `ProcessPoolExecutor` allows 8 worker processes to each grab image paths from a list, read the file, apply heavy matrix transformations on the CPU, and write the output, achieving an 8x speedup on an octa-core machine.

#### Performance Notes
Process creation is extremely heavy compared to thread creation. You are booting an entire Python runtime. Therefore, worker pools must be reused. The crossover point where multiprocessing wins against threading is entirely dependent on the ratio of CPU crunching versus the overhead of pickling arguments and IPC. If pickling massive lists of dictionaries takes 2 seconds, and the math operation takes 1 second, multiprocessing is a net loss.

#### Quick Recap
- `multiprocessing` defeats the GIL by creating independent OS processes.
- Memory is fully isolated; sharing requires IPC queues, pipes, or explicit shared memory (`Value`, `Array`).
- Arguments passed to processes must be serializable by `pickle`.
- Use `Pool` or `ProcessPoolExecutor` to map workloads across multiple cores.
- Always use the `__main__` guard to prevent recursive process spawning on Windows.

#### Interview Questions
1. **Junior**: Why can't I just use multithreading to speed up a mathematical loop in Python?
   *Answer*: The GIL prevents multiple threads from executing Python bytecodes simultaneously. The threads will just take turns, and the overhead of context switching will make it slower than a single thread. You must use multiprocessing for math loops.
2. **Mid**: What is the difference between `multiprocessing.Queue` and `queue.Queue`?
   *Answer*: `queue.Queue` is thread-safe and relies on shared memory. `multiprocessing.Queue` is process-safe; it uses OS pipes and `pickle` serialization to serialize and transmit data across isolated process memory spaces.
3. **Senior**: When using `multiprocessing.Pool`, when would you choose `imap_unordered` over `map`?
   *Answer*: `map` blocks until all results are computed, returning an ordered list. It can consume massive memory if the result set is huge. `imap_unordered` yields results as an iterator exactly as soon as any worker finishes, completely out of order. It is optimal for processing massive streams of independent data where order doesn't matter, saving RAM and improving pipeline latency.


# Topic 3.5 — asyncio

#### Concept Overview
Python's `asyncio` implements cooperative multitasking on a single thread using an event loop. Instead of the operating system preempting threads, your code explicitly yields control at designated `await` points. This concurrency model shines for I/O-bound tasks. A coroutine, defined with `async def`, is a specialized generator that can suspend its execution. When it suspends, the event loop runs other scheduled coroutines. This avoids the overhead of context switching and thread synchronization.

#### The Mind Picture
```
[Event Loop] <------------------------------------+
     |                                            |
     | (schedules)                                | (yields control via await)
     v                                            |
[Task 1 (coroutine)] -- running --> await I/O ----+
                                                  |
[Task 2 (coroutine)] <--- resumes (I/O ready) ----+
```

#### Syntax Anatomy
```python
import asyncio

async def fetch_data(id: int) -> str:
    # 'async def' defines a coroutine function
    # 'await' suspends execution until the awaitable completes
    await asyncio.sleep(1)
    return f"Data {id}"

# asyncio.run() creates the event loop, runs the main coroutine, and closes the loop
# result = asyncio.run(fetch_data(1))
```



````python
import asyncio
from typing import List

async def fetch_item(item_id: int, delay: float) -> str:
    await asyncio.sleep(delay)
    return f"Item {item_id}"

````

#### Practical Examples
We start by scheduling a single coroutine. Then, we use `asyncio.create_task` to schedule a coroutine in the background. Finally, `asyncio.gather` runs multiple coroutines concurrently and waits for all of them.


````python
async def main() -> None:
    # 1. Awaiting directly
    result = await fetch_item(1, 0.1)
    print(result)

    # 2. Creating a background task
    task = asyncio.create_task(fetch_item(2, 0.2))
    print("Task created, doing other work...")
    await asyncio.sleep(0.1)
    print(await task)

    # 3. Concurrent execution
    coros = [fetch_item(i, 0.1) for i in range(3, 6)]
    results: List[str] = await asyncio.gather(*coros)
    print(results)

asyncio.run(main())

````

#### Common Mistakes and Pitfalls
The most destructive mistake is calling synchronous blocking code inside an `async def`. This freezes the entire event loop, stopping all tasks.


````python
import time

async def bad_coroutine() -> None:
    # WRONG: time.sleep blocks the thread, starving the event loop
    time.sleep(1)

async def fixed_coroutine() -> None:
    # RIGHT: await asyncio.to_thread runs blocking code in a worker thread
    await asyncio.to_thread(time.sleep, 1)

````

#### Best Practices
1. Never block the event loop with synchronous I/O or heavy CPU work. Use `asyncio.to_thread` or `ProcessPoolExecutor`.
2. Do not mix `asyncio.run()` with existing event loops. Use it strictly as the top-level entry point.
3. Catch `asyncio.CancelledError` if you need specific cleanup, but always re-raise it.
4. Keep references to background tasks. If you do not store the result of `asyncio.create_task()`, the garbage collector might destroy it mid-execution.

#### Real-World Use Case
Fetching data concurrently from multiple APIs. Frameworks like FastAPI and libraries like `httpx` rely on this exact pattern.


````python
# Using a mock HTTP fetcher to simulate httpx.AsyncClient
import asyncio
from typing import Dict, Any

async def mock_http_get(url: str) -> Dict[str, Any]:
    await asyncio.sleep(0.1)
    return {"status": 200, "url": url}

async def fetch_all(urls: list[str]) -> list[Dict[str, Any]]:
    tasks = [asyncio.create_task(mock_http_get(url)) for url in urls]
    return await asyncio.gather(*tasks)

urls = ["http://api.example.com/users", "http://api.example.com/posts"]
print(asyncio.run(fetch_all(urls)))

````

#### Performance Notes
`asyncio` excels for I/O bound work where latency is high. It handles thousands of concurrent socket connections efficiently. It provides zero benefit for CPU-bound tasks, as Python's GIL ensures only one thread executes Python bytecode at a time.

#### Quick Recap
- `asyncio` provides concurrent execution on a single thread.
- `async def` makes a function a coroutine.
- `await` yields control back to the event loop.
- `asyncio.create_task` schedules a coroutine for execution.
- `asyncio.gather` aggregates results from multiple tasks concurrently.
- Blocking the event loop destroys concurrency.

#### Interview Questions
- **Junior**: What is the difference between a normal function and an `async def` function?
  - *Answer*: A normal function returns its result immediately. An `async def` function returns a coroutine object when called. You must `await` it or schedule it on an event loop to execute its body.
- **Mid**: You have a heavy image processing function. Can you run it inside an `async def` to avoid freezing the API?
  - *Answer*: No. `asyncio` is single-threaded. CPU-heavy work will block the event loop. You must run it in a `ProcessPoolExecutor` or use `asyncio.to_thread` for threads (if it releases the GIL).
- **Senior**: How do you gracefully shut down an asyncio application?
  - *Answer*: You retrieve all running tasks via `asyncio.all_tasks()`, iterate over them and call `.cancel()`. Then you use `asyncio.gather(*tasks, return_exceptions=True)` to wait for them to finish handling `CancelledError` before closing the loop.

# Topic 3.6 — Metaclasses

#### Concept Overview
In Python, classes are objects too. Just as an instance is created by a class, a class is created by a metaclass. By default, the metaclass for all classes is `type`. A metaclass intercepts the class creation process, allowing you to modify the class dictionary, alter inheritance, or automatically register the class before it is ever instantiated.

#### The Mind Picture
```
[ type (Metaclass) ]
       |
       | (creates)
       v
[ MyClass (Class) ]
       |
       | (creates)
       v
[ my_obj (Instance) ]
```

#### Syntax Anatomy
```python
# The standard class definition syntax:
class Foo(metaclass=type):
    pass

# Is syntactic sugar for:
# Foo = type("Foo", (), {})
# type(name, bases, namespace_dict)
```



````python
from typing import Any, Dict, Tuple

class RegistryMeta(type):
    registry: Dict[str, type] = {}

    def __new__(mcs, name: str, bases: Tuple[type, ...], namespace: Dict[str, Any]) -> type:
        # Intercept class creation
        new_class = super().__new__(mcs, name, bases, namespace)
        mcs.registry[name] = new_class
        return new_class

````

#### Practical Examples
Metaclasses allow for implicit registration patterns. Any class that uses `RegistryMeta` will automatically insert itself into the registry dictionary upon creation.


````python
class BasePlugin(metaclass=RegistryMeta):
    pass

class AudioPlugin(BasePlugin):
    pass

class VideoPlugin(BasePlugin):
    pass

print(RegistryMeta.registry)

````

#### Common Mistakes and Pitfalls
Using a metaclass when `__init_subclass__` or a class decorator would suffice. Metaclasses introduce complexity and can conflict with each other (metaclass conflicts).


````python
# WRONG: Using a metaclass just to modify a class attribute
class BadMeta(type):
    def __init__(cls, name: str, bases: tuple, namespace: dict) -> None:
        cls.is_registered = True
        super().__init__(name, bases, namespace)

# RIGHT: Use __init_subclass__ for simple class modifications
class GoodBase:
    is_registered: bool = False

    def __init_subclass__(cls, **kwargs: Any) -> None:
        super().__init_subclass__(**kwargs)
        cls.is_registered = True

````

#### Best Practices
1. Avoid metaclasses unless you absolutely need to modify class creation mechanics that `__init_subclass__` cannot handle.
2. If you do write a metaclass, subclass `type`.
3. Use `__new__` to modify the class structure before it is created.
4. Keep metaclasses isolated; combining classes with different metaclasses raises a `TypeError`.

#### Real-World Use Case
Object-Relational Mappers (ORMs) like Django or SQLAlchemy. They use metaclasses to read declarative class attributes (fields), collect them, and build the database schema mapping behind the scenes.

#### Performance Notes
Metaclass execution happens strictly at import time (class definition time). It has zero impact on instance creation or method dispatch time.

#### Quick Recap
- A metaclass is the class of a class.
- The default metaclass is `type`.
- `__new__` on a metaclass receives `(name, bases, namespace)`.
- Use metaclasses for framework-level automation.
- Prefer `__init_subclass__` for simpler subclass tracking.

#### Interview Questions
- **Junior**: What is `type` in Python?
  - *Answer*: `type` is a built-in function that returns the type of an object, but it is also the default metaclass. Calling `type(name, bases, namespace)` dynamically creates a new class.
- **Mid**: What is the difference between `__new__` and `__init__` in a metaclass?
  - *Answer*: `__new__` allocates the new class object and can modify the namespace dict before the class is created. `__init__` is called after the class object exists to initialize it.
- **Senior**: When would you encounter a metaclass conflict and how do you resolve it?
  - *Answer*: A conflict occurs when a class inherits from multiple base classes that have different, unrelated metaclasses. You resolve it by writing a new metaclass that inherits from all the conflicting metaclasses, and using that for your class.

# Topic 3.7 — Descriptors

#### Concept Overview
The descriptor protocol allows you to customize attribute access on objects. Any class that implements `__get__`, `__set__`, or `__delete__` is a descriptor. Python uses descriptors internally to implement properties, methods, static methods, and class methods. They provide a unified way to manage state logic, validation, and type-checking across attributes.

#### The Mind Picture
```
[ Instance: obj ] --> access obj.attr
       |
[ Class: ObjClass ] --> looks up attr
       |
[ Descriptor ] --> has __get__ ? Calls attr.__get__(obj, ObjClass)
```

#### Syntax Anatomy
```python
class Descriptor:
    def __get__(self, obj: Any, objtype: type = None) -> Any: ...
    def __set__(self, obj: Any, value: Any) -> None: ...
    def __delete__(self, obj: Any) -> None: ...
```



````python
from typing import Any, Type, Optional

class ValidatedString:
    def __set_name__(self, owner: Type[Any], name: str) -> None:
        self.public_name = name
        self.private_name = '_' + name

    def __get__(self, obj: Any, objtype: Type[Any] = None) -> Any:
        if obj is None:
            return self
        return getattr(obj, self.private_name)

    def __set__(self, obj: Any, value: str) -> None:
        if not isinstance(value, str):
            raise TypeError(f"Expected {self.public_name} to be a string")
        setattr(obj, self.private_name, value)

````

#### Practical Examples
By using the `ValidatedString` descriptor, we enforce type checking every time the attribute is assigned. `__set_name__` binds the descriptor to the attribute name dynamically.


````python
class User:
    username = ValidatedString()

u = User()
u.username = "admin"
print(u.username)

````

#### Common Mistakes and Pitfalls
Storing data on the descriptor instance itself instead of the owner instance. Descriptors are instantiated at the class level, so modifying `self.value` inside the descriptor shares that state across all instances.


````python
# WRONG: Shared state across instances
class BadDescriptor:
    def __init__(self) -> None:
        self.val = 0
    def __get__(self, obj: Any, objtype: Any) -> int: return self.val
    def __set__(self, obj: Any, val: int) -> None: self.val = val

# RIGHT: Storing state on the owner object
class GoodDescriptor:
    def __set_name__(self, owner: Any, name: str) -> None:
        self.name = name
    def __get__(self, obj: Any, objtype: Any) -> Any:
        if obj is None: return self
        return obj.__dict__.get(self.name)
    def __set__(self, obj: Any, val: Any) -> None:
        obj.__dict__[self.name] = val

````

#### Best Practices
1. Use `__set_name__` (added in 3.6) to capture the attribute name assigned to the descriptor.
2. Return the descriptor instance itself from `__get__` when `obj` is `None` (i.e., accessed on the class).
3. Use the object's `__dict__` or a private attribute on the object to store instance data.

#### Real-World Use Case
The `@property` decorator is implemented purely in Python using the descriptor protocol. SQLAlchemy columns (`Column`) are descriptors that handle mapping attribute access to SQL statements.

#### Performance Notes
Descriptor access is highly optimized in CPython, but it still introduces a function call overhead compared to direct dictionary lookups. For attributes accessed in tight loops, local variable caching is advised.

#### Quick Recap
- Descriptors intercept attribute access via `__get__`, `__set__`, and `__delete__`.
- A data descriptor implements `__set__` or `__delete__` and overrides instance attributes.
- A non-data descriptor implements only `__get__` (e.g., methods).
- `__set_name__` binds the attribute name automatically.

#### Interview Questions
- **Junior**: What is the difference between `@property` and a descriptor?
  - *Answer*: `@property` is a built-in decorator that implements the descriptor protocol. You use `@property` for simple, one-off computed attributes, and write custom descriptors when you need to reuse attribute logic across multiple classes.
- **Mid**: How does method binding work in Python?
  - *Answer*: Functions implement `__get__`. When a function is accessed via an instance (`obj.method`), the function's `__get__` method is invoked, which returns a bound method object wrapping the function and the instance.
- **Senior**: Why does a data descriptor take precedence over the instance dictionary, but a non-data descriptor does not?
  - *Answer*: This is designed specifically so that methods (non-data descriptors) can be overridden by instance attributes, while properties (data descriptors) provide strict control and cannot be bypassed simply by assigning to the instance dict.

# Topic 3.8 — Advanced Decorators

#### Concept Overview
A decorator is a callable that takes a callable and returns a modified callable. Advanced decorators include parameterized decorators, which require a nested factory structure, and class-based decorators, which encapsulate state. Proper decorator design requires preserving the original function's metadata to ensure debugging tools and documentation generation continue to work correctly.

#### The Mind Picture
```
Parameterized Decorator Execution:
@retry(retries=3)
def func(): ...

1. factory = retry(retries=3)
2. wrapper = factory(func)
3. func = wrapper
```

#### Syntax Anatomy
```python
from typing import Callable, Any
import functools

def decorator_with_args(arg: int) -> Callable:
    def decorator(func: Callable) -> Callable:
        @functools.wraps(func)
        def wrapper(*args: Any, **kwargs: Any) -> Any:
            # use arg, call func
            return func(*args, **kwargs)
        return wrapper
    return decorator
```



````python
import functools
import time
from typing import Callable, Any

def retry(max_attempts: int, delay: float = 0.1) -> Callable:
    def decorator(func: Callable) -> Callable:
        @functools.wraps(func)
        def wrapper(*args: Any, **kwargs: Any) -> Any:
            for attempt in range(max_attempts):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    if attempt == max_attempts - 1:
                        raise
                    time.sleep(delay)
        return wrapper
    return decorator

````

#### Practical Examples
Applying the retry decorator to a function that might fail intermittently. Using `functools.wraps` preserves the `__name__` and `__doc__` of `flaky_network_call`.


````python
@retry(max_attempts=3, delay=0.01)
def flaky_network_call() -> str:
    # Simulate random failure
    return "Success"

print(flaky_network_call())
print(flaky_network_call.__name__)

````

#### Common Mistakes and Pitfalls
Forgetting `functools.wraps`. This causes the decorated function to lose its signature, name, and docstring, masking errors and breaking IDE autocompletion.


````python
# WRONG
def bad_decorator(func: Callable) -> Callable:
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        return func(*args, **kwargs)
    return wrapper

@bad_decorator
def example() -> None:
    '''Does something.'''
    pass

# example.__name__ is now 'wrapper', and the docstring is lost.

````

#### Best Practices
1. Always use `functools.wraps`.
2. Keep decorators orthogonal. A decorator should do exactly one thing (e.g., logging, retrying) so they can be stacked.
3. Be mindful of argument modifications. If your wrapper alters `args` or `kwargs`, document the new signature clearly.
4. When stacking, remember decorators evaluate bottom-to-top, but execute top-to-bottom.

#### Real-World Use Case
Web frameworks like Flask use parameterized decorators (`@app.route('/path')`) to map URLs to handler functions, appending metadata to a central registry without altering the underlying function execution logic.

#### Performance Notes
Each decorator adds an extra function call to the stack. Deeply nested decorators can slightly increase execution overhead and complicate stack traces during exception handling.

#### Quick Recap
- Parameterized decorators require three levels of nested functions.
- `functools.wraps` is mandatory for maintaining metadata.
- Decorators can be classes by implementing `__call__`.
- Stacking order determines execution flow.

#### Interview Questions
- **Junior**: What is `functools.wraps` and why do we use it?
  - *Answer*: It is a helper decorator applied to the wrapper function inside a decorator. It copies attributes like `__name__`, `__doc__`, and `__annotations__` from the original function to the wrapper so introspection tools work correctly.
- **Mid**: How do you create a decorator that can be used both with and without parentheses (e.g., `@log` and `@log(level='DEBUG')`)?
  - *Answer*: You inspect the arguments passed to the outer function. If the first argument is a callable and `kwargs` are empty, it was used without parentheses. Otherwise, return a decorator factory.
- **Senior**: When would you implement a decorator as a class instead of a function closure?
  - *Answer*: When the decorator requires complex state management across invocations. For example, a rate limiter decorator that needs to track timestamps and counts internally is cleaner as a class with state rather than using `nonlocal` variables in a closure.



#### Concept Overview — Abstract Base Classes and Mixins
Abstract Base Classes (ABCs) provide a formal way to define interfaces in Python. Before ABCs, developers relied on raising `NotImplementedError` in base class methods, but this only caught errors at runtime when the method was actually called. The `abc` module allows you to define classes that cannot be instantiated directly and require subclasses to implement specific methods. This shifts the failure point from method execution time to instantiation time, making errors easier to trace.

Mixins are a design pattern rather than a language feature. A mixin is a class designed to provide a specific piece of functionality to subclasses but is not meant to be instantiated on its own. They are "mixed in" via multiple inheritance. By combining ABCs to define interfaces and Mixins to provide modular implementations, you build highly flexible and reusable class hierarchies.

#### The Mind Picture
```text
[ Abstract Interface ]         [ Modular Behavior ]
     Shape (ABC)                  ColorMixin
     - area() (abstract)          - get_color()
             \                      /
              \                    /
               [ Concrete Class ]
                 ColoredCircle
                 - area() -> implemented
                 - get_color() -> inherited
```
The ABC acts as a contract template. The Mixin acts as a backpack of features. The concrete class fulfills the contract and wears the backpack.

#### Syntax Anatomy
```python
from abc import ABC, abstractmethod

class MyInterface(ABC):               # Inherit from ABC (sets metaclass to ABCMeta)
    
    @abstractmethod                   # Decorator forcing subclass implementation
    def required_method(self) -> int:
        pass                          # Usually pass, but can contain default implementation
        
    @property
    @abstractmethod                   # Combine properties and abstract methods
    def required_property(self) -> str:
        pass

class MyMixin:                        # Convention: Name ends with Mixin
    def provided_behavior(self) -> None:
        print("Behavior added!")
```

#### Practical Examples
Defining a basic ABC and seeing it enforce implementation.


````python
from abc import ABC, abstractmethod

class DataStore(ABC):
    @abstractmethod
    def save(self, data: dict[str, str]) -> None:
        pass

    @abstractmethod
    def load(self) -> dict[str, str]:
        pass

# This fails: TypeError: Can't instantiate abstract class MissingImplementation with abstract methods load
class MissingImplementation(DataStore):
    def save(self, data: dict[str, str]) -> None:
        print("Saved")
        
try:
    bad_store = MissingImplementation()
except TypeError as e:
    print(f"Expected failure: {e}")

````

A working implementation combined with a Mixin for logging.


````python
from typing import Any

class LoggingMixin:
    def log(self, message: str) -> None:
        # In a real app, use the logging module
        print(f"[{self.__class__.__name__} LOG]: {message}")

class MemoryStore(DataStore, LoggingMixin):
    def __init__(self) -> None:
        self._storage: dict[str, str] = {}
        
    def save(self, data: dict[str, str]) -> None:
        self._storage.update(data)
        self.log(f"Saved {len(data)} items.")
        
    def load(self) -> dict[str, str]:
        self.log("Loading all items.")
        return self._storage.copy()

store = MemoryStore()
store.save({"user": "admin", "id": "123"})

````

Using `collections.abc` to get free methods. If you implement `__getitem__` and `__len__`, `Sequence` gives you `__contains__`, `__iter__`, `__reversed__`, `index`, and `count`.


````python
from collections.abc import Sequence

class CustomList(Sequence):
    def __init__(self, data: list[int]) -> None:
        self._data = data
        
    def __getitem__(self, index: int) -> int:
        return self._data[index]
        
    def __len__(self) -> int:
        return len(self._data)

c_list = CustomList([10, 20, 30])
print(f"Contains 20? {20 in c_list}")  # Free from Sequence
print(f"Count of 10: {c_list.count(10)}") # Free from Sequence

````

#### Common Mistakes and Pitfalls
Forgetting to inherit from `ABC` when using `@abstractmethod`. The decorator alone does nothing without the `ABCMeta` metaclass.


````python
from abc import abstractmethod

class FakeABC:
    @abstractmethod
    def do_work(self) -> None:
        pass

# This succeeds! The developer thought they made an ABC, but they didn't.
# They will get a None return when calling do_work instead of a TypeError on instantiation.
fake = FakeABC()
print(f"Should have failed, but got: {fake}")

````

Incorrect order when combining `@property` and `@abstractmethod`. `@property` must be the outermost decorator.


````python
from abc import ABC, abstractmethod

class BadPropertyOrder(ABC):
    @abstractmethod
    @property  # WRONG: @property must be on top
    def name(self) -> str:
        pass
        
# Fix:
class GoodPropertyOrder(ABC):
    @property
    @abstractmethod
    def name(self) -> str:
        pass

````

#### Best Practices
1. Suffix mixins with `Mixin` (e.g., `SerializableMixin`) to clarify intent.
2. Mixins should not have `__init__` methods that require arguments, as it complicates `super()` chains.
3. Inherit from Mixins first, then the base class: `class MyClass(MyMixin, BaseClass):`.
4. Use `collections.abc` for structural typing and duck typing checks (`isinstance(obj, Sequence)`).
5. Only use ABCs when defining a clear contract for external implementers; don't overuse them for simple internal structures.

#### Real-World Use Case
The Django web framework uses Mixins extensively for Class-Based Views (CBVs). For example, `LoginRequiredMixin` is mixed into a view class to ensure only authenticated users can access it. By combining `LoginRequiredMixin` with `TemplateView`, developers compose behavior cleanly without deep inheritance trees.

#### Performance Notes
ABCs introduce a slight overhead at import time due to the `ABCMeta` metaclass machinery checking abstract methods. Instantiation also has a tiny overhead to check that all abstract methods are implemented. In practice, this is completely negligible and rarely a bottleneck in any standard application.

#### Quick Recap
- ABCs formalize interfaces and prevent instantiation if methods are missing.
- Define ABCs by inheriting from `abc.ABC` and using `@abstractmethod`.
- Combine `@property` and `@abstractmethod` with `@property` on top.
- Mixins are stateless classes meant to add behavior via multiple inheritance.
- Name mixins with the `Mixin` suffix.
- `collections.abc` provides ready-to-use interfaces like `Sequence` and `Mapping`.

#### Interview Questions
1. **Junior**: How does an ABC differ from a regular class raising `NotImplementedError`?
   *Answer*: Raising `NotImplementedError` fails when the method is called. An ABC fails at instantiation time, catching the error earlier.
2. **Mid**: Explain how `register()` works in the `abc` module.
   *Answer*: `MyABC.register(ThirdPartyClass)` marks a class as a virtual subclass of the ABC. `isinstance(obj, MyABC)` will return True, even though `ThirdPartyClass` doesn't actually inherit from `MyABC`. This allows integrating third-party code into your type checks without modifying their source.
3. **Senior**: Can an abstract method have a concrete implementation in the base class? Why would you do this?
   *Answer*: Yes, an abstract method can contain code. The subclass is forced to override the method, but it can use `super().method_name()` to call the base class implementation. This is useful for providing common boilerplate or validation logic that every subclass must execute while still forcing them to provide their own specific logic.

#### Concept Overview — Multiple Inheritance and C3 Linearization
Python supports multiple inheritance, allowing a class to inherit from multiple base classes. To resolve method calls, Python uses a specific algorithm called C3 Linearization to compute the Method Resolution Order (MRO). The MRO defines the exact sequence in which base classes are searched when you call a method or access an attribute.

C3 Linearization guarantees two properties: Monotonicity (a class is always searched before its parent classes) and Local Precedence Order consistency (the order of bases listed in the class definition is preserved). Understanding MRO is crucial for using `super()` correctly in complex hierarchies, known as cooperative multiple inheritance.

#### The Mind Picture
```text
Diamond Inheritance:
      [ Base ]
      /      \
 [Left]      [Right]
      \      /
     [ Bottom ]

MRO determines the path: Bottom -> Left -> Right -> Base -> object
```
C3 merges the MROs of the parents while preserving the local order (`Left` before `Right`).

#### Syntax Anatomy
```python
class Base: pass
class Left(Base): pass
class Right(Base): pass
class Bottom(Left, Right): pass

# View MRO via the __mro__ attribute or mro() method
# print(Bottom.__mro__)
# (<class 'Bottom'>, <class 'Left'>, <class 'Right'>, <class 'Base'>, <class 'object'>)
```

#### Practical Examples
A standard diamond inheritance pattern. Notice how `super()` propagates.


````python
class Root:
    def __init__(self) -> None:
        print("Root init")
        super().__init__()

class BranchA(Root):
    def __init__(self) -> None:
        print("BranchA init")
        super().__init__()

class BranchB(Root):
    def __init__(self) -> None:
        print("BranchB init")
        super().__init__()

class Leaf(BranchA, BranchB):
    def __init__(self) -> None:
        print("Leaf init")
        super().__init__()

print(f"MRO: {[c.__name__ for c in Leaf.__mro__]}")
print("--- Instantiation ---")
l = Leaf()

````

A case where C3 linearization fails and raises a TypeError. You cannot define a class that breaks consistency.


````python
class X: pass
class Y(X): pass

# Y inherits from X, so Y must come before X.
# But here we declare X before Y in the bases list.
try:
    class Bad(X, Y):
        pass
except TypeError as e:
    print(f"C3 Failure: {e}")

````

#### Common Mistakes and Pitfalls
Failing to call `super()` breaks the chain in cooperative multiple inheritance.


````python
class ChainRoot:
    def process(self) -> None:
        print("Root process")

class LinkA(ChainRoot):
    def process(self) -> None:
        print("LinkA process")
        # Missing super().process()!

class LinkB(ChainRoot):
    def process(self) -> None:
        print("LinkB process")
        super().process()

class BrokenChain(LinkA, LinkB):
    def process(self) -> None:
        print("BrokenChain process")
        super().process()

b = BrokenChain()
b.process() # Notice LinkB and ChainRoot are completely skipped!

````

#### Best Practices
1. Always call `super()` in overridden methods if the class is designed for multiple inheritance.
2. Design base classes to accept `**kwargs` and pass them to `super()` to handle unknown initialization arguments gracefully.
3. Keep inheritance hierarchies flat. Prefer composition over deep inheritance.
4. List mixins first in the inheritance list: `class View(AuthMixin, BaseView)`.
5. Inspect `Class.__mro__` when debugging unexpected method resolution behavior.

#### Real-World Use Case
In standard library `socketserver`, multiple inheritance is used to combine server types and concurrency models. You have `TCPServer` and `UDPServer`. You also have `ThreadingMixIn` and `ForkingMixIn`. To create a threaded TCP server, you simply declare `class ThreadingTCPServer(ThreadingMixIn, TCPServer): pass`. The mixin intercepts the request handling and spins up a thread.

#### Performance Notes
MRO computation happens strictly at class definition time, not at runtime. Therefore, complex inheritance hierarchies do not impact instantiation or method lookup speed. Method resolution itself is highly optimized in CPython via method caching.

#### Quick Recap
- Multiple inheritance relies on C3 linearization to create the MRO.
- C3 ensures a class is checked before its parents, and listed bases are checked in order.
- `__mro__` shows the exact method resolution order.
- `super()` delegates to the next class in the MRO, not necessarily the direct parent.
- Failing to call `super()` in one class breaks the entire cooperative chain.

#### Interview Questions
1. **Junior**: How do you check the method resolution order of a class?
   *Answer*: Using the `__mro__` attribute or the `mro()` method on the class.
2. **Mid**: Why does Python use C3 linearization instead of simple depth-first search?
   *Answer*: Simple depth-first search in diamond inheritance would cause the base class to be searched before the second branch. C3 ensures monotonicity, meaning subclasses are always searched before their common base classes.
3. **Senior**: Explain what `super()` actually does under the hood in a multiple inheritance scenario.
   *Answer*: `super()` does not call the parent class. It creates a proxy object that delegates method calls to classes that appear *after* the current class in the instance's MRO. It uses the `__class__` of the instance and the lexical class where `super()` was called to determine the next step in the MRO list.

#### Concept Overview — Design Patterns in Python
Design patterns are typical solutions to common problems in software design. However, patterns created for statically typed, rigidly object-oriented languages (like Java or C++) often look different in Python. Python's dynamic nature, first-class functions, and modules often render classic implementation boilerplate unnecessary. The goal is to implement patterns "Pythonically" rather than translating them verbatim from GoF (Gang of Four) books.

We focus on four patterns: Singleton (ensuring a class has one instance), Factory (creating objects without specifying the exact class), Observer (notifying dependents of state changes), and Strategy (encapsulating a family of algorithms).

#### The Mind Picture
```text
Classic OOP Strategy      vs.     Pythonic Strategy
[ Interface ]                     sorted(data, key=my_func)
      |                             (Functions are first-class!)
 [ Concrete ]
```

#### Syntax Anatomy
```python
# Classic Factory vs Pythonic Factory
class ClassicFactory:
    def create(self, type_name: str) -> BaseClass: ...

def pythonic_factory(type_name: str) -> BaseClass:
    # A simple dictionary mapping names to classes
    registry = {"a": ClassA, "b": ClassB}
    return registry[type_name]()
```

#### Practical Examples
Singleton pattern. Classic OOP uses a private constructor. In Python, modules are natural singletons, but you can also use `__new__`.


````python
class Singleton:
    _instance = None
    
    def __new__(cls, *args, **kwargs) -> 'Singleton':
        if not cls._instance:
            cls._instance = super().__new__(cls)
        return cls._instance

s1 = Singleton()
s2 = Singleton()
print(f"Same instance? {s1 is s2}")

# Pythonic alternative: Just put state in a module!
# config.py
# settings = {}
# Every file that imports config gets the exact same settings dict.

````

Factory pattern using a classmethod and a registry.


````python
from typing import Protocol

class PaymentProcessor(Protocol):
    def pay(self, amount: float) -> None: ...

class StripeProcessor:
    def pay(self, amount: float) -> None:
        print(f"Stripe: {amount}")

class PayPalProcessor:
    def pay(self, amount: float) -> None:
        print(f"PayPal: {amount}")

class ProcessorFactory:
    _registry: dict[str, type[PaymentProcessor]] = {
        "stripe": StripeProcessor,
        "paypal": PayPalProcessor
    }
    
    @classmethod
    def create(cls, provider: str) -> PaymentProcessor:
        if provider not in cls._registry:
            raise ValueError(f"Unknown provider: {provider}")
        return cls._registry[provider]()

p = ProcessorFactory.create("stripe")
p.pay(100.0)

````

Observer pattern. Python makes this trivial with standard sets and functions.


````python
from typing import Callable

ObserverType = Callable[[str], None]

class EventNotifier:
    def __init__(self) -> None:
        self._observers: set[ObserverType] = set()
        
    def subscribe(self, observer: ObserverType) -> None:
        self._observers.add(observer)
        
    def notify(self, message: str) -> None:
        for observer in self._observers:
            observer(message)

def email_alert(msg: str) -> None:
    print(f"Email sent: {msg}")

def log_alert(msg: str) -> None:
    print(f"Log written: {msg}")

notifier = EventNotifier()
notifier.subscribe(email_alert)
notifier.subscribe(log_alert)
notifier.notify("System Rebooting")

````

Strategy pattern. Using first-class functions instead of deep class hierarchies.


````python
from typing import Callable

# Strategy signature
DiscountStrategy = Callable[[float], float]

def no_discount(price: float) -> float:
    return price

def half_off(price: float) -> float:
    return price * 0.5

class ShoppingCart:
    def __init__(self, strategy: DiscountStrategy) -> None:
        self.total = 100.0
        self.strategy = strategy
        
    def checkout(self) -> float:
        return self.strategy(self.total)

cart = ShoppingCart(strategy=half_off)
print(f"Final price: {cart.checkout()}")

````

#### Common Mistakes and Pitfalls
Over-engineering. Writing a `StrategyFactoryProxySingleton` in Python is usually a mistake.


````python
# BAD: Java-style boilerplate in Python
class AbstractStrategy:
    def execute(self) -> None:
        raise NotImplementedError

class ConcreteStrategy(AbstractStrategy):
    def execute(self) -> None:
        print("Executing")

# FIX: Just use a function.
def strategy() -> None:
    print("Executing")

````

#### Best Practices
1. Modules are singletons. Use them for global state when appropriate.
2. First-class functions often replace the need for single-method classes in patterns like Strategy and Command.
3. Use `typing.Protocol` (structural subtyping) to define interfaces for patterns without forcing explicit inheritance.
4. Keep factories simple, often just using dictionaries mapping strings to callable class objects.

#### Real-World Use Case
The Strategy pattern is heavily used in standard Python via the `key` argument in `sorted()`, `min()`, and `max()`. Instead of implementing `Comparable` interfaces on classes, you pass a small function (the strategy) that extracts the sorting metric.

#### Performance Notes
Pythonic patterns (like using functions for Strategy) are generally faster than classic OOP equivalents because they avoid object instantiation and method lookup overhead. Using module-level singletons avoids the `__new__` execution cost entirely.

#### Quick Recap
- Patterns provide common vocabulary but implementations vary by language.
- Python modules act as natural Singletons.
- First-class functions replace heavy class hierarchies in Strategy and Command patterns.
- `typing.Protocol` allows for strict interface definitions without rigid inheritance trees.
- Avoid writing Java/C++ style boilerplate in Python.

#### Interview Questions
1. **Junior**: What is the most Pythonic way to implement a Singleton?
   *Answer*: Creating a module with variables and functions. Python modules are imported only once per runtime, making them natural, thread-safe singletons.
2. **Mid**: How does Python's dynamic typing alter the Factory pattern?
   *Answer*: You don't need a massive switch statement or complex registration. You can store class objects (which are first-class) directly in a dictionary and instantiate them dynamically by calling the dictionary value.
3. **Senior**: Explain the difference between `abc.ABC` and `typing.Protocol` when implementing the Strategy pattern.
   *Answer*: `ABC` is nominal subtyping; the concrete strategy must explicitly inherit from the ABC. `Protocol` is structural subtyping (static duck typing); the concrete strategy doesn't need to inherit anything, it just needs to implement the required methods, and type checkers will validate it.

### Topic 3.13 — Big-O Behavior of Python Built-in Data Structures

#### Concept Overview
Python abstracts away memory management, but performance characteristics remain bound to underlying data structures. Understanding Big-O complexity for core types—`list`, `dict`, `set`, and `deque`—is essential for writing performant code. A list is a dynamic array, meaning appending is fast but inserting in the middle requires shifting elements. Dicts and sets use hash tables, providing constant time lookups on average, but susceptible to hash collisions causing linear degradation. 

Python uses Timsort for list sorting, a hybrid of merge sort and insertion sort, offering O(n log n) average performance and O(n) on nearly sorted data. Choosing the right data structure changes an operation from timing out to executing instantly. Membership testing in lists requires scanning, while sets offer direct addressing via hashing.

#### The Mind Picture
```text
List (Dynamic Array):
[ ptr0 | ptr1 | ptr2 | ptr3 | ptr4 | ... unused capacity ... ]
Index access: Base_Address + (Index * Pointer_Size) -> O(1)
Insert at 0: Shift all elements right -> O(N)

Dict / Set (Hash Table):
Hash(Key) -> Index in sparse array
[ empty | Hash1/Key1/Val1 | empty | empty | Hash2/Key2/Val2 | ... ]
Lookup: Hash the key, go directly to index -> O(1) average
```

#### Syntax Anatomy
| Operation | Structure | Complexity (Average) | Complexity (Worst) | Reason |
|---|---|---|---|---|
| `append()` | `list` | O(1) | O(N) | Amortized constant; occasional reallocation copies all elements. |
| `insert(0, x)` | `list` | O(N) | O(N) | Must shift every existing element right by one slot. |
| `x in data` | `list` | O(N) | O(N) | Linear scan from index 0 to N-1. |
| `x in data` | `set` / `dict`| O(1) | O(N) | Hash lookup; worst case happens on many hash collisions. |
| `append()` / `popleft()` | `collections.deque` | O(1) | O(1) | Doubly-linked list blocks; no shifting required. |
| `sort()` | `list` | O(N log N) | O(N log N) | Timsort algorithm. |

#### Practical Examples
Measuring membership testing differences between list and set.


````python
import timeit
import collections
from typing import List, Set, Dict, Deque

# Example 1: Membership Testing (List vs Set)
N = 1_000_000
test_list: List[int] = list(range(N))
test_set: Set[int] = set(test_list)

# Timing list lookup (worst case, item at end)
list_time = timeit.timeit(lambda: (N - 1) in test_list, number=100)
# Timing set lookup
set_time = timeit.timeit(lambda: (N - 1) in test_set, number=100)

print(f"List lookup: {list_time:.5f}s")
print(f"Set lookup:  {set_time:.5f}s")

````

Using `collections.deque` for fast left-side operations.


````python
# Example 2: Deque vs List for Left Insertions
dq: Deque[int] = collections.deque()
lst: List[int] = []

dq_time = timeit.timeit(lambda: dq.appendleft(1), number=50000)
list_time = timeit.timeit(lambda: lst.insert(0, 1), number=50000)

print(f"Deque appendleft: {dq_time:.5f}s")
print(f"List insert(0):   {list_time:.5f}s")

````

#### Common Mistakes and Pitfalls
Using a list for extensive lookup operations causes massive performance degradation.


````python
from typing import Iterable, List

# BAD: O(N * M) complexity
def find_common_bad(a: Iterable[int], b: List[int]) -> List[int]:
    # Using 'in b' where b is a list means an O(N) scan per item in a.
    return [x for x in a if x in b]

# GOOD: O(N + M) complexity
def find_common_good(a: Iterable[int], b: List[int]) -> List[int]:
    b_set = set(b)  # O(M) to create set
    return [x for x in a if x in b_set]  # O(1) lookup, O(N) total

````

#### Best Practices
1. Default to `list` for ordered collections of homogeneous items where operations happen at the tail.
2. Always convert lookup targets to a `set` before running membership tests (`in`) inside a loop.
3. Use `collections.deque` when implementing queues or needing fast insertions/deletions at both ends.
4. Avoid using `dict.keys()` as a list; use set operations directly on `dict.keys()` objects (view objects).
5. Pre-allocate size when possible, though Python lists handle dynamic growth efficiently.
6. Understand that `del list[0]` is O(N). Never use lists for queue implementations.

#### Real-World Use Case
In web frameworks handling routing (like FastAPI or Flask), endpoint paths are matched against requested URLs. A list of regexes requires an O(N) scan. Exact string matching for static routes uses a dictionary mapping path strings to handler functions, providing O(1) routing regardless of application size.

#### Performance Notes
Python lists over-allocate backing arrays to achieve O(1) amortized append operations. Specifically, they grow by approximately 1.125x when exhausted. Dicts double in size when two-thirds full, rehashing all keys. This means occasionally `append()` or `dict[key] = val` takes O(N) time, but mathematically averages out to O(1) across many operations.

#### Quick Recap
- Lists are dynamic arrays: fast at the end, slow at the beginning.
- Dictionaries and sets use hash tables: fast lookups, fast inserts.
- Deques are doubly-linked lists: fast at both ends, slow in the middle.
- Timsort makes `list.sort()` extremely fast on partially sorted data.
- Amortized complexity means most operations are fast, with occasional slow reallocations.

#### Interview Questions
- **Junior:** Why shouldn't you use a list to check if an item exists out of 10 million records?
  *Answer:* Checking existence in a list is O(N) as it requires a linear scan. A set uses hashing for O(1) average lookup time.
- **Mid:** Explain why `list.append()` is O(1) amortized rather than strictly O(1).
  *Answer:* The list's underlying array has a fixed capacity. Once full, appending requires allocating a larger array and copying all elements (O(N) operation). Because this reallocation happens infrequently, the cost averages out to O(1) per append.
- **Senior:** Describe a scenario where dictionary lookups degrade to O(N) and how you might trigger it intentionally.
  *Answer:* Lookups degrade to O(N) upon numerous hash collisions, where many keys map to the same hash bucket. You could intentionally trigger this by creating custom objects whose `__hash__` method always returns the same constant integer, forcing linear probing or chaining depending on internal implementation.

### Topic 3.14 — Advanced Generators: send(), throw(), yield from

#### Concept Overview
Generators are not just for producing sequences of values; they support a two-way communication channel. Execution pauses at `yield` and the generator can receive data back when resumed. The `.send(value)` method resumes execution and assigns the sent value to the `yield` expression. 

The `.throw(Exception)` method raises an exception inside the generator at the exact point it yielded. `.close()` raises `GeneratorExit` to shut it down cleanly. The `yield from` expression delegates operations to a sub-generator, tunneling `send()`, `throw()`, and `close()` automatically. This two-way communication formed the foundation for asynchronous programming (coroutines) before `async`/`await` became standard.

#### The Mind Picture
```text
Caller                     Generator
  |                            |
  |--- send(None) ------------>| run until yield
  |<-- yields 'A' -------------| PAUSED
  |                            |
  |--- send(42) -------------->| yield evaluates to 42
  |<-- yields 'B' -------------| PAUSED
  |                            |
  |--- throw(ValueError) ----->| ValueError raised at yield
```

#### Syntax Anatomy
| Method/Syntax | Description |
|---|---|
| `gen.send(None)` | Starts the generator (must be None initially). |
| `val = yield x` | Yields `x` to caller; pauses; when resumed via `send(v)`, `val` becomes `v`. |
| `gen.throw(Exc)` | Raises `Exc` at the paused `yield` expression. |
| `gen.close()` | Raises `GeneratorExit` at the paused `yield`. |
| `yield from sub` | Exhausts `sub`, passing yielded values to caller and passing sent values to `sub`. |

#### Practical Examples
A simple two-way generator.


````python
from typing import Generator, Any

def accumulator() -> Generator[int, int, None]:
    total = 0
    while True:
        # yield 'total' to caller, wait for 'value' from caller
        value = yield total
        if value is None:
            break
        total += value

gen = accumulator()
print(gen.send(None))  # Start generator, prints 0
print(gen.send(10))    # total becomes 10, prints 10
print(gen.send(20))    # total becomes 30, prints 30
gen.close()

````

Delegating generators with `yield from` and capturing return values.


````python
def sub_generator() -> Generator[str, int, str]:
    v1 = yield "Give me a number"
    v2 = yield f"Got {v1}, give me another"
    return f"Sum is {v1 + v2}"

def main_generator() -> Generator[str, int, None]:
    print("Delegating...")
    # yield from passes sent values down to sub_generator
    # result gets the Return value (StopIteration.value)
    result = yield from sub_generator()
    print(f"Sub-generator returned: {result}")
    yield "Done"

gen = main_generator()
print(gen.send(None))
print(gen.send(5))
try:
    print(gen.send(7))
except StopIteration:
    pass

````

#### Common Mistakes and Pitfalls
Forgetting to prime the generator with `send(None)` before sending a value.


````python
def simple_gen() -> Generator[None, int, None]:
    x = yield
    print(x)

g = simple_gen()
try:
    # BAD: TypeError: can't send non-None value to a just-started generator
    g.send(42) 
except TypeError as e:
    print(f"Caught error: {e}")

# FIX: Prime it first
g = simple_gen()
g.send(None)
try:
    g.send(42)
except StopIteration:
    pass

````

#### Best Practices
1. Always type hint generators with `Generator[YieldType, SendType, ReturnType]`.
2. Ensure you initialize a two-way generator by calling `next(gen)` or `gen.send(None)`.
3. Use `yield from` whenever delegating to another iterable or generator to avoid manually writing loops that tunnel `send` and `throw`.
4. Handle `GeneratorExit` gracefully if your generator manages external resources.
5. Do not suppress `GeneratorExit`; let it propagate to ensure clean closure.
6. Prefer `async`/`await` for I/O bound concurrency rather than building custom coroutine schedulers via generators.

#### Real-World Use Case
In parsing libraries or state machines, data arrives in chunks. A parsing generator yields its required bytes count and suspends. The network layer reads the bytes and `send()`s them back into the generator. This cleanly separates stateful parsing logic from asynchronous network I/O.

#### Performance Notes
Context switching into and out of a generator is highly optimized in CPython. `yield from` is implemented at the C level, making it significantly faster than a manual Python `for` loop yielding individual items. This optimization handles the complex exception routing and StopIteration unwrapping transparently.

#### Quick Recap
- Generators can consume data using `.send(value)`.
- The first interaction must be `.send(None)` or `next()` to reach the first `yield`.
- `.throw()` injects exceptions at the point of suspension.
- `yield from` transparently delegates all state and communication to a sub-generator.
- Returning a value in a generator attaches it to the `StopIteration` exception, which `yield from` captures.

#### Interview Questions
- **Junior:** What happens if you call `gen.send(10)` on a newly created generator?
  *Answer:* A `TypeError` is raised. A generator must be advanced to its first `yield` via `send(None)` or `next()` before it can accept sent values.
- **Mid:** How does `yield from` differ from a regular `for` loop containing a `yield`?
  *Answer:* A `for` loop only yields values outward. `yield from` establishes a bidirectional channel, passing sent values inward, tunneling exceptions via `throw()`, and capturing the sub-generator's return value.
- **Senior:** Explain how `GeneratorExit` is handled inside a generator and why it is raised.
  *Answer:* Calling `gen.close()` raises `GeneratorExit` at the active `yield`. The generator should clean up resources and exit. If it catches `GeneratorExit` and yields another value instead of exiting, Python raises a `RuntimeError`.

### Topic 3.15 — Advanced typing: generics, Protocol, overload, TypedDict, Literal

#### Concept Overview
Python's type system extends far beyond basic types. `Generic` and `TypeVar` enable classes and functions to operate on multiple types while maintaining type safety. `Protocol` implements structural subtyping (duck typing verified statically); a class fulfills a protocol simply by matching its signature, without inheritance. 

`@overload` handles functions returning different types based on argument types. `TypedDict` defines precise schemas for dictionaries, crucial for parsing JSON APIs. `Literal` constrains variables to specific exact values rather than broad types. `TYPE_CHECKING` provides a safe way to import types solely for static analysis without creating circular dependency crashes at runtime.

#### The Mind Picture
```text
Protocol (Structural Subtyping):
Does Bird have .fly()? Yes. Does Airplane have .fly()? Yes.
Both implicitly satisfy class Flyer(Protocol): def fly(): ...

Generics:
Box[T] -> Box[int] stores ints, Box[str] stores strs.
The type analyzer remembers what you put in the box.
```

#### Syntax Anatomy
| Feature | Usage | Purpose |
|---|---|---|
| `TypeVar('T')` | `T = TypeVar('T')` | Defines a generic type variable. |
| `Generic[T]` | `class Box(Generic[T]):` | Makes a class generic. |
| `Protocol` | `class Drawable(Protocol):` | Defines expected methods statically. |
| `@overload` | `@overload` before `def` | Specifies multiple valid signatures. |
| `TypedDict` | `class User(TypedDict):` | Specifies keys/types in dictionaries. |
| `Literal` | `mode: Literal['r', 'w']` | Restricts parameter to specific values. |

#### Practical Examples
Using Protocols and Generics for flexible, type-safe APIs.


````python
from typing import TypeVar, Protocol, Generic, List

T = TypeVar('T')

# Protocol definition
class Sized(Protocol):
    def __len__(self) -> int: ...

# Function using Protocol
def print_size(item: Sized) -> None:
    print(f"Size is {len(item)}")

# Generic class
class Stack(Generic[T]):
    def __init__(self) -> None:
        self._items: List[T] = []
        
    def push(self, item: T) -> None:
        self._items.append(item)
        
    def pop(self) -> T:
        return self._items.pop()

# Satisfies Sized protocol implicitly
s = Stack[int]()
s.push(1)
print_size([1, 2, 3])  # List satisfies Sized

````

Using `@overload`, `Literal`, and `TypedDict`.


````python
from typing import overload, Literal, TypedDict, Union

class Config(TypedDict):
    host: str
    port: int
    secure: bool

@overload
def fetch_data(mode: Literal['text']) -> str: ...

@overload
def fetch_data(mode: Literal['binary']) -> bytes: ...

def fetch_data(mode: Literal['text', 'binary']) -> Union[str, bytes]:
    if mode == 'text':
        return "data"
    return b"data"

config: Config = {"host": "localhost", "port": 8080, "secure": False}
text_result = fetch_data('text')      # Type checker knows this is str

````

#### Common Mistakes and Pitfalls
Providing implementation logic inside `@overload` definitions.


````python
from typing import overload, Any

# BAD: Overloads should only have `...` or `pass`
@overload
def process(x: int) -> int:
    return x * 2  # Incorrect

# FIX: Overloads define signatures; implementation comes last
@overload
def process_good(x: int) -> int: ...

@overload
def process_good(x: str) -> str: ...

def process_good(x: Any) -> Any:
    return x * 2 if isinstance(x, int) else x + x

````

#### Best Practices
1. Use `Protocol` when writing libraries requiring specific behaviors from objects, rather than forcing consumers to inherit from ABCs.
2. Ensure the actual implementation function immediately follows its `@overload` decorators and accepts a union of all valid types.
3. Use `Literal` for string-based flags (e.g., `'read'` vs `'write'`) to catch typos statically.
4. Bound `TypeVar` using `bound=BaseClass` when the generic type must have certain methods.
5. Avoid deeply nested `Generic` types; they make signatures unreadable and degrade type checker performance.
6. Use `if TYPE_CHECKING:` guards to import types needed only for annotations to prevent circular imports.

#### Real-World Use Case
In SQLAlchemy or Django ORM, query objects are heavily typed using `Generic[Model]`. When you write `session.query(User).first()`, the type system knows it returns a `User` object because `query()` accepts `Type[T]` and returns `Query[T]`. `TypedDict` is extensively used when handling parsed JSON from external APIs matching a known static schema.

#### Performance Notes
Type hints generally impose zero runtime performance cost because Python evaluates them at import time but does not enforce them during execution. However, massive typing structures can slow down import times. Features like `typing.get_type_hints()` do reflection at runtime and can be slow.

#### Quick Recap
- `Generic` and `TypeVar` preserve type identity through classes and functions.
- `Protocol` offers duck typing with rigid static checks (structural subtyping).
- `@overload` clarifies return types for dynamic parameters.
- `TypedDict` enforces dict schemas.
- `TYPE_CHECKING` breaks circular dependencies for imports.

#### Interview Questions
- **Junior:** What is the difference between `Literal['r', 'w']` and `str`?
  *Answer:* `str` accepts any string, whereas `Literal['r', 'w']` restricts the argument strictly to those two exact string values, catching errors before runtime.
- **Mid:** How does a `Protocol` differ from an Abstract Base Class (ABC)?
  *Answer:* An ABC requires explicit inheritance (`class MyClass(MyABC):`). A Protocol checks structural compatibility implicitly; if a class has the required methods, it satisfies the Protocol automatically.
- **Senior:** How do you resolve a circular import issue caused strictly by type annotations?
  *Answer:* Import the `typing.TYPE_CHECKING` boolean constant. Place the conflicting imports inside an `if TYPE_CHECKING:` block. During execution, this block is false, avoiding the circular import. Type checkers evaluate it as true.

### Topic 3.16 — Dataclasses

#### Concept Overview
The `@dataclass` decorator, introduced in Python 3.7, automatically generates boilerplate methods like `__init__`, `__repr__`, and `__eq__` based on class attributes with type hints. It is the standard library's answer to writing classes that primarily hold data.

Dataclasses handle mutable defaults correctly using `field(default_factory=...)`, preventing the shared-state trap. Setting `frozen=True` creates immutable instances that are automatically hashable, allowing them to be used as dictionary keys. `__post_init__` handles custom validation logic immediately after the auto-generated initialization. They provide a massive productivity boost over manual boilerplate.

#### The Mind Picture
```text
Without Dataclass:                 With Dataclass:
class Point:                       @dataclass
  def __init__(self, x, y):        class Point:
      self.x = x                       x: int
      self.y = y                       y: int
  def __repr__(self):              
      return f"Point({self.x})"    (Python writes __init__, 
  def __eq__(self, other):          __repr__, __eq__ automatically)
      return self.x == other.x
```

#### Syntax Anatomy
| Parameter / Tool | Description |
|---|---|
| `@dataclass` | Decorator to trigger code generation. |
| `frozen=True` | Makes attributes read-only; generates `__hash__`. |
| `order=True` | Generates `__lt__`, `__le__`, `__gt__`, `__ge__`. |
| `field(...)` | Customizes field behavior (default factory, hide from repr). |
| `__post_init__` | Hook for executing code after the generated `__init__`. |
| `InitVar` | Field used only in `__init__`/`__post_init__`, not stored. |

#### Practical Examples
Basic usage, mutable defaults, and freezing.


````python
from dataclasses import dataclass, field
from typing import List

@dataclass
class InventoryItem:
    name: str
    unit_price: float
    # Mutable defaults MUST use default_factory to avoid sharing across instances
    tags: List[str] = field(default_factory=list)

@dataclass(frozen=True, order=True)
class Coordinate:
    x: int
    y: int

item1 = InventoryItem("Widget", 3.50)
item2 = InventoryItem("Gadget", 5.00, ["sale"])
item1.tags.append("new")

c1 = Coordinate(1, 2)
c2 = Coordinate(3, 4)
print(c1 < c2)  # True, derived from order=True comparing fields sequentially

````

Using `__post_init__` for validation and `InitVar`.


````python
from dataclasses import dataclass, InitVar

@dataclass
class User:
    username: str
    email: str
    # InitVar is passed to __post_init__ but not saved as self.password
    password: InitVar[str]
    hashed_password: str = field(init=False)

    def __post_init__(self, password: str) -> None:
        if "@" not in self.email:
            raise ValueError("Invalid email")
        # Fake hashing mechanism
        self.hashed_password = f"hash_{password}"

user = User("alice", "alice@example.com", "secret123")
print(user.hashed_password)

````

#### Common Mistakes and Pitfalls
Using a mutable object as a default value directly.


````python
from dataclasses import dataclass
from typing import List

# BAD: TypeError will be raised by dataclass decorator
# @dataclass
# class BrokenItem:
#     tags: List[str] = []  # ValueError: mutable default

# FIX: Use field(default_factory=)
from dataclasses import field

@dataclass
class WorkingItem:
    tags: List[str] = field(default_factory=list)

````

#### Best Practices
1. Use `frozen=True` by default unless you explicitly need mutation. Immutable data reduces bugs.
2. Always use `field(default_factory=list)` (or `dict`, `set`) for mutable defaults.
3. Use `InitVar` for parameters required to construct the object but unnecessary to store.
4. Omit fields from `__repr__` using `field(repr=False)` if they contain sensitive data (passwords) or huge structures.
5. Use plain classes when behavior (methods) vastly outweighs state (data).
6. Use Pydantic over dataclasses if you require automatic runtime data parsing and strict validation (e.g., API boundaries).

#### Real-World Use Case
In configuring a test environment, multiple settings parameters (DB string, timeouts, feature flags) are grouped into a `@dataclass(frozen=True)`. This ensures settings cannot be accidentally mutated mid-test, and the auto-generated `__repr__` makes logging the exact test configuration effortless.

#### Performance Notes
Dataclasses are fast because the generated `__init__` is standard Python code; there is no metaclass overhead or proxy wrapper dynamically intercepting assignments. Applying `@dataclass(slots=True)` (Python 3.10+) reduces memory usage per instance by creating `__slots__` instead of an instance `__dict__`, speeding up attribute access.

#### Quick Recap
- `@dataclass` generates structural boilerplate based on type hints.
- Protect mutable defaults with `field(default_factory=...)`.
- `frozen=True` provides immutability and hashability.
- `order=True` creates comparison operators.
- `__post_init__` executes logic right after assignment.
- They are standard Python classes under the hood, with zero runtime penalty compared to manual implementations.

#### Interview Questions
- **Junior:** What does `@dataclass` do for you?
  *Answer:* It automatically writes boilerplate methods like `__init__`, `__repr__`, and `__eq__` based on class attributes, saving time and preventing typos.
- **Mid:** How do you handle a list attribute with an empty list as the default in a dataclass?
  *Answer:* You must use `tags: list = field(default_factory=list)`. Setting it directly to `[]` creates a shared list across all instances, which Python prevents by raising a ValueError.
- **Senior:** Compare standard library `@dataclass` with Pydantic's `BaseModel`. When would you choose one over the other?
  *Answer:* `@dataclass` assumes data is already correct and performs no runtime validation or coercion; it is lightweight and fast, ideal for internal data structures. Pydantic's `BaseModel` actively parses, coerces, and validates data at runtime. Use Pydantic at the application boundaries (like parsing JSON from a web request), and dataclasses for internal application logic.



### Topic 3.17 — Introspection

#### Concept Overview
Introspection is the ability of a program to examine its own state, types, and properties at runtime. While statically typed languages like C++ or Java require you to know a lot about an object at compile time, Python's dynamic nature means everything is an object—and every object carries its metadata with it.

Python added introspection capabilities because of its "duck typing" philosophy. If it walks like a duck, it's a duck. But sometimes you actually need to know if the object you just received from a third-party library has a `quack` method before you call it, or what arguments that method expects. Introspection solves the problem of dynamic discovery, allowing you to build flexible systems like plugins, ORMs, and dependency injectors that figure out how to wire things up on the fly.


#### The Mind Picture
Every Python object is essentially a struct containing a pointer to its type object (which defines its behavior) and a dictionary (or slots) holding its attributes. When you introspect an object, you are peeking under the hood at these internal structures.

```text
+-----------------------+
|      MyObject()       |
+-----------------------+
| __class__  ---------->| +-----------------------+
| __dict__   ---------->| |      MyClass          |
|                       | +-----------------------+
|                       | | __bases__: (object,)  |
+-----------------------+ | __dict__: {           |
                          |   'method1': <func>,  |
                          |   'attr1': 'value'    |
                          | }                     |
                          +-----------------------+
```


#### Syntax Anatomy
Python provides built-in functions and the `inspect` module for introspection. Here is the core syntax you will use every day.

```python
# Basic introspection
dir(obj)            # Returns a list of all attribute names on the object
vars(obj)           # Returns the __dict__ containing instance attributes
type(obj)           # Returns the class of the object
isinstance(obj, cls)# True if obj is an instance of cls or a subclass
issubclass(cls, b)  # True if cls is a subclass of base b
callable(obj)       # True if the object can be called like a function

# Dynamic attribute access
getattr(obj, 'name', default) # Fetches obj.name, returning default if not found
setattr(obj, 'name', value)   # Sets obj.name = value
hasattr(obj, 'name')          # True if obj has attribute 'name'
delattr(obj, 'name')          # Deletes obj.name

# Advanced introspection via inspect module
inspect.signature(func)       # Gets the exact argument signature of a function
inspect.getsource(func)       # Returns the source code of the function as a string
inspect.getmembers(obj, pred) # Returns all (name, value) pairs matching predicate
inspect.isfunction(obj)       # True if obj is a function (there are many is* methods)
```


#### Practical Examples
Let's start with basic built-in introspection to examine objects dynamically.


````python
from typing import Any

class DatabaseConnection:
    def __init__(self, host: str, port: int) -> None:
        self.host = host
        self.port = port
        
    def connect(self) -> bool:
        return True

def analyze_object(obj: Any) -> None:
    print(f"Type: {type(obj).__name__}")
    print(f"Is callable? {callable(obj)}")
    if hasattr(obj, '__dict__'):
        print(f"Instance vars: {vars(obj)}")
    print(f"Methods available: {[attr for attr in dir(obj) if callable(getattr(obj, attr)) and not attr.startswith('__')]}")

db = DatabaseConnection("localhost", 5432)
analyze_object(db)
````

Dynamic dispatch is a pattern where you determine which method to call at runtime based on string names. This avoids massive `if-elif` chains.


````python
class ReportGenerator:
    def generate_pdf(self) -> str:
        return "PDF Content"
        
    def generate_csv(self) -> str:
        return "CSV Content"
        
    def generate_json(self) -> str:
        return "JSON Content"

def export_report(generator: ReportGenerator, format_type: str) -> str:
    method_name = f"generate_{format_type}"
    # Safely get the method, defaulting to None if it doesn't exist
    method = getattr(generator, method_name, None)
    
    if method and callable(method):
        return method()
    raise ValueError(f"Unsupported format: {format_type}")

gen = ReportGenerator()
print(export_report(gen, "csv"))
# print(export_report(gen, "xml")) # Raises ValueError
````

The `inspect` module provides deeper insights, such as reading type hints and function signatures at runtime. This is how frameworks like FastAPI validate endpoints.


````python
import inspect
from typing import get_type_hints

def process_payment(amount: float, currency: str = "USD") -> bool:
    '''Process a payment transaction.'''
    return True

# 1. Get the signature
sig = inspect.signature(process_payment)
print(f"Signature: {sig}")

# 2. Inspect parameters
for name, param in sig.parameters.items():
    print(f"Param: {name}, Default: {param.default}, Kind: {param.kind}")

# 3. Read type hints (evaluated)
hints = get_type_hints(process_payment)
print(f"Type hints: {hints}")

# 4. Read source code
print(f"Source code:\n{inspect.getsource(process_payment)}")
````

#### Common Mistakes and Pitfalls
A common mistake is catching `AttributeError` instead of using `hasattr`, or worse, failing to provide a default to `getattr`.


````python
# WRONG
class Config:
    timeout = 30

conf = Config()
# If 'retries' doesn't exist, this throws an unhandled exception if we just use it
# value = getattr(conf, 'retries') # Raises AttributeError

# RIGHT
value = getattr(conf, 'retries', 3) # Safe fallback
print(f"Retries: {value}")
````

Another pitfall is misunderstanding `dir()` vs `vars()`. `dir()` gives all attributes including inherited class methods, while `vars()` only gives instance variables.


````python
class Base:
    base_attr = 1
    
class Child(Base):
    def __init__(self) -> None:
        self.child_attr = 2

# WRONG context usage
c = Child()
# Thinking vars() gives you everything
print("Vars:", vars(c)) # Only {'child_attr': 2}

# RIGHT context usage
# Use dir() to see what you can actually access on the object
print("Dir contains base_attr?", 'base_attr' in dir(c))
````

#### Best Practices
- **Prefer `isinstance()` over `type(obj) == cls`**: `isinstance` correctly handles inheritance, while `type()` checking breaks for subclasses.
- **Use `getattr` with defaults**: Always provide a sensible default to `getattr` to avoid `AttributeError` exceptions, unless you explicitly want to fail.
- **Cache `inspect` calls**: The `inspect` module functions (like `getsource` or `signature`) can be slow. Cache their results if you are doing introspection in a hot loop.
- **Respect dunder privacy**: Avoid calling `__dunder__` methods directly or relying on `_private` attributes unless absolutely necessary for a hack or patch.


#### Real-World Use Case
Building a simple plugin system where you discover and load all classes in a module that subclass a specific base class.


````python
import inspect
import sys

# Simulate a module with various classes
class PluginBase:
    pass

class AuthPlugin(PluginBase):
    name = "auth"

class LoggingPlugin(PluginBase):
    name = "logging"

class NotAPlugin:
    pass

def discover_plugins(module_name: str) -> list[type[PluginBase]]:
    module = sys.modules[module_name]
    plugins = []
    
    # inspect.getmembers lets us filter the module namespace
    for name, obj in inspect.getmembers(module, inspect.isclass):
        if issubclass(obj, PluginBase) and obj is not PluginBase:
            plugins.append(obj)
            
    return plugins

# Call on current module (__main__)
found_plugins = discover_plugins(__name__)
print(f"Found plugins: {[p.name for p in found_plugins]}")
````

#### Performance Notes
Introspection is inherently slow compared to direct attribute access or static typing. Calling `hasattr` or `getattr` adds dictionary lookup overhead. The `inspect` module is even slower because it often parses Python bytecode or source files. Do not use introspection in the inner loops of performance-critical code; resolve dynamically at startup, then cache the references.


#### Quick Recap
- `dir()` lists all accessible attribute names (including inherited).
- `vars()` returns the instance `__dict__`.
- `getattr`, `setattr`, `hasattr`, `delattr` allow string-based attribute manipulation.
- `isinstance` is the correct way to check types, supporting inheritance.
- `callable()` tells you if an object can be invoked.
- The `inspect` module is for heavy-lifting: signatures, source code, and module member discovery.
- `typing.get_type_hints()` evaluates type annotations safely at runtime.


#### Interview Questions
- **Junior**: What is the difference between `type(obj) == MyClass` and `isinstance(obj, MyClass)`?  
  **Answer**: `isinstance` accounts for inheritance, returning `True` if `obj` is an instance of `MyClass` or any subclass. `type(obj) == MyClass` strictly checks for exact class identity, failing on subclasses. `isinstance` is almost always preferred.
- **Mid**: How would you safely invoke a method on an object if you only have the method's name as a string?  
  **Answer**: Use `method = getattr(obj, method_name, None)`. Check if `callable(method)` is True, then invoke it `method()`. This pattern is called dynamic dispatch.
- **Senior**: Explain how a framework like FastAPI uses Python's introspection and type hints to build its API routing and validation.  
  **Answer**: It uses `inspect.signature(endpoint_func)` to get the parameter names and defaults. It then uses `typing.get_type_hints(endpoint_func)` to read the type annotations for those parameters. When a request arrives, it parses the JSON/Query parameters, coerces them to the types defined in the hints using Pydantic, and dynamically passes them to the endpoint function using `**kwargs` matching the parameter names.


### Topic 3.18 — Logging Done Properly

#### Concept Overview
Logging is how your application communicates its state, warnings, and errors to the outside world. Using `print()` is fine for simple scripts, but in production, `print()` is a liability: it doesn't rotate files, it doesn't provide severity levels, and it blocks execution unnecessarily. 

The standard library `logging` module solves this by providing a robust, thread-safe hierarchy of loggers, handlers, and formatters. You define *what* severity a message has, and the configuration determines *where* it goes and *how* it looks.


#### The Mind Picture
The logging system is a tree. Events bubble up from child loggers to parent loggers until they hit a handler that writes them out.

```text
[ Logger: 'app.database' ] --(emit)--> [ Filter ]
        |                                   |
    (propagate)                             v
        v                         [ Handler: FileHandler ]
[ Logger: 'app' ]                           |
        |                                   v
    (propagate)                     [ Formatter ]
        v                                   |
[ Root Logger ]                             v
        |---> [ Handler: StreamHandler ] -> stdout / my_app.log
```


#### Syntax Anatomy
Logging involves setting up loggers, setting levels, and sending messages.

```python
import logging

# 1. Get a logger for the current module
logger = logging.getLogger(__name__)

# 2. Set the severity level (DEBUG, INFO, WARNING, ERROR, CRITICAL)
# logger.setLevel(logging.INFO)

# 3. Log messages
logger.debug("Fine-grained info for devs")
logger.info("Standard operational messages")
logger.warning("Something unexpected happened, but we recovered")
logger.error("A task failed")
logger.critical("The system is going down")
logger.exception("Logs ERROR level + the stack trace")
```


#### Practical Examples
Let's start with setting up a basic logger for an application entry point. We use `dictConfig` because it is the most predictable way to configure logging.


````python
import logging
import logging.config

LOGGING_CONFIG = {
    'version': 1,
    'disable_existing_loggers': False,
    'formatters': {
        'standard': {
            'format': '%(asctime)s [%(levelname)s] %(name)s: %(message)s'
        },
    },
    'handlers': {
        'console': {
            'level': 'INFO',
            'formatter': 'standard',
            'class': 'logging.StreamHandler',
            'stream': 'ext://sys.stdout',
        },
    },
    'loggers': {
        '': {  # root logger
            'handlers': ['console'],
            'level': 'WARNING',
            'propagate': False
        },
        'my_app': { 
            'handlers': ['console'],
            'level': 'DEBUG',
            'propagate': False
        },
    }
}

logging.config.dictConfig(LOGGING_CONFIG)
logger = logging.getLogger('my_app.service')

logger.debug("Connecting to database...")
logger.info("Service started successfully.")
````

In library code, you should **never** configure logging. You only emit logs. You attach a `NullHandler` to prevent 'No handler found' warnings if the user doesn't configure logging.


````python
import logging

# In my_awesome_lib/client.py
logger = logging.getLogger(__name__)
# Add NullHandler so this logger doesn't complain if app doesn't configure logging
logger.addHandler(logging.NullHandler())

class APIClient:
    def fetch_data(self) -> dict[str, str]:
        logger.debug("Fetching data from API...")
        # Simulate network error
        logger.error("Failed to fetch data due to timeout.")
        return {}

# Note: No basicConfig or dictConfig here!
````

Structured logging (adding context without messing up the string message) using `extra`.


````python
import logging
import sys

logger = logging.getLogger("structured_app")
handler = logging.StreamHandler(sys.stdout)
# Formatter explicitly expects 'user_id' and 'ip'
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - User:%(user_id)s IP:%(ip)s - %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)
logger.setLevel(logging.INFO)

# Passing the required extra context
logger.info("User logged in", extra={"user_id": 42, "ip": "192.168.1.1"})
````

#### Common Mistakes and Pitfalls
Calling `logging.basicConfig()` in multiple modules or inside library code. `basicConfig` only works once; subsequent calls do nothing.


````python
import logging

# WRONG: Doing this in a library or a sub-module
# logging.basicConfig(level=logging.DEBUG)
# logger.info("This forces the whole app to use DEBUG, angering users")

# RIGHT: Do this ONLY in your main.py or __main__ block
if __name__ == "__main__":
    logging.basicConfig(
        level=logging.INFO,
        format='%(asctime)s %(levelname)s: %(message)s'
    )
````

Failing to use `logger.exception()` in except blocks, which results in losing the stack trace.


````python
import logging
logger = logging.getLogger(__name__)

# WRONG
try:
    1 / 0
except ZeroDivisionError as e:
    logger.error(f"Failed: {e}") # Stack trace is lost forever!

# RIGHT
try:
    1 / 0
except ZeroDivisionError:
    logger.exception("Failed to compute division") # Automatically attaches full traceback
````

#### Best Practices
- **Use `logging.getLogger(__name__)`**: This automatically creates a hierarchical logger based on your module path (e.g., `package.module`).
- **Never configure logging in library code**: Use `NullHandler` and let the application entry point dictate handlers and formats.
- **Use lazy formatting**: Write `logger.info("User %s", user)` instead of `logger.info(f"User {user}")`. The former skips string interpolation entirely if the INFO level is disabled, saving CPU cycles.
- **Use `logging.config.dictConfig`**: It is significantly more powerful and readable than `basicConfig` or file-based configs.
- **Log exceptions with `logger.exception`**: It automatically includes the traceback.


#### Real-World Use Case
Rotating log files are critical for long-running daemon processes to prevent filling up the disk.


````python
import logging
import logging.handlers
import sys

logger = logging.getLogger("daemon_logger")
logger.setLevel(logging.INFO)

# Rotate file when it hits 1MB, keeping 3 backup copies
file_handler = logging.handlers.RotatingFileHandler(
    "daemon.log", maxBytes=1024*1024, backupCount=3
)
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

# Also output to stdout for development
stdout_handler = logging.StreamHandler(sys.stdout)
stdout_handler.setFormatter(formatter)
logger.addHandler(stdout_handler)

logger.info("Daemon started. Logs are writing to daemon.log with rotation.")
````

#### Performance Notes
Logging is an I/O bound operation. Extensive logging (like DEBUG level in production) will slow down your application due to disk writes and string allocations. By using lazy formatting (`%s` style instead of f-strings), the logger skips evaluating the string entirely if the logger's level is higher than the message's level. For extreme high-performance applications, consider asynchronous logging via the `QueueHandler`.


#### Quick Recap
- Levels matter: DEBUG < INFO < WARNING < ERROR < CRITICAL.
- Loggers form a tree structure via dot-notation (e.g., `app.db`).
- Handlers route logs to destinations (console, files, syslog).
- Formatters shape the text output.
- `dictConfig` is the preferred way to configure loggers in applications.
- Libraries should only define loggers and add a `NullHandler`.
- Use `logger.exception` inside `except` blocks to capture stack traces.


#### Interview Questions
- **Junior**: Why should you use `logger.info("User %s", name)` instead of `logger.info(f"User {name}")`?  
  **Answer**: Using `%s` defers the string interpolation to the logging module. If the logger is set to WARNING, the INFO message is discarded before the string is constructed, saving CPU time and memory.
- **Mid**: You are writing a pip package. How should you set up logging in your `__init__.py`?  
  **Answer**: Instantiate a logger using `__name__` and attach a `logging.NullHandler()`. I will not call `basicConfig` or attach `StreamHandlers`, as that is the responsibility of the application consuming my library.
- **Senior**: How does the `propagate` attribute work in the logging module?  
  **Answer**: `propagate=True` means events logged to a child logger are passed up to the parent loggers' handlers. If `app.module` logs an error, it gets handled by `app.module`'s handlers, and then passed to `app`'s handlers, and then the root handler. Setting `propagate=False` stops this bubble-up, preventing duplicate log entries.


### Topic 3.19 — Packaging and Distribution

#### Concept Overview
Writing code is only half the battle; the other half is getting it onto other developers' machines. Python's packaging ecosystem used to be a fragmented mess of `setup.py` scripts executing arbitrary code at install time. PEP 517, 518, and 621 standardized packaging into a declarative format: `pyproject.toml`.

By using `pyproject.toml`, you define metadata, dependencies, and build requirements statically. This makes builds faster, safer, and tooling-independent. You can choose any build backend (Hatch, Setuptools, Flit) without changing how users install your package.


#### The Mind Picture
The modern build pipeline transforms source code into a wheel (binary distribution) via a standard interface.

```text
+------------------+     PEP 517     +-----------------+
| pyproject.toml   | --(Front-end)-> |  Build Backend  | (e.g. hatchling)
| src/mypackage/   |  (e.g. build)   +-----------------+
+------------------+                          |
                                              v
                                   +-----------------------+
                                   | dist/                 |
                                   |  mypackage-1.0.tar.gz | (sdist)
                                   |  mypackage-1.0.whl    | (wheel)
                                   +-----------------------+
```


#### Syntax Anatomy
The `pyproject.toml` file is divided into tables (sections).

```toml
# 1. Build System Setup (PEP 518)
[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"

# 2. Project Metadata (PEP 621)
[project]
name = "my-awesome-tool"
version = "1.0.0"
description = "Does awesome things"
readme = "README.md"
requires-python = ">=3.11"
dependencies = [
    "requests>=2.31.0",
]

# 3. Entry points (CLI tools)
[project.scripts]
my-cli = "mypackage.cli:main"
```


#### Practical Examples
Let's structure a modern Python project using the `src/` layout. The `src` layout prevents accidental imports of the local source directory when testing, forcing tests to run against the built package.


````python
# Directory Structure:
# my_project/
# ├── pyproject.toml
# ├── src/
# │   └── mypackage/
# │       ├── __init__.py
# │       └── core.py
# └── tests/
#     └── test_core.py

import os

# Simulating writing the pyproject.toml
toml_content = '''
[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"

[project]
name = "math-utils-demo"
version = "0.1.0"
description = "A simple math utility"
requires-python = ">=3.11"
dependencies = []

[project.scripts]
math-cli = "mypackage.core:main"
'''
print("pyproject.toml configured for Hatchling backend.")
````

Building the package requires the standard `build` frontend. You should never invoke `setup.py` directly.


````python
import subprocess
import sys

def build_package():
    print("Run this in your terminal:")
    print("python -m pip install build")
    print("python -m build")
    print("\nThis produces:")
    print("dist/math_utils_demo-0.1.0-py3-none-any.whl")
    print("dist/math_utils_demo-0.1.0.tar.gz")

build_package()
````

Publishing to PyPI uses Twine. Always test on TestPyPI first.


````python
def publish_package():
    print("To upload securely using Twine:")
    print("python -m pip install twine")
    print("\n# 1. Upload to TestPyPI first")
    print("twine upload --repository testpypi dist/*")
    print("\n# 2. Verify installation")
    print("python -m pip install --index-url https://test.pypi.org/simple/ math-utils-demo")
    print("\n# 3. Upload to actual PyPI")
    print("twine upload dist/*")

publish_package()
````

#### Common Mistakes and Pitfalls
A major pitfall is using the 'flat layout' (putting the package directory next to `pyproject.toml`) and then running `pytest`. Python will import the local directory instead of the installed package, masking issues with missing package data or faulty installation.


````python
# WRONG (Flat Layout)
# my_project/
# ├── mypackage/
# │   └── __init__.py
# ├── pyproject.toml
# └── tests/

# RIGHT (Src Layout)
# my_project/
# ├── src/
# │   └── mypackage/
# │       └── __init__.py
# ├── pyproject.toml
# └── tests/
````

Another mistake is hardcoding absolute versions in `dependencies` inside `pyproject.toml`. Libraries should use broad version ranges, while applications should pin exact versions in a `requirements.txt` or `Pipfile.lock`.


````python
# WRONG (in pyproject.toml for a library)
# dependencies = ["requests==2.31.0"]  # Too restrictive, causes dependency conflicts

# RIGHT
# dependencies = ["requests>=2.28.0,<3.0.0"] # Semantic versioning friendly
````

#### Best Practices
- **Use `pyproject.toml`**: `setup.py` and `setup.cfg` are legacy. Move everything to `pyproject.toml`.
- **Use the `src/` layout**: It is the industry standard to prevent local import masking.
- **Semantic Versioning**: Follow `MAJOR.MINOR.PATCH`. Increment MAJOR for breaking changes, MINOR for features, PATCH for bugfixes.
- **Include a README and License**: The `[project]` table should point to these so they render on PyPI.
- **Use `twine` for uploads**: Never use `python setup.py upload`, it is insecure and deprecated.


#### Real-World Use Case
Defining an entry point for a CLI application. When a user `pip installs` your package, pip reads `[project.scripts]` and automatically creates an executable shim in the user's `bin` or `Scripts` folder.


````python
# In pyproject.toml
'''
[project.scripts]
my-linter = "my_linter.cli:main"
'''

# In src/my_linter/cli.py
def main() -> int:
    print("Running linter...")
    return 0

# When installed, the user can just type:
# $ my-linter
# in their terminal.
````

#### Performance Notes
Packaging has little runtime performance impact, but using Wheels (`.whl`) dramatically improves installation performance. A source distribution (`.tar.gz`) requires the user's machine to build the package, which is extremely slow for packages with C-extensions (like NumPy). Distributing pre-compiled Wheels for different OS/architectures saves end-users massive amounts of time.


#### Quick Recap
- `pyproject.toml` is the single source of truth for modern Python packaging.
- The `[build-system]` defines the backend (e.g., Hatch, Setuptools).
- The `[project]` defines the metadata and dependencies.
- `python -m build` generates Wheels and source distributions.
- `twine` handles secure uploads to PyPI.
- The `src/` directory layout is strongly recommended to isolate source code from test environments.
- `[project.scripts]` easily exposes python functions as command-line tools.


#### Interview Questions
- **Junior**: What is the difference between a source distribution (`sdist`) and a wheel (`whl`)?  
  **Answer**: An `sdist` contains the raw source code and requires the package to be built on the user's machine during installation. A wheel is a pre-built binary distribution. Installing a wheel is much faster because it just needs to be unpacked into `site-packages`.
- **Mid**: Why should you use the `src` layout instead of placing your package directory directly in the project root?  
  **Answer**: To prevent Python from importing the local directory when running tests from the project root. The `src` layout forces you to test the package exactly as it is installed in `site-packages`, catching missing files or build issues.
- **Senior**: Explain the roles of PEP 517 and PEP 518 in the Python packaging ecosystem.  
  **Answer**: Prior to these PEPs, `setup.py` was assumed to be executable and heavily relied on `setuptools` being installed. PEP 518 introduced `pyproject.toml` to declare build dependencies *before* the build starts. PEP 517 defined a standard API for build backends. Together, they decouple the packaging frontend (like `pip` or `build`) from the backend, allowing alternatives like `flit` or `hatch` to flourish.


### Topic 3.20 — Talking to the Outside World: sqlite3, requests, secrets/hashing

#### Concept Overview

#### Practical Examples
Let's interact with a local `sqlite3` database securely. We use the connection as a context manager to automatically commit or rollback transactions.


````python
import sqlite3
from typing import Any

def setup_db(db_path: str = ':memory:') -> sqlite3.Connection:
    # row_factory allows accessing columns by name (like a dict)
    conn = sqlite3.connect(db_path)
    conn.row_factory = sqlite3.Row
    
    with conn:
        conn.execute('''
            CREATE TABLE IF NOT EXISTS users (
                id INTEGER PRIMARY KEY,
                username TEXT UNIQUE NOT NULL,
                api_key TEXT
            )
        ''')
    return conn

def add_user(conn: sqlite3.Connection, username: str, api_key: str) -> int | None:
    try:
        with conn: # Handles transaction commit/rollback
            cursor = conn.execute(
                "INSERT INTO users (username, api_key) VALUES (?, ?)",
                (username, api_key)
            )
            return cursor.lastrowid
    except sqlite3.IntegrityError:
        print("User already exists!")
        return None

conn = setup_db()
add_user(conn, "alice", "key_123")
````

Making HTTP requests properly means handling timeouts and checking status codes. A `Session` pools connections for faster subsequent requests.


````python
import requests
from requests.exceptions import RequestException
from typing import Any

def fetch_github_user(username: str) -> dict[str, Any] | None:
    # Use a session for connection pooling
    with requests.Session() as session:
        session.headers.update({"Accept": "application/vnd.github.v3+json"})
        
        try:
            # ALWAYS define a timeout
            resp = session.get(
                f"https://api.github.com/users/{username}", 
                timeout=5.0
            )
            resp.raise_for_status() # Raises HTTPError for bad responses
            return resp.json()
        except RequestException as e:
            print(f"Network error occurred: {e}")
            return None

# data = fetch_github_user("torvalds")
# if data: print(data['name'])
````

Handling passwords requires generating salts and using a slow hashing algorithm like `scrypt`. Never use `sha256` for passwords; it's too fast and vulnerable to brute force.


````python
import hashlib
import secrets
import base64

def hash_password(password: str) -> str:
    # Generate 16 secure random bytes for the salt
    salt = secrets.token_bytes(16)
    
    # scrypt is designed to be slow and memory-hard
    hashed = hashlib.scrypt(
        password.encode('utf-8'),
        salt=salt,
        n=16384,
        r=8,
        p=1
    )
    
    # Store salt and hash together
    combined = salt + hashed
    return base64.b64encode(combined).decode('utf-8')

def verify_password(stored_hash_b64: str, password: str) -> bool:
    combined = base64.b64decode(stored_hash_b64)
    salt = combined[:16]
    stored_hash = combined[16:]
    
    # Re-hash provided password with the extracted salt
    new_hash = hashlib.scrypt(
        password.encode('utf-8'),
        salt=salt,
        n=16384,
        r=8,
        p=1
    )
    
    # Use hmac.compare_digest to prevent timing attacks
    import hmac
    return hmac.compare_digest(stored_hash, new_hash)

h = hash_password("super_secret_123")
print("Verified:", verify_password(h, "super_secret_123"))
````

#### Common Mistakes and Pitfalls
The most dangerous mistake in Python is string formatting SQL queries. This leads directly to SQL injection.


````python
# WRONG (SQL INJECTION VULNERABILITY)
def get_user_bad(conn, username):
    # If username is "admin' --", the query becomes:
    # SELECT * FROM users WHERE username='admin' --'
    query = f"SELECT * FROM users WHERE username='{username}'"
    return conn.execute(query).fetchall()

# RIGHT
def get_user_good(conn, username):
    # The database driver safely escapes the parameter
    query = "SELECT * FROM users WHERE username=?"
    return conn.execute(query, (username,)).fetchall()
````

Omitting the `timeout` parameter in `requests` is a guaranteed way to bring down a production service. The default timeout is infinity, meaning a hanging server will freeze your worker indefinitely.


````python
import requests

# WRONG
# response = requests.get('https://example.com') # Will block forever if server hangs

# RIGHT
# response = requests.get('https://example.com', timeout=(3.0, 10.0)) # (connect_timeout, read_timeout)
````

#### Best Practices
- **SQLite**: Always use `?` parameterization. Never use f-strings or `.format()` for SQL values. Set `conn.row_factory = sqlite3.Row` to get dictionary-like rows.
- **Requests**: Always pass `timeout=5.0`. Call `response.raise_for_status()` to fail fast on HTTP errors. Use `requests.Session()` if calling the same host multiple times to reuse the TCP connection.
- **Secrets**: Use the `secrets` module for tokens, session IDs, and salts. `random.randint` is deterministic and predictable.
- **Hashing**: Use `hashlib.scrypt` or the external `passlib` library for passwords. Store the salt alongside the hash. Compare hashes using `hmac.compare_digest` to prevent timing attacks.


#### Real-World Use Case
Building a robust API client wrapper that authenticates using a secure token and manages retries/timeouts.


````python
import requests
import secrets

class SecureAPIClient:
    def __init__(self, base_url: str):
        self.base_url = base_url.rstrip('/')
        self.session = requests.Session()
        # Generate a secure session ID for tracking
        self.session_id = secrets.token_hex(8)
        self.session.headers.update({"X-Session-ID": self.session_id})
        
    def get_resource(self, endpoint: str) -> dict:
        url = f"{self.base_url}/{endpoint.lstrip('/')}"
        # Always use timeouts
        response = self.session.get(url, timeout=5.0)
        # Always check for HTTP errors
        response.raise_for_status()
        return response.json()

# client = SecureAPIClient("https://api.example.com")
# data = client.get_resource("/users/1")
````

#### Performance Notes
- **SQLite**: Disk I/O bound. Batch inserts using `executemany()` are orders of magnitude faster than a loop of `execute()`. Wrapping operations in a transaction (`with conn:`) speeds up writes by reducing disk fsyncs.
- **Requests**: Network I/O bound. Using `Session` speeds up HTTPS requests by reusing the TLS handshake and TCP connection.
- **Hashing**: `scrypt` is intentionally slow to defeat brute-forcing. Offload password hashing to a background task or thread in high-throughput APIs so it doesn't block the async event loop.


#### Quick Recap
- `sqlite3` provides a built-in relational database. Use `?` parameters to prevent SQL injection.
- `requests.get/post` must ALWAYS include a `timeout` argument.
- Call `raise_for_status()` on requests to ensure you aren't silently ignoring 404s or 500s.
- The `secrets` module generates cryptographically secure random numbers for tokens.
- Passwords must be hashed using a slow algorithm like `scrypt` with a random salt.
- Use `hmac.compare_digest` to compare hashes securely without timing vulnerabilities.


#### Interview Questions
- **Junior**: Why should you never use `random.choice` to generate a security token?  
  **Answer**: The `random` module is a Mersenne Twister pseudo-random number generator. It is predictable if an attacker observes enough outputs. The `secrets` module uses the operating system's cryptographically secure random source (e.g., `/dev/urandom`), which is unpredictable.
- **Mid**: What is SQL injection and how does Python's `sqlite3` module help prevent it?  
  **Answer**: SQL injection is when an attacker inputs malicious SQL syntax that gets executed by the database because it was unsafely concatenated into the query string. `sqlite3` prevents this by using parameterized queries (the `?` placeholder). The database driver safely escapes the inputs, treating them strictly as literal values, not executable code.
- **Senior**: Why is `sha256` considered insecure for password hashing, and what makes `scrypt` better?  
  **Answer**: `sha256` is a fast cryptographic hash designed for verifying file integrity quickly. Modern GPUs can calculate billions of `sha256` hashes per second, making brute-force and dictionary attacks trivial. `scrypt` is a key derivation function that is both CPU-intensive and memory-hard. It forces the attacker to use significant RAM to compute each hash, defeating parallelization on GPUs and ASICs.
