## Pandas Pro-Tips

A few habits that separate someone who knows Pandas syntax from someone who can actually clean data fast:

- **Look before you touch.** Before writing a single line of transformation code, run `df.head()` to see the actual shape of the data, `df.info()` to check dtypes and null counts, and `df.describe()` to sanity-check the numeric ranges. Every real data-cleaning task starts here, not with code.
- **`Shift + Tab`** inside any method's parentheses pops up its docstring -- use it constantly instead of guessing parameter names.
- **Avoid `.iterrows()`.** It is one of the slowest ways to touch a DataFrame. If you find yourself reaching for a loop, ask first: is there a vectorized method, `.apply()`, `np.where()`, or `np.select()` that does this without iterating row by row in Python?
- **Chain deliberately.** Method chaining (`df.dropna().reset_index()...`) is powerful, but each link should be a single clear step -- if a chain gets hard to read, break it up or add a comment.
