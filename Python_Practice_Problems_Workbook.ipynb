### How This Workbook Works

105 problems across 7 topics. Each topic runs Basic -> Intermediate -> Advanced, 5 problems per level, 15 problems per topic.

No solutions are included in this workbook. That's intentional. Read the problem, write the code in the cell directly below it, run it, break it, fix it. If a problem doesn't click in under 10 minutes, that's information, not failure — note where you got stuck and move on. Come back to it after finishing the topic.

**One rule:** don't look up the answer before you've tried and failed at least twice.

## Topic 1: Python Basics & Control Flow

### Basic

**Problem 1:** Write a script that takes a temperature in Celsius from the user and converts it to Fahrenheit, printing the result rounded to 2 decimal places.

**Problem 2:** Write a program that checks whether a given year is a leap year, accounting for the century exception rule (divisible by 100 but not 400).

**Problem 3:** Given a string, write code that counts and prints the number of vowels, consonants, digits, and spaces in it separately.

**Problem 4:** Write a program that swaps the values of two variables without using a third temporary variable.

**Problem 5:** Write a script that takes three numbers as input and prints them in ascending order without using the built-in `sorted()` function.

### Intermediate

**Problem 6:** Write a program that prints the multiplication table of a number in a formatted grid (not a plain list) for numbers 1 through 10.

**Problem 7:** Write a script that validates whether a given password meets these rules: at least 8 characters, one uppercase letter, one digit, and one special character. Print exactly which rule failed if it's invalid.

**Problem 8:** Given a sentence, write a program that reverses the order of words but keeps each word itself unreversed (e.g., 'Code is life' becomes 'life is Code').

**Problem 9:** Write a program that simulates a basic ATM: it should loop asking for an action (deposit, withdraw, check balance, exit) and correctly handle insufficient funds and invalid input without crashing.

**Problem 10:** Write a script that determines whether a given number is an Armstrong number (a number equal to the sum of its own digits each raised to the power of the number of digits).

### Advanced

**Problem 11:** Write a program that prints a diamond pattern of asterisks whose size is determined by user input, without hardcoding logic for a single size.

**Problem 12:** Write a script that parses a raw date string in multiple possible formats (e.g., '2024-01-05', '05/01/2024', 'Jan 5, 2024') and normalizes them all to 'YYYY-MM-DD' using only conditional logic and string methods (no external libraries).

**Problem 13:** Write a program that calculates someone's exact age in years, months, and days given their birth date and today's date, correctly handling month/day borrowing.

**Problem 14:** Write a script that implements a basic calculator supporting +, -, *, /, and % that reads a full expression like '12 + 4 * 2' from a string and evaluates it respecting operator precedence, without using `eval()`.

**Problem 15:** Write a program that finds all 'self dividing' numbers in a given range (numbers where every digit divides the number evenly, and no digit is zero), and prints them along with the total count.

### Advanced

**Problem 29:** Write a program that deep-copies a nested dictionary structure of arbitrary depth without using the `copy` module.

### Basic

**Problem 97:** Write a script that safely parses a CSV-like text file (comma-separated, no external library) where some rows may be malformed, skipping bad rows and logging which line numbers failed and why.

**Problem 98:** Write a custom context manager (using a class with `__enter__` and `__exit__`) that times how long the code inside the `with` block takes to run.

**Problem 104:** Write a script that safely handles reading and writing to the same file for an update operation (read, transform, write back) in a way that prevents data loss if the program crashes mid-write (hint: write to a temp file then replace).

**Problem 105:** Write a context manager that acquires and releases a simple in-memory 'lock' (a shared flag/counter) to simulate protecting a critical section, and correctly handles the case where an exception occurs while the lock is held.
